# PHASE 01 — Setup & Dataset Loading

```python
# ============================================================
# PHASE 01 — SETUP & DATASET LOADING
# CELL 1: Import Libraries and Project Configuration
# ============================================================

import os
import sys
import json
import random
import warnings

import numpy as np
import pandas as pd

warnings.filterwarnings("ignore")


# ------------------------------------------------------------
# Reproducibility
# ------------------------------------------------------------
SEED = 42

random.seed(SEED)
np.random.seed(SEED)


# ------------------------------------------------------------
# Dataset Path
# ------------------------------------------------------------
DATASET_DIR = "/kaggle/input/datasets/blastchar/telco-customer-churn"

DATA_PATH = os.path.join(
    DATASET_DIR,
    "WA_Fn-UseC_-Telco-Customer-Churn.csv"
)


# ------------------------------------------------------------
# Main Working Directory
# ------------------------------------------------------------
WORK_DIR = "/kaggle/working/telco_churn_research"

os.makedirs(WORK_DIR, exist_ok=True)


# ------------------------------------------------------------
# Phase-wise Output Directories
# ------------------------------------------------------------
OUTPUT_DIRS = {
    "01_data_audit": os.path.join(WORK_DIR, "01_data_audit"),
    "02_eda": os.path.join(WORK_DIR, "02_eda"),
    "03_split": os.path.join(WORK_DIR, "03_split"),
    "04_rq1_imbalance": os.path.join(WORK_DIR, "04_rq1_imbalance"),
    "05_tuning": os.path.join(WORK_DIR, "05_tuning"),
    "06_oof": os.path.join(WORK_DIR, "06_oof"),
    "07_threshold": os.path.join(WORK_DIR, "07_threshold"),
    "08_utility": os.path.join(WORK_DIR, "08_utility"),
    "09_final_model": os.path.join(WORK_DIR, "09_final_model"),
    "10_shap": os.path.join(WORK_DIR, "10_shap"),
    "11_paper_outputs": os.path.join(WORK_DIR, "11_paper_outputs"),
}

for folder in OUTPUT_DIRS.values():
    os.makedirs(folder, exist_ok=True)


print("✅ Project setup completed.")
print(f"Random Seed : {SEED}")
print(f"Working Dir : {WORK_DIR}")
```

# PHASE 02 — Data Quality Audit 

```python
# ============================================================
# PHASE 02 — DATA QUALITY AUDIT
# CELL 1: Basic Dataset Audit
# ============================================================

print("=" * 75)
print("PHASE 02 — BASIC DATASET AUDIT")
print("=" * 75)

print(f"Rows                : {df.shape[0]:,}")
print(f"Columns             : {df.shape[1]}")
print(f"Exact duplicate rows: {df.duplicated().sum():,}")

print("\nColumn type counts:")
print(df.dtypes.astype(str).value_counts())

print("\n✅ Basic audit completed.")
```

```python
# ============================================================
# CELL 2: Missing Values and Hidden Blank Strings
# ============================================================

audit_rows = []

for col in df.columns:
    
    # Standard missing values
    missing_count = int(df[col].isna().sum())

    # Blank / whitespace-only values for object columns
    if df[col].dtype == "object":
        blank_count = int(
            df[col]
            .astype(str)
            .str.strip()
            .eq("")
            .sum()
        )
    else:
        blank_count = 0

    total_problem_count = missing_count + blank_count

    audit_rows.append({
        "Column": col,
        "Data_Type": str(df[col].dtype),
        "Missing_NaN": missing_count,
        "Blank_or_Whitespace": blank_count,
        "Total_Missing_or_Blank": total_problem_count,
        "Problem_Percentage": round(
            (total_problem_count / len(df)) * 100, 4
        )
    })


missing_blank_audit = pd.DataFrame(audit_rows)

display(
    missing_blank_audit[
        missing_blank_audit["Total_Missing_or_Blank"] > 0
    ]
)

print(
    "\nTotal standard NaN values:",
    int(df.isna().sum().sum())
)

print(
    "Total hidden blank/whitespace values:",
    int(missing_blank_audit["Blank_or_Whitespace"].sum())
)
```

```python
# ============================================================
# CELL 3: TotalCharges Detailed Audit
# ============================================================

TOTAL_CHARGES_COL = "TotalCharges"

print("=" * 75)
print("TOTALCHARGES AUDIT")
print("=" * 75)

print("Raw dtype:")
print(df[TOTAL_CHARGES_COL].dtype)


# ------------------------------------------------------------
# Detect blank / whitespace-only values
# ------------------------------------------------------------
totalcharges_blank_mask = (
    df[TOTAL_CHARGES_COL]
    .astype(str)
    .str.strip()
    .eq("")
)

totalcharges_blank_count = int(totalcharges_blank_mask.sum())

print(
    "\nBlank / whitespace-only TotalCharges values:",
    totalcharges_blank_count
)


# ------------------------------------------------------------
# Temporary numeric conversion FOR AUDIT ONLY
# df is NOT modified
# ------------------------------------------------------------
totalcharges_numeric_audit = pd.to_numeric(
    df[TOTAL_CHARGES_COL],
    errors="coerce"
)

numeric_conversion_nan = int(
    totalcharges_numeric_audit.isna().sum()
)

print(
    "Values becoming NaN after numeric conversion:",
    numeric_conversion_nan
)


# ------------------------------------------------------------
# Display suspicious rows
# ------------------------------------------------------------
suspicious_totalcharges_rows = df.loc[
    totalcharges_numeric_audit.isna(),
    [
        "customerID",
        "tenure",
        "MonthlyCharges",
        "TotalCharges",
        "Churn"
    ]
]

print("\nSuspicious TotalCharges rows:")
display(suspicious_totalcharges_rows)


print("\nIMPORTANT:")
print("The original df has NOT been modified.")
```

```python
# ============================================================
# CELL 4: Duplicate Rows and Customer ID Integrity
# ============================================================

duplicate_rows = int(df.duplicated().sum())

customer_id_missing = int(df["customerID"].isna().sum())

customer_id_blank = int(
    df["customerID"]
    .astype(str)
    .str.strip()
    .eq("")
    .sum()
)

customer_id_duplicate = int(
    df["customerID"].duplicated().sum()
)

customer_id_unique = int(
    df["customerID"].nunique(dropna=True)
)


print("=" * 75)
print("DUPLICATE & CUSTOMER ID AUDIT")
print("=" * 75)

print(f"Exact duplicate rows       : {duplicate_rows:,}")
print(f"customerID missing         : {customer_id_missing:,}")
print(f"customerID blank           : {customer_id_blank:,}")
print(f"customerID duplicates      : {customer_id_duplicate:,}")
print(f"Unique customerID values   : {customer_id_unique:,}")
print(f"Total rows                 : {len(df):,}")


if (
    customer_id_missing == 0
    and customer_id_blank == 0
    and customer_id_duplicate == 0
    and customer_id_unique == len(df)
):
    print("\n✅ customerID integrity looks valid.")
else:
    print("\n⚠️ customerID integrity requires review.")
```

```python
# ============================================================
# CELL 5: Unique Values and Constant Column Audit
# ============================================================

unique_summary = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str).values,
    "Unique_Count": [
        df[col].nunique(dropna=False)
        for col in df.columns
    ]
})

unique_summary["Unique_Percentage"] = (
    unique_summary["Unique_Count"] / len(df) * 100
).round(2)


print("Unique value summary:")
display(unique_summary)


constant_columns = unique_summary.loc[
    unique_summary["Unique_Count"] <= 1,
    "Column"
].tolist()


print("\nConstant columns:")
print(
    constant_columns
    if constant_columns
    else "None"
)
```

```python
# ============================================================
# CELL 6: Numerical Range Audit
# ============================================================

numeric_audit_df = pd.DataFrame({
    "SeniorCitizen": pd.to_numeric(
        df["SeniorCitizen"],
        errors="coerce"
    ),
    "tenure": pd.to_numeric(
        df["tenure"],
        errors="coerce"
    ),
    "MonthlyCharges": pd.to_numeric(
        df["MonthlyCharges"],
        errors="coerce"
    ),
    "TotalCharges": totalcharges_numeric_audit
})


numeric_range_summary = pd.DataFrame({
    "Column": numeric_audit_df.columns,
    "Minimum": [
        numeric_audit_df[col].min()
        for col in numeric_audit_df.columns
    ],
    "Maximum": [
        numeric_audit_df[col].max()
        for col in numeric_audit_df.columns
    ],
    "Mean": [
        numeric_audit_df[col].mean()
        for col in numeric_audit_df.columns
    ],
    "Median": [
        numeric_audit_df[col].median()
        for col in numeric_audit_df.columns
    ],
    "Missing_after_numeric_conversion": [
        numeric_audit_df[col].isna().sum()
        for col in numeric_audit_df.columns
    ]
})


display(numeric_range_summary)


print("\nBasic suspicious-value checks:")

print(
    "Negative tenure:",
    int((numeric_audit_df["tenure"] < 0).sum())
)

print(
    "Negative MonthlyCharges:",
    int((numeric_audit_df["MonthlyCharges"] < 0).sum())
)

print(
    "Negative TotalCharges:",
    int((numeric_audit_df["TotalCharges"] < 0).sum())
)

print(
    "Invalid SeniorCitizen values:",
    int(
        (~numeric_audit_df["SeniorCitizen"].isin([0, 1]))
        .sum()
    )
)
```

```python
# ============================================================
# CELL 7: Categorical Value Audit
# ============================================================

categorical_columns = df.select_dtypes(
    include="object"
).columns.tolist()

print("=" * 75)
print("CATEGORICAL VALUE AUDIT")
print("=" * 75)


for col in categorical_columns:
    
    print(f"\n--- {col} ---")
    
    # customerID has thousands of unique values,
    # so do not print every ID
    if col == "customerID":
        print(
            f"Unique values: {df[col].nunique(dropna=False):,}"
        )
        continue

    value_counts = df[col].value_counts(
        dropna=False
    )

    print(value_counts)
```

```python
# ============================================================
# CELL 8: Target Distribution Audit
# ============================================================

target_distribution = (
    df[TARGET]
    .value_counts(dropna=False)
    .rename_axis("Class")
    .reset_index(name="Count")
)

target_distribution["Percentage"] = (
    target_distribution["Count"]
    / len(df)
    * 100
).round(2)


display(target_distribution)


if (
    "Yes" in df[TARGET].values
    and "No" in df[TARGET].values
):

    churn_count = int(
        (df[TARGET] == "Yes").sum()
    )

    non_churn_count = int(
        (df[TARGET] == "No").sum()
    )

    churn_rate = churn_count / len(df) * 100
    non_churn_rate = non_churn_count / len(df) * 100

    imbalance_ratio = (
        non_churn_count / churn_count
    )

    print(f"Churn rate       : {churn_rate:.2f}%")
    print(f"Non-churn rate   : {non_churn_rate:.2f}%")
    print(
        f"Non-Churn : Churn ratio = "
        f"{imbalance_ratio:.2f} : 1"
    )
```

```python
# ============================================================
# CELL 9: Save Phase 02 Audit Outputs
# ============================================================

AUDIT_DIR = OUTPUT_DIRS["01_data_audit"]


missing_blank_path = os.path.join(
    AUDIT_DIR,
    "missing_blank_audit.csv"
)

unique_summary_path = os.path.join(
    AUDIT_DIR,
    "unique_value_summary.csv"
)

numeric_range_path = os.path.join(
    AUDIT_DIR,
    "numeric_range_summary.csv"
)

target_distribution_path = os.path.join(
    AUDIT_DIR,
    "target_distribution.csv"
)

suspicious_totalcharges_path = os.path.join(
    AUDIT_DIR,
    "suspicious_totalcharges_rows.csv"
)


missing_blank_audit.to_csv(
    missing_blank_path,
    index=False
)

unique_summary.to_csv(
    unique_summary_path,
    index=False
)

numeric_range_summary.to_csv(
    numeric_range_path,
    index=False
)

target_distribution.to_csv(
    target_distribution_path,
    index=False
)

suspicious_totalcharges_rows.to_csv(
    suspicious_totalcharges_path,
    index=False
)


print("✅ Phase 02 audit outputs saved.")

print("\nSaved files:")
print(missing_blank_path)
print(unique_summary_path)
print(numeric_range_path)
print(target_distribution_path)
print(suspicious_totalcharges_path)
```

```python
# ============================================================
# PHASE 02 — FINAL SUMMARY
# Run this cell LAST and send me the complete output
# ============================================================

print("=" * 78)
print("PHASE 02 — DATA QUALITY AUDIT SUMMARY")
print("=" * 78)


# ------------------------------------------------------------
# 1. Dataset Integrity
# ------------------------------------------------------------
print("\n[1] DATASET INTEGRITY")

print(f"Rows                  : {len(df):,}")
print(f"Columns               : {df.shape[1]}")
print(f"Exact duplicate rows  : {duplicate_rows:,}")


# ------------------------------------------------------------
# 2. Missing / Blank Values
# ------------------------------------------------------------
print("\n[2] MISSING / BLANK VALUES")

problem_columns = missing_blank_audit[
    missing_blank_audit["Total_Missing_or_Blank"] > 0
]

if len(problem_columns) == 0:
    print("No missing or blank values detected.")

else:
    for _, row in problem_columns.iterrows():
        print(
            f"{row['Column']:<20} "
            f"NaN={int(row['Missing_NaN'])}, "
            f"Blank={int(row['Blank_or_Whitespace'])}, "
            f"Total={int(row['Total_Missing_or_Blank'])}"
        )


# ------------------------------------------------------------
# 3. TotalCharges
# ------------------------------------------------------------
print("\n[3] TOTALCHARGES AUDIT")

print(f"Raw dtype                         : {df['TotalCharges'].dtype}")
print(f"Blank / whitespace values         : {totalcharges_blank_count}")
print(
    f"NaN after temporary numeric cast  : "
    f"{numeric_conversion_nan}"
)


# ------------------------------------------------------------
# 4. Customer ID Integrity
# ------------------------------------------------------------
print("\n[4] CUSTOMER ID INTEGRITY")

print(f"Missing customerID       : {customer_id_missing}")
print(f"Blank customerID         : {customer_id_blank}")
print(f"Duplicate customerID     : {customer_id_duplicate}")
print(f"Unique customerID        : {customer_id_unique:,}")


# ------------------------------------------------------------
# 5. Constant Columns
# ------------------------------------------------------------
print("\n[5] CONSTANT COLUMNS")

if constant_columns:
    print(constant_columns)
else:
    print("None")


# ------------------------------------------------------------
# 6. Basic Numeric Problems
# ------------------------------------------------------------
negative_tenure = int(
    (numeric_audit_df["tenure"] < 0).sum()
)

negative_monthly = int(
    (numeric_audit_df["MonthlyCharges"] < 0).sum()
)

negative_total = int(
    (numeric_audit_df["TotalCharges"] < 0).sum()
)

invalid_senior = int(
    (~numeric_audit_df["SeniorCitizen"].isin([0, 1]))
    .sum()
)


print("\n[6] BASIC NUMERIC VALIDITY")

print(f"Negative tenure          : {negative_tenure}")
print(f"Negative MonthlyCharges  : {negative_monthly}")
print(f"Negative TotalCharges    : {negative_total}")
print(f"Invalid SeniorCitizen    : {invalid_senior}")


# ------------------------------------------------------------
# 7. Target
# ------------------------------------------------------------
print("\n[7] TARGET DISTRIBUTION")

print(f"Non-Churn (No) : {non_churn_count:,} ({non_churn_rate:.2f}%)")
print(f"Churn (Yes)    : {churn_count:,} ({churn_rate:.2f}%)")
print(
    f"Class ratio    : "
    f"{imbalance_ratio:.2f} Non-Churn per Churn"
)


# ------------------------------------------------------------
# 8. Important Findings
# ------------------------------------------------------------
print("\n[8] IMPORTANT AUDIT FLAGS")

flags = []

if duplicate_rows > 0:
    flags.append(
        f"{duplicate_rows} exact duplicate row(s)"
    )

if totalcharges_blank_count > 0:
    flags.append(
        f"{totalcharges_blank_count} blank TotalCharges value(s)"
    )

if customer_id_duplicate > 0:
    flags.append(
        f"{customer_id_duplicate} duplicated customerID(s)"
    )

if negative_tenure > 0:
    flags.append("Negative tenure detected")

if negative_monthly > 0:
    flags.append("Negative MonthlyCharges detected")

if negative_total > 0:
    flags.append("Negative TotalCharges detected")

if invalid_senior > 0:
    flags.append("Invalid SeniorCitizen value detected")


if flags:
    for flag in flags:
        print("⚠️", flag)
else:
    print("✅ No major raw-data integrity problems detected.")


# ------------------------------------------------------------
# 9. Phase Status
# ------------------------------------------------------------
print("\n" + "=" * 78)

critical_issue = any([
    customer_id_duplicate > 0,
    customer_id_missing > 0,
    negative_tenure > 0,
    negative_monthly > 0,
    negative_total > 0,
    invalid_senior > 0,
])

if critical_issue:
    print("⚠️ PHASE 02 STATUS: REVIEW REQUIRED")
else:
    print("✅ PHASE 02 STATUS: AUDIT COMPLETED")

print("=" * 78)

print(
    "\nNOTE: No cleaning, imputation, encoding, "
    "scaling, or resampling has been performed."
)
```

#  PHASE 03 — Cleaning & Target Setup 

```python
# ============================================================
# PHASE 03 — CLEANING & TARGET SETUP
# CELL 1: Preserve Original Raw Dataset
# ============================================================

df_raw = df.copy(deep=True)

df_clean = df.copy(deep=True)

print("✅ Raw dataset preserved.")
print("Original shape:", df_raw.shape)
print("Working copy shape:", df_clean.shape)
```

```python
# ============================================================
# CELL 2: Remove Identifier Column
# ============================================================

ID_COLUMN = "customerID"

if ID_COLUMN not in df_clean.columns:
    raise ValueError(
        f"Expected identifier column '{ID_COLUMN}' was not found."
    )


print("Before removing customerID:")
print("Shape:", df_clean.shape)


df_clean = df_clean.drop(
    columns=[ID_COLUMN]
)


print("\nAfter removing customerID:")
print("Shape:", df_clean.shape)

print(
    "\ncustomerID present:",
    ID_COLUMN in df_clean.columns
)
```

```python
# ============================================================
# CELL 3: Convert TotalCharges to Numeric
# ============================================================

TOTAL_CHARGES_COL = "TotalCharges"

print("Before conversion:")
print("dtype:", df_clean[TOTAL_CHARGES_COL].dtype)


df_clean[TOTAL_CHARGES_COL] = pd.to_numeric(
    df_clean[TOTAL_CHARGES_COL],
    errors="coerce"
)


print("\nAfter conversion:")
print("dtype:", df_clean[TOTAL_CHARGES_COL].dtype)

print(
    "Missing values after conversion:",
    df_clean[TOTAL_CHARGES_COL].isna().sum()
)
```

```python
# ============================================================
# CELL 4: Verify Missing TotalCharges Rows
# ============================================================

missing_totalcharges_rows = df_clean[
    df_clean[TOTAL_CHARGES_COL].isna()
].copy()


print(
    "Rows with missing TotalCharges:",
    len(missing_totalcharges_rows)
)

display(
    missing_totalcharges_rows[
        [
            "tenure",
            "MonthlyCharges",
            "TotalCharges",
            "Churn"
        ]
    ]
)
```

```python
# ============================================================
# CELL 5: Encode Target Variable
# ============================================================

TARGET = "Churn"

print("Target values before encoding:")
print(df_clean[TARGET].value_counts(dropna=False))


target_mapping = {
    "No": 0,
    "Yes": 1
}


unexpected_target_values = set(
    df_clean[TARGET].dropna().unique()
) - set(target_mapping.keys())


if unexpected_target_values:
    raise ValueError(
        f"Unexpected target values found: "
        f"{unexpected_target_values}"
    )


df_clean[TARGET] = df_clean[TARGET].map(
    target_mapping
).astype("int8")


print("\nTarget values after encoding:")
print(df_clean[TARGET].value_counts(dropna=False))

print("\nTarget dtype:")
print(df_clean[TARGET].dtype)
```

```python
# ============================================================
# CELL 6: Cleaning Integrity Check
# ============================================================

print("=" * 75)
print("CLEANING INTEGRITY CHECK")
print("=" * 75)


print(
    "Original row count:",
    len(df_raw)
)

print(
    "Cleaned row count :",
    len(df_clean)
)


print(
    "\nRows removed:",
    len(df_raw) - len(df_clean)
)


print(
    "\nCurrent columns:",
    df_clean.shape[1]
)


print(
    "\nTotalCharges dtype:",
    df_clean["TotalCharges"].dtype
)


print(
    "TotalCharges missing:",
    df_clean["TotalCharges"].isna().sum()
)


print(
    "\nTarget unique values:",
    sorted(df_clean[TARGET].unique().tolist())
)
```

```python
# ============================================================
# CELL 7: Remaining Missing Values After Cleaning
# ============================================================

remaining_missing = (
    df_clean
    .isna()
    .sum()
    .sort_values(ascending=False)
)

remaining_missing = remaining_missing[
    remaining_missing > 0
]


print("Remaining missing values:")

if remaining_missing.empty:
    print("None")
else:
    print(remaining_missing)


print(
    "\nTotal remaining missing values:",
    int(df_clean.isna().sum().sum())
)
```

```python
# ============================================================
# CELL 8: Feature Type Summary
# ============================================================

feature_columns = [
    col for col in df_clean.columns
    if col != TARGET
]


numerical_columns_current = (
    df_clean[feature_columns]
    .select_dtypes(include=["number"])
    .columns
    .tolist()
)


categorical_columns_current = (
    df_clean[feature_columns]
    .select_dtypes(include=["object", "category"])
    .columns
    .tolist()
)


print("=" * 75)
print("FEATURE TYPE SUMMARY")
print("=" * 75)


print(
    f"\nPredictor features: "
    f"{len(feature_columns)}"
)


print(
    f"\nNumerical columns "
    f"({len(numerical_columns_current)}):"
)

for col in numerical_columns_current:
    print(" -", col)


print(
    f"\nCategorical columns "
    f"({len(categorical_columns_current)}):"
)

for col in categorical_columns_current:
    print(" -", col)
```

```python
# ============================================================
# PHASE 03 — FINAL SUMMARY
# Run this cell LAST and send me the complete output
# ============================================================

print("=" * 78)
print("PHASE 03 — CLEANING & TARGET SETUP SUMMARY")
print("=" * 78)


# ------------------------------------------------------------
# 1. Dataset Shape
# ------------------------------------------------------------
print("\n[1] DATASET SHAPE")

print(
    f"Original : "
    f"{df_raw.shape[0]:,} rows × "
    f"{df_raw.shape[1]} columns"
)

print(
    f"Cleaned  : "
    f"{df_clean.shape[0]:,} rows × "
    f"{df_clean.shape[1]} columns"
)

print(
    f"Rows removed : "
    f"{len(df_raw) - len(df_clean)}"
)


# ------------------------------------------------------------
# 2. customerID
# ------------------------------------------------------------
print("\n[2] IDENTIFIER HANDLING")

print(
    "customerID removed:",
    "customerID" not in df_clean.columns
)


# ------------------------------------------------------------
# 3. TotalCharges
# ------------------------------------------------------------
print("\n[3] TOTALCHARGES")

print(
    "Original dtype :",
    df_raw["TotalCharges"].dtype
)

print(
    "Cleaned dtype  :",
    df_clean["TotalCharges"].dtype
)

print(
    "Remaining NaN  :",
    int(
        df_clean["TotalCharges"]
        .isna()
        .sum()
    )
)


# ------------------------------------------------------------
# 4. Target
# ------------------------------------------------------------
print("\n[4] TARGET ENCODING")

print(
    "Target dtype:",
    df_clean[TARGET].dtype
)

print(
    "Target values:",
    sorted(
        df_clean[TARGET]
        .unique()
        .tolist()
    )
)

print("\nTarget counts:")

target_clean_counts = (
    df_clean[TARGET]
    .value_counts()
    .sort_index()
)

for label, count in target_clean_counts.items():

    class_name = (
        "Non-Churn"
        if label == 0
        else "Churn"
    )

    print(
        f"    {label} "
        f"({class_name}) : "
        f"{count:,}"
    )


# ------------------------------------------------------------
# 5. Missing Values
# ------------------------------------------------------------
print("\n[5] REMAINING MISSING VALUES")

remaining_missing_summary = (
    df_clean
    .isna()
    .sum()
)

remaining_missing_summary = (
    remaining_missing_summary[
        remaining_missing_summary > 0
    ]
)

if remaining_missing_summary.empty:
    print("None")

else:
    for col, count in (
        remaining_missing_summary.items()
    ):
        print(
            f"    {col}: {count}"
        )


# ------------------------------------------------------------
# 6. Feature Types
# ------------------------------------------------------------
print("\n[6] FEATURE STRUCTURE")

print(
    "Predictor features :",
    len(feature_columns)
)

print(
    "Numeric dtype predictors :",
    len(numerical_columns_current)
)

print(
    "Categorical predictors   :",
    len(categorical_columns_current)
)


# ------------------------------------------------------------
# 7. Leakage Protection
# ------------------------------------------------------------
print("\n[7] LEAKAGE-SAFETY STATUS")

print(
    "Missing-value imputation performed : False"
)

print(
    "Feature scaling performed           : False"
)

print(
    "Categorical encoding performed      : False"
)

print(
    "Resampling performed                : False"
)

print(
    "Train/test split performed          : False"
)


# ------------------------------------------------------------
# 8. Saved Files
# ------------------------------------------------------------
print("\n[8] SAVED OUTPUTS")

print(
    "Cleaned dataset:",
    os.path.isfile(
        cleaned_dataset_path
    )
)

print(
    cleaned_dataset_path
)

print(
    "\nCleaning log:",
    os.path.isfile(
        cleaning_log_path
    )
)

print(
    cleaning_log_path
)


# ------------------------------------------------------------
# Integrity Checks
# ------------------------------------------------------------
phase03_checks = {
    "No rows accidentally removed":
        len(df_clean) == len(df_raw),

    "customerID removed":
        "customerID" not in df_clean.columns,

    "TotalCharges numeric":
        pd.api.types.is_numeric_dtype(
            df_clean["TotalCharges"]
        ),

    "Exactly 11 TotalCharges NaN":
        int(
            df_clean["TotalCharges"]
            .isna()
            .sum()
        ) == 11,

    "Target binary encoded":
        set(
            df_clean[TARGET]
            .unique()
        ) == {0, 1},

    "Target counts preserved":
        (
            int(
                (df_clean[TARGET] == 0)
                .sum()
            ) == 5174
            and
            int(
                (df_clean[TARGET] == 1)
                .sum()
            ) == 1869
        )
}


print("\n" + "-" * 78)
print("PHASE 03 INTEGRITY CHECKS")
print("-" * 78)


for name, status in phase03_checks.items():

    symbol = (
        "✅"
        if status
        else "❌"
    )

    print(
        f"{symbol} {name}"
    )


all_phase03_passed = all(
    phase03_checks.values()
)


print("\n" + "=" * 78)

if all_phase03_passed:

    print(
        "✅ PHASE 03 STATUS: PASSED"
    )

else:

    print(
        "❌ PHASE 03 STATUS: REVIEW REQUIRED"
    )

print("=" * 78)


print(
    "\nNOTE: The 11 missing TotalCharges values "
    "have intentionally NOT been imputed yet."
)

print(
    "Imputation will be learned from training data "
    "inside the preprocessing pipeline after splitting."
)
```

```python
# ============================================================
# PHASE 04 — RESEARCH-FOCUSED EDA
# CELL 1: EDA Setup
# ============================================================

import matplotlib.pyplot as plt

EDA_DIR = OUTPUT_DIRS["02_eda"]

os.makedirs(EDA_DIR, exist_ok=True)

# Preserve the cleaned dataset
df_eda = df_clean.copy(deep=True)

TARGET = "Churn"

CHURN_LABELS = {
    0: "Non-Churn",
    1: "Churn"
}

print("=" * 75)
print("PHASE 04 — RESEARCH-FOCUSED EDA")
print("=" * 75)

print(f"EDA rows    : {df_eda.shape[0]:,}")
print(f"EDA columns : {df_eda.shape[1]}")
print(f"Target      : {TARGET}")

print("\n✅ EDA working copy created.")
```

```python
# ============================================================
# CELL 2: Target Distribution
# ============================================================

target_eda = (
    df_eda[TARGET]
    .value_counts()
    .sort_index()
    .rename_axis("Churn")
    .reset_index(name="Count")
)

target_eda["Class"] = target_eda["Churn"].map(
    CHURN_LABELS
)

target_eda["Percentage"] = (
    target_eda["Count"] / len(df_eda) * 100
).round(2)

target_eda = target_eda[
    ["Churn", "Class", "Count", "Percentage"]
]

display(target_eda)


# ------------------------------------------------------------
# Save table
# ------------------------------------------------------------

target_eda_path = os.path.join(
    EDA_DIR,
    "target_distribution.csv"
)

target_eda.to_csv(
    target_eda_path,
    index=False
)

print("\n✅ Target distribution table saved.")
```

```python
# ============================================================
# CELL 3: Target Distribution Figure
# ============================================================

fig, ax = plt.subplots(figsize=(7, 5))

bars = ax.bar(
    target_eda["Class"],
    target_eda["Count"]
)

ax.set_title(
    "Customer Churn Class Distribution",
    fontsize=13
)

ax.set_xlabel("Customer Status")
ax.set_ylabel("Number of Customers")

ax.grid(
    axis="y",
    alpha=0.25
)


# Add count and percentage above bars
for bar, count, pct in zip(
    bars,
    target_eda["Count"],
    target_eda["Percentage"]
):
    ax.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height(),
        f"{count:,}\n({pct:.2f}%)",
        ha="center",
        va="bottom"
    )


plt.tight_layout()

target_plot_path = os.path.join(
    EDA_DIR,
    "target_distribution.png"
)

plt.savefig(
    target_plot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()

print("✅ Target distribution figure saved.")
print(target_plot_path)
```

```python
# ============================================================
# CELL 4: Numerical Features vs Churn
# ============================================================

continuous_features = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]


numeric_summary_rows = []

for feature in continuous_features:

    for churn_class in [0, 1]:

        values = df_eda.loc[
            df_eda[TARGET] == churn_class,
            feature
        ]

        numeric_summary_rows.append({
            "Feature": feature,
            "Churn_Class": churn_class,
            "Class_Name": CHURN_LABELS[churn_class],
            "Count": int(values.notna().sum()),
            "Missing": int(values.isna().sum()),
            "Mean": round(values.mean(), 4),
            "Median": round(values.median(), 4),
            "Std": round(values.std(), 4),
            "Minimum": round(values.min(), 4),
            "Maximum": round(values.max(), 4)
        })


numeric_churn_summary = pd.DataFrame(
    numeric_summary_rows
)

display(numeric_churn_summary)


numeric_summary_path = os.path.join(
    EDA_DIR,
    "numeric_features_by_churn.csv"
)

numeric_churn_summary.to_csv(
    numeric_summary_path,
    index=False
)

print("\n✅ Numerical churn summary saved.")
```

```python
# ============================================================
# CELL 5: Numerical Feature Boxplots
# ============================================================

fig, axes = plt.subplots(
    1,
    3,
    figsize=(16, 5)
)


for ax, feature in zip(
    axes,
    continuous_features
):

    non_churn_values = (
        df_eda.loc[
            df_eda[TARGET] == 0,
            feature
        ]
        .dropna()
    )

    churn_values = (
        df_eda.loc[
            df_eda[TARGET] == 1,
            feature
        ]
        .dropna()
    )

    ax.boxplot(
        [
            non_churn_values,
            churn_values
        ],
        labels=[
            "Non-Churn",
            "Churn"
        ],
        showfliers=False
    )

    ax.set_title(
        f"{feature} by Churn Status"
    )

    ax.set_xlabel(
        "Customer Status"
    )

    ax.set_ylabel(
        feature
    )

    ax.grid(
        axis="y",
        alpha=0.20
    )


plt.suptitle(
    "Numerical Feature Distributions by Churn Status",
    fontsize=14
)

plt.tight_layout()


numeric_boxplot_path = os.path.join(
    EDA_DIR,
    "numeric_features_vs_churn.png"
)

plt.savefig(
    numeric_boxplot_path,
    dpi=300,
    bbox_inches="tight"
)

plt.show()
plt.close()


print("✅ Numerical feature figure saved.")
print(numeric_boxplot_path)
```

```python
# ============================================================
# CELL 6: Numerical Median Comparison
# ============================================================

numeric_median_comparison = (
    df_eda
    .groupby(TARGET)[continuous_features]
    .median()
    .T
    .reset_index()
)

numeric_median_comparison.columns = [
    "Feature",
    "Non_Churn_Median",
    "Churn_Median"
]

numeric_median_comparison[
    "Absolute_Difference"
] = (
    numeric_median_comparison[
        "Churn_Median"
    ]
    -
    numeric_median_comparison[
        "Non_Churn_Median"
    ]
).round(4)


display(
    numeric_median_comparison
)


numeric_median_path = os.path.join(
    EDA_DIR,
    "numeric_median_comparison.csv"
)

numeric_median_comparison.to_csv(
    numeric_median_path,
    index=False
)

print(
    "✅ Numerical median comparison saved."
)
```

```python
# ============================================================
# CELL 7: Categorical Churn Rate Analysis
# ============================================================

categorical_features_eda = [
    "gender",
    "SeniorCitizen",
    "Partner",
    "Dependents",
    "PhoneService",
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaperlessBilling",
    "PaymentMethod"
]


categorical_summary_list = []


for feature in categorical_features_eda:

    summary = (
        df_eda
        .groupby(
            feature,
            dropna=False
        )[TARGET]
        .agg(
            Total_Customers="count",
            Churned_Customers="sum",
            Churn_Rate="mean"
        )
        .reset_index()
    )

    summary["Feature"] = feature

    summary["Churn_Rate_Percentage"] = (
        summary["Churn_Rate"] * 100
    ).round(2)

    summary = summary[
        [
            "Feature",
            feature,
            "Total_Customers",
            "Churned_Customers",
            "Churn_Rate",
            "Churn_Rate_Percentage"
        ]
    ]

    # Standard category column name
    summary = summary.rename(
        columns={
            feature: "Category"
        }
    )

    categorical_summary_list.append(
        summary
    )


categorical_churn_summary = pd.concat(
    categorical_summary_list,
    ignore_index=True
)


display(
    categorical_churn_summary.head(30)
)


categorical_summary_path = os.path.join(
    EDA_DIR,
    "categorical_churn_summary.csv"
)

categorical_churn_summary.to_csv(
    categorical_summary_path,
    index=False
)


print(
    "\n✅ Full categorical churn summary saved."
)
```

```python
# ============================================================
# CELL 8: Highest Churn Category per Feature
# ============================================================

highest_churn_categories = (
    categorical_churn_summary
    .sort_values(
        [
            "Feature",
            "Churn_Rate"
        ],
        ascending=[
            True,
            False
        ]
    )
    .groupby(
        "Feature",
        as_index=False
    )
    .first()
)


highest_churn_categories = (
    highest_churn_categories[
        [
            "Feature",
            "Category",
            "Total_Customers",
            "Churned_Customers",
            "Churn_Rate_Percentage"
        ]
    ]
)


display(
    highest_churn_categories
)


highest_category_path = os.path.join(
    EDA_DIR,
    "highest_churn_category_per_feature.csv"
)


highest_churn_categories.to_csv(
    highest_category_path,
    index=False
)


print(
    "✅ Highest-risk category summary saved."
)
```

```python
# ============================================================
# CELL 9: Categorical Churn Plot Function
# ============================================================

def plot_categorical_churn_rate(
    feature,
    dataframe,
    target,
    output_directory
):

    plot_data = (
        dataframe
        .groupby(
            feature,
            dropna=False
        )[target]
        .agg(
            Customers="count",
            Churn_Rate="mean"
        )
        .reset_index()
    )

    plot_data["Churn_Rate_Percentage"] = (
        plot_data["Churn_Rate"] * 100
    )

    plot_data = plot_data.sort_values(
        "Churn_Rate_Percentage",
        ascending=False
    )


    fig, ax = plt.subplots(
        figsize=(8, 5)
    )

    bars = ax.bar(
        plot_data[feature].astype(str),
        plot_data["Churn_Rate_Percentage"]
    )


    ax.set_title(
        f"Churn Rate by {feature}"
    )

    ax.set_xlabel(
        feature
    )

    ax.set_ylabel(
        "Churn Rate (%)"
    )

    ax.grid(
        axis="y",
        alpha=0.20
    )


    plt.xticks(
        rotation=30,
        ha="right"
    )


    for bar, rate in zip(
        bars,
        plot_data[
            "Churn_Rate_Percentage"
        ]
    ):

        ax.text(
            bar.get_x()
            + bar.get_width() / 2,

            bar.get_height(),

            f"{rate:.1f}%",

            ha="center",
            va="bottom",
            fontsize=9
        )


    plt.tight_layout()


    safe_name = (
        feature
        .replace(" ", "_")
        .lower()
    )


    save_path = os.path.join(
        output_directory,
        f"{safe_name}_vs_churn.png"
    )


    plt.savefig(
        save_path,
        dpi=300,
        bbox_inches="tight"
    )

    plt.show()
    plt.close()


    return plot_data, save_path


print(
    "✅ Categorical plotting function ready."
)
```

```python
# ============================================================
# CELL 10: Contract vs Churn
# ============================================================

contract_plot_data, contract_plot_path = (
    plot_categorical_churn_rate(
        feature="Contract",
        dataframe=df_eda,
        target=TARGET,
        output_directory=EDA_DIR
    )
)


display(
    contract_plot_data
)

print(
    "\nSaved:",
    contract_plot_path
)
```


# PHASE 05 — Hold-Out Split                           [ 80% Development / 20% Final Test ]


```python
# ============================================================
# PHASE 05 — STRATIFIED HOLD-OUT SPLIT
# CELL 1: Import Split Utility
# ============================================================

from sklearn.model_selection import train_test_split

print("✅ train_test_split imported successfully.")
```

```python
# ============================================================
# CELL 2: Separate X and y
# ============================================================

TARGET = "Churn"

X = df_clean.drop(
    columns=[TARGET]
).copy()

y = df_clean[TARGET].copy()


print("=" * 75)
print("FEATURE / TARGET SEPARATION")
print("=" * 75)

print("X shape:", X.shape)
print("y shape:", y.shape)

print("\nNumber of predictor features:")
print(X.shape[1])

print("\nTarget distribution:")
print(y.value_counts().sort_index())
```

```python
# ============================================================
# CELL 3: Preserve Original Row Indices
# ============================================================

source_indices = df_clean.index.to_series(
    name="source_index"
)

print("Source index count :", len(source_indices))
print("Unique indices     :", source_indices.nunique())

print("\nFirst 5 source indices:")
print(source_indices.head())
```

```python
# ============================================================
# CELL 4: Stratified 80/20 Hold-Out Split
# ============================================================

TEST_SIZE = 0.20
SPLIT_SEED = 42


X_dev, X_test, y_dev, y_test = train_test_split(
    X,
    y,
    test_size=TEST_SIZE,
    random_state=SPLIT_SEED,
    stratify=y
)


print("✅ Stratified split completed.")

print("\nDevelopment set:")
print("X_dev:", X_dev.shape)
print("y_dev:", y_dev.shape)

print("\nFinal test set:")
print("X_test:", X_test.shape)
print("y_test:", y_test.shape)
```

```python
# ============================================================
# CELL 5: Verify Development/Test Separation
# ============================================================

dev_indices = set(X_dev.index)

test_indices = set(X_test.index)

overlap_indices = dev_indices.intersection(
    test_indices
)


print("=" * 75)
print("SPLIT SEPARATION CHECK")
print("=" * 75)

print(
    "Development indices:",
    len(dev_indices)
)

print(
    "Test indices       :",
    len(test_indices)
)

print(
    "Overlapping indices:",
    len(overlap_indices)
)


print(
    "\nCombined unique rows:",
    len(
        dev_indices.union(
            test_indices
        )
    )
)


if len(overlap_indices) == 0:
    print(
        "\n✅ Development and final test sets are completely separate."
    )
else:
    print(
        "\n❌ ERROR: Development/Test overlap detected."
    )
```

```python
# ============================================================
# CELL 6: Verify Stratified Class Distribution
# ============================================================

def class_distribution_table(
    target_series,
    dataset_name
):

    counts = (
        target_series
        .value_counts()
        .sort_index()
    )

    percentages = (
        target_series
        .value_counts(
            normalize=True
        )
        .sort_index()
        * 100
    )


    summary = pd.DataFrame({
        "Dataset": dataset_name,
        "Class": counts.index,
        "Count": counts.values,
        "Percentage": percentages.values
    })

    return summary


full_distribution = (
    class_distribution_table(
        y,
        "Full Dataset"
    )
)

dev_distribution = (
    class_distribution_table(
        y_dev,
        "Development"
    )
)

test_distribution = (
    class_distribution_table(
        y_test,
        "Final Test"
    )
)


split_class_distribution = pd.concat(
    [
        full_distribution,
        dev_distribution,
        test_distribution
    ],
    ignore_index=True
)


split_class_distribution[
    "Class_Name"
] = split_class_distribution[
    "Class"
].map({
    0: "Non-Churn",
    1: "Churn"
})


split_class_distribution[
    "Percentage"
] = split_class_distribution[
    "Percentage"
].round(2)


display(
    split_class_distribution[
        [
            "Dataset",
            "Class",
            "Class_Name",
            "Count",
            "Percentage"
        ]
    ]
)
```

```python
# ============================================================
# CELL 7: Missing Values After Split
# ============================================================

dev_missing = (
    X_dev
    .isna()
    .sum()
)

dev_missing = dev_missing[
    dev_missing > 0
]


test_missing = (
    X_test
    .isna()
    .sum()
)

test_missing = test_missing[
    test_missing > 0
]


print("=" * 75)
print("DEVELOPMENT SET MISSING VALUES")
print("=" * 75)

if dev_missing.empty:
    print("None")
else:
    print(dev_missing)


print("\n" + "=" * 75)
print("FINAL TEST SET MISSING VALUES")
print("=" * 75)

if test_missing.empty:
    print("None")
else:
    print(test_missing)


print(
    "\nTotal missing values across both sets:",
    int(
        X_dev.isna().sum().sum()
        +
        X_test.isna().sum().sum()
    )
)
```

```python
# ============================================================
# CELL 8: Final Test Lock
# ============================================================

FINAL_TEST_LOCKED = True


print("=" * 75)
print("FINAL TEST SET STATUS")
print("=" * 75)

print(
    "FINAL_TEST_LOCKED =",
    FINAL_TEST_LOCKED
)

print(
    "\n🔒 Final test set is now LOCKED."
)

print(
    "It must NOT be used for:"
)

print(" - preprocessing decisions")
print(" - imbalance strategy selection")
print(" - model selection")
print(" - hyperparameter tuning")
print(" - threshold optimization")
print(" - feature selection")


print(
    "\nIt will be opened only after the final "
    "model and threshold have been selected."
)
```

```python
# ============================================================
# CELL 9: Save Reproducible Split Indices
# ============================================================

SPLIT_DIR = OUTPUT_DIRS["03_split"]

os.makedirs(
    SPLIT_DIR,
    exist_ok=True
)


split_index_table = pd.DataFrame({
    "source_index": df_clean.index
})


split_index_table["split"] = np.where(
    split_index_table[
        "source_index"
    ].isin(
        X_test.index
    ),
    "final_test",
    "development"
)


split_index_path = os.path.join(
    SPLIT_DIR,
    "split_indices.csv"
)


split_index_table.to_csv(
    split_index_path,
    index=False
)


print(
    "✅ Split indices saved."
)

print(
    split_index_path
)


print("\nSplit counts:")

print(
    split_index_table[
        "split"
    ].value_counts()
)
```

```python
# ============================================================
# CELL 10: Save Development Dataset
# ============================================================

development_data = X_dev.copy()

development_data[TARGET] = y_dev


development_path = os.path.join(
    SPLIT_DIR,
    "development_set.csv"
)


development_data.to_csv(
    development_path,
    index=True,
    index_label="source_index"
)


print(
    "✅ Development set saved."
)

print(
    development_path
)

print(
    "\nShape:",
    development_data.shape
)
```

```python
# ============================================================
# CELL 11: Save Locked Final Test Dataset
# ============================================================

final_test_data = X_test.copy()

final_test_data[TARGET] = y_test


final_test_path = os.path.join(
    SPLIT_DIR,
    "final_test_LOCKED.csv"
)


final_test_data.to_csv(
    final_test_path,
    index=True,
    index_label="source_index"
)


print(
    "✅ Final test dataset saved."
)

print(
    final_test_path
)

print(
    "\nShape:",
    final_test_data.shape
)

print(
    "\n🔒 DO NOT USE THIS DATASET UNTIL PHASE 13."
)
```

```python
# ============================================================
# CELL 12: Save Split Metadata
# ============================================================

split_metadata = {

    "total_samples":
        int(len(df_clean)),

    "development_samples":
        int(len(X_dev)),

    "final_test_samples":
        int(len(X_test)),

    "test_size":
        TEST_SIZE,

    "random_state":
        SPLIT_SEED,

    "stratified":
        True,

    "target":
        TARGET,

    "development_non_churn":
        int(
            (y_dev == 0).sum()
        ),

    "development_churn":
        int(
            (y_dev == 1).sum()
        ),

    "test_non_churn":
        int(
            (y_test == 0).sum()
        ),

    "test_churn":
        int(
            (y_test == 1).sum()
        ),

    "development_totalcharges_missing":
        int(
            X_dev[
                "TotalCharges"
            ].isna().sum()
        ),

    "test_totalcharges_missing":
        int(
            X_test[
                "TotalCharges"
            ].isna().sum()
        ),

    "final_test_locked":
        True
}


split_metadata_path = os.path.join(
    SPLIT_DIR,
    "phase05_split_metadata.json"
)


with open(
    split_metadata_path,
    "w"
) as f:

    json.dump(
        split_metadata,
        f,
        indent=4
    )


print(
    "✅ Split metadata saved."
)

print(
    split_metadata_path
)
```

```python
# ============================================================
# CELL 13: Verify Saved Split Files
# ============================================================

required_split_files = [
    split_index_path,
    development_path,
    final_test_path,
    split_metadata_path
]


print("=" * 75)
print("SAVED SPLIT FILES")
print("=" * 75)


for path in required_split_files:

    status = os.path.isfile(path)

    symbol = (
        "✅"
        if status
        else "❌"
    )

    print(
        symbol,
        path
    )
```

# PHASE 06 — Leakage-Safe Preprocessing Pipeline Design

```python
# ============================================================
# PHASE 06 — LEAKAGE-SAFE PREPROCESSING PIPELINE
# CELL 1: Import Required Tools
# ============================================================

import sklearn
import imblearn

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline as SkPipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import (
    StandardScaler,
    OneHotEncoder,
    OrdinalEncoder
)
from sklearn.model_selection import StratifiedKFold

from imblearn.over_sampling import SMOTENC
from imblearn.under_sampling import EditedNearestNeighbours
from imblearn.pipeline import Pipeline as ImbPipeline


print("=" * 75)
print("PHASE 06 — PREPROCESSING ENVIRONMENT")
print("=" * 75)

print("scikit-learn version :", sklearn.__version__)
print("imbalanced-learn     :", imblearn.__version__)

print("\n✅ Preprocessing libraries imported successfully.")
```

```python
# ============================================================
# CELL 2: Define Feature Groups
# ============================================================

CONTINUOUS_FEATURES = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]


CATEGORICAL_FEATURES = [
    col
    for col in X_dev.columns
    if col not in CONTINUOUS_FEATURES
]


print("=" * 75)
print("FEATURE GROUPS")
print("=" * 75)

print(
    f"\nContinuous features "
    f"({len(CONTINUOUS_FEATURES)}):"
)

for feature in CONTINUOUS_FEATURES:
    print(" -", feature)


print(
    f"\nCategorical / Binary features "
    f"({len(CATEGORICAL_FEATURES)}):"
)

for feature in CATEGORICAL_FEATURES:
    print(" -", feature)


print("\nTotal predictors:")
print(
    len(CONTINUOUS_FEATURES)
    +
    len(CATEGORICAL_FEATURES)
)
```

```python
# ============================================================
# CELL 3: Validate Feature Groups
# ============================================================

all_defined_features = (
    CONTINUOUS_FEATURES
    +
    CATEGORICAL_FEATURES
)


duplicate_defined_features = [
    feature
    for feature in set(
        CONTINUOUS_FEATURES
    ).intersection(
        CATEGORICAL_FEATURES
    )
]


missing_defined_features = (
    set(X_dev.columns)
    -
    set(all_defined_features)
)


extra_defined_features = (
    set(all_defined_features)
    -
    set(X_dev.columns)
)


print("=" * 75)
print("FEATURE GROUP INTEGRITY")
print("=" * 75)

print(
    "Duplicate between groups:",
    duplicate_defined_features
)

print(
    "Features not assigned:",
    list(missing_defined_features)
)

print(
    "Unknown defined features:",
    list(extra_defined_features)
)


print(
    "\nAll 19 predictors assigned:",
    (
        len(all_defined_features)
        ==
        X_dev.shape[1]
        and
        len(duplicate_defined_features) == 0
        and
        len(missing_defined_features) == 0
        and
        len(extra_defined_features) == 0
    )
)
```

```python
# ============================================================
# CELL 4: Version-Safe OneHotEncoder
# ============================================================

def make_onehot_encoder():

    try:
        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse_output=False
        )

    except TypeError:

        encoder = OneHotEncoder(
            handle_unknown="ignore",
            sparse=False
        )

    return encoder


test_encoder = make_onehot_encoder()

print(
    "✅ Version-safe OneHotEncoder created."
)

print(test_encoder)
```

```python
# ============================================================
# CELL 5: Build Pre-Sampling Transformer
# ============================================================

def make_pre_sampler_transformer():

    # --------------------------------------------------------
    # Continuous variables
    # --------------------------------------------------------
    numeric_pipeline = SkPipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="median"
                )
            ),
            (
                "scaler",
                StandardScaler()
            )
        ]
    )


    # --------------------------------------------------------
    # Categorical / Binary variables
    # --------------------------------------------------------
    categorical_pipeline = SkPipeline(
        steps=[
            (
                "imputer",
                SimpleImputer(
                    strategy="most_frequent"
                )
            ),
            (
                "ordinal",
                OrdinalEncoder(
                    handle_unknown="use_encoded_value",
                    unknown_value=-1
                )
            )
        ]
    )


    pre_sampler = ColumnTransformer(
        transformers=[
            (
                "continuous",
                numeric_pipeline,
                CONTINUOUS_FEATURES
            ),
            (
                "categorical",
                categorical_pipeline,
                CATEGORICAL_FEATURES
            )
        ],
        remainder="drop",
        verbose_feature_names_out=False,
        sparse_threshold=0
    )

    return pre_sampler


pre_sampler_example = (
    make_pre_sampler_transformer()
)

print(
    "✅ Pre-sampling transformer created."
)

print(pre_sampler_example)
```

```python
# ============================================================
# CELL 6: Define SMOTENC Feature Positions
# ============================================================

N_CONTINUOUS = len(
    CONTINUOUS_FEATURES
)

N_CATEGORICAL = len(
    CATEGORICAL_FEATURES
)


CONTINUOUS_POSITIONS = list(
    range(
        0,
        N_CONTINUOUS
    )
)


CATEGORICAL_POSITIONS = list(
    range(
        N_CONTINUOUS,
        N_CONTINUOUS
        +
        N_CATEGORICAL
    )
)


print(
    "Continuous positions:"
)

print(
    CONTINUOUS_POSITIONS
)


print(
    "\nCategorical positions for SMOTENC:"
)

print(
    CATEGORICAL_POSITIONS
)


print(
    "\nTotal stage-1 features:",
    (
        len(CONTINUOUS_POSITIONS)
        +
        len(CATEGORICAL_POSITIONS)
    )
)
```

```python
# ============================================================
# CELL 7: Define Future Sampling Components
# ============================================================

def make_smotenc():

    return SMOTENC(
        categorical_features=
            CATEGORICAL_POSITIONS,

        random_state=SEED,

        k_neighbors=5
    )


def make_enn():

    return EditedNearestNeighbours(
        n_neighbors=3
    )


smotenc_example = make_smotenc()
enn_example = make_enn()


print("✅ Future samplers configured.")

print("\nSMOTENC:")
print(smotenc_example)

print("\nENN:")
print(enn_example)


print(
    "\nNOTE: No resampling has been "
    "performed in PHASE 06."
)
```

```python
# ============================================================
# CELL 8: Build Post-Sampling Encoder
# ============================================================

def make_post_sampler_encoder():

    post_encoder = ColumnTransformer(
        transformers=[
            (
                "continuous",
                "passthrough",
                CONTINUOUS_POSITIONS
            ),

            (
                "categorical",
                make_onehot_encoder(),
                CATEGORICAL_POSITIONS
            )
        ],

        remainder="drop",

        verbose_feature_names_out=False,

        sparse_threshold=0
    )

    return post_encoder


post_encoder_example = (
    make_post_sampler_encoder()
)


print(
    "✅ Post-sampling One-Hot encoder created."
)

print(post_encoder_example)
```

```python
# ============================================================
# CELL 9: Create Fixed Stratified 5-Fold CV
# ============================================================

CV_FOLDS = 5

cv = StratifiedKFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)


print(
    "✅ Stratified 5-Fold CV configured."
)

print(
    "Number of folds:",
    CV_FOLDS
)

print(
    "Shuffle:",
    True
)

print(
    "Random state:",
    SEED
)
```

```python
# ============================================================
# CELL 10: Create Diagnostic Development Fold
# ============================================================

diagnostic_split = next(
    cv.split(
        X_dev,
        y_dev
    )
)


diag_train_idx = diagnostic_split[0]
diag_valid_idx = diagnostic_split[1]


X_diag_train = (
    X_dev
    .iloc[diag_train_idx]
    .copy()
)

y_diag_train = (
    y_dev
    .iloc[diag_train_idx]
    .copy()
)


X_diag_valid = (
    X_dev
    .iloc[diag_valid_idx]
    .copy()
)

y_diag_valid = (
    y_dev
    .iloc[diag_valid_idx]
    .copy()
)


print("=" * 75)
print("DIAGNOSTIC DEVELOPMENT FOLD")
print("=" * 75)

print(
    "Training fold:",
    X_diag_train.shape
)

print(
    "Validation fold:",
    X_diag_valid.shape
)


print("\nTraining target distribution:")

print(
    y_diag_train
    .value_counts()
    .sort_index()
)


print("\nValidation target distribution:")

print(
    y_diag_valid
    .value_counts()
    .sort_index()
)
```

```python
# ============================================================
# CELL 11: Diagnostic Stage-1 Preprocessing
# ============================================================

diagnostic_pre_sampler = (
    make_pre_sampler_transformer()
)


X_diag_train_stage1 = (
    diagnostic_pre_sampler
    .fit_transform(
        X_diag_train,
        y_diag_train
    )
)


X_diag_valid_stage1 = (
    diagnostic_pre_sampler
    .transform(
        X_diag_valid
    )
)


print("=" * 75)
print("STAGE-1 PREPROCESSING CHECK")
print("=" * 75)

print(
    "Train output shape:",
    X_diag_train_stage1.shape
)

print(
    "Validation output shape:",
    X_diag_valid_stage1.shape
)


print(
    "\nTrain NaN count:",
    int(
        np.isnan(
            X_diag_train_stage1
        ).sum()
    )
)


print(
    "Validation NaN count:",
    int(
        np.isnan(
            X_diag_valid_stage1
        ).sum()
    )
)
```

```python
# ============================================================
# CELL 12: Verify Fold-Based TotalCharges Imputation
# ============================================================

numeric_pipeline_fitted = (
    diagnostic_pre_sampler
    .named_transformers_[
        "continuous"
    ]
)


numeric_imputer_fitted = (
    numeric_pipeline_fitted
    .named_steps[
        "imputer"
    ]
)


imputer_statistics = (
    numeric_imputer_fitted
    .statistics_
)


imputation_statistics = pd.DataFrame({
    "Feature":
        CONTINUOUS_FEATURES,

    "Training_Fold_Imputation_Value":
        imputer_statistics
})


display(
    imputation_statistics
)


totalcharges_position = (
    CONTINUOUS_FEATURES
    .index(
        "TotalCharges"
    )
)


diagnostic_totalcharges_median = (
    imputer_statistics[
        totalcharges_position
    ]
)


manual_training_fold_median = (
    X_diag_train[
        "TotalCharges"
    ]
    .median()
)


print(
    "\nPipeline TotalCharges median:",
    diagnostic_totalcharges_median
)


print(
    "Manual TRAINING FOLD median:",
    manual_training_fold_median
)


print(
    "\nValues match:",
    np.isclose(
        diagnostic_totalcharges_median,
        manual_training_fold_median
    )
)
```

```python
# ============================================================
# CELL 13: Diagnostic Post-Sampling Encoding
# ============================================================

diagnostic_post_encoder = (
    make_post_sampler_encoder()
)


X_diag_train_final = (
    diagnostic_post_encoder
    .fit_transform(
        X_diag_train_stage1,
        y_diag_train
    )
)


X_diag_valid_final = (
    diagnostic_post_encoder
    .transform(
        X_diag_valid_stage1
    )
)


print("=" * 75)
print("FINAL ENCODED FEATURE CHECK")
print("=" * 75)

print(
    "Training encoded shape:",
    X_diag_train_final.shape
)

print(
    "Validation encoded shape:",
    X_diag_valid_final.shape
)


print(
    "\nTraining NaN:",
    int(
        np.isnan(
            X_diag_train_final
        ).sum()
    )
)


print(
    "Validation NaN:",
    int(
        np.isnan(
            X_diag_valid_final
        ).sum()
    )
)


print(
    "\nSame number of encoded features:",
    (
        X_diag_train_final.shape[1]
        ==
        X_diag_valid_final.shape[1]
    )
)
```

```python
# ============================================================
# CELL 14: Extract Encoded Feature Names
# ============================================================

stage1_feature_names = (
    CONTINUOUS_FEATURES
    +
    CATEGORICAL_FEATURES
)


encoded_feature_names = (
    diagnostic_post_encoder
    .get_feature_names_out(
        stage1_feature_names
    )
    .tolist()
)


encoded_feature_table = pd.DataFrame({
    "Feature_Index":
        range(
            len(
                encoded_feature_names
            )
        ),

    "Encoded_Feature":
        encoded_feature_names
})


print(
    "Number of final encoded features:",
    len(
        encoded_feature_names
    )
)


display(
    encoded_feature_table
    .head(50)
)
```

```python
# ============================================================
# PHASE 07 — RQ1 IMBALANCE EXPERIMENTS
# CELL 1: Import Models and Evaluation Metrics
# ============================================================

import time
import joblib

from sklearn.base import clone

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    average_precision_score,
    confusion_matrix
)


# ------------------------------------------------------------
# XGBoost
# ------------------------------------------------------------
try:
    from xgboost import XGBClassifier

    XGBOOST_AVAILABLE = True

except ImportError:
    XGBOOST_AVAILABLE = False


# ------------------------------------------------------------
# LightGBM
# ------------------------------------------------------------
try:
    from lightgbm import LGBMClassifier

    LIGHTGBM_AVAILABLE = True

except ImportError:
    LIGHTGBM_AVAILABLE = False


print("=" * 78)
print("PHASE 07 — MODEL ENVIRONMENT")
print("=" * 78)

print(
    "XGBoost available :",
    XGBOOST_AVAILABLE
)

print(
    "LightGBM available:",
    LIGHTGBM_AVAILABLE
)


if not XGBOOST_AVAILABLE:
    raise ImportError(
        "XGBoost is not available in this Kaggle environment."
    )


if not LIGHTGBM_AVAILABLE:
    raise ImportError(
        "LightGBM is not available in this Kaggle environment."
    )


print(
    "\n✅ All five research models are available."
)
```

```python
# ============================================================
# CELL 2: Create Fixed Stratified CV Splits
# ============================================================

cv_splits = list(
    cv.split(
        X_dev,
        y_dev
    )
)


print("=" * 78)
print("FIXED STRATIFIED CV SPLITS")
print("=" * 78)

print(
    "Number of folds:",
    len(cv_splits)
)


cv_distribution_rows = []


for fold_number, (
    train_idx,
    valid_idx
) in enumerate(
    cv_splits,
    start=1
):

    y_train_fold = (
        y_dev
        .iloc[train_idx]
    )

    y_valid_fold = (
        y_dev
        .iloc[valid_idx]
    )

    cv_distribution_rows.append({
        "Fold": fold_number,

        "Train_Rows":
            len(train_idx),

        "Train_NonChurn":
            int(
                (y_train_fold == 0).sum()
            ),

        "Train_Churn":
            int(
                (y_train_fold == 1).sum()
            ),

        "Validation_Rows":
            len(valid_idx),

        "Validation_NonChurn":
            int(
                (y_valid_fold == 0).sum()
            ),

        "Validation_Churn":
            int(
                (y_valid_fold == 1).sum()
            )
    })


cv_distribution_table = pd.DataFrame(
    cv_distribution_rows
)


display(
    cv_distribution_table
)
```

```python
# ============================================================
# CELL 3: Define Baseline Model Factory
# ============================================================

MODEL_NAMES = [
    "Logistic Regression",
    "Decision Tree",
    "Random Forest",
    "XGBoost",
    "LightGBM"
]


STRATEGIES = [
    "Original",
    "Class Weight",
    "SMOTENC",
    "SMOTENC + ENN"
]


def make_model(
    model_name,
    strategy,
    y_training
):

    # --------------------------------------------------------
    # Training-fold class ratio
    # --------------------------------------------------------
    negative_count = int(
        (y_training == 0).sum()
    )

    positive_count = int(
        (y_training == 1).sum()
    )

    scale_pos_weight = (
        negative_count
        /
        positive_count
    )


    use_class_weight = (
        strategy
        ==
        "Class Weight"
    )


    # --------------------------------------------------------
    # Logistic Regression
    # --------------------------------------------------------
    if model_name == "Logistic Regression":

        return LogisticRegression(
            max_iter=2000,

            class_weight=(
                "balanced"
                if use_class_weight
                else None
            ),

            random_state=SEED
        )


    # --------------------------------------------------------
    # Decision Tree
    # --------------------------------------------------------
    elif model_name == "Decision Tree":

        return DecisionTreeClassifier(
            class_weight=(
                "balanced"
                if use_class_weight
                else None
            ),

            random_state=SEED
        )


    # --------------------------------------------------------
    # Random Forest
    # --------------------------------------------------------
    elif model_name == "Random Forest":

        return RandomForestClassifier(
            n_estimators=200,

            class_weight=(
                "balanced"
                if use_class_weight
                else None
            ),

            random_state=SEED,

            n_jobs=-1
        )


    # --------------------------------------------------------
    # XGBoost
    # --------------------------------------------------------
    elif model_name == "XGBoost":

        return XGBClassifier(
            n_estimators=250,

            learning_rate=0.05,

            max_depth=4,

            subsample=0.90,

            colsample_bytree=0.90,

            objective="binary:logistic",

            eval_metric="logloss",

            scale_pos_weight=(
                scale_pos_weight
                if use_class_weight
                else 1.0
            ),

            random_state=SEED,

            n_jobs=-1,

            tree_method="hist"
        )


    # --------------------------------------------------------
    # LightGBM
    # --------------------------------------------------------
    elif model_name == "LightGBM":

        return LGBMClassifier(
            n_estimators=250,

            learning_rate=0.05,

            class_weight=(
                "balanced"
                if use_class_weight
                else None
            ),

            random_state=SEED,

            n_jobs=-1,

            verbosity=-1
        )


    else:

        raise ValueError(
            f"Unknown model: {model_name}"
        )


print("Models:")
for model in MODEL_NAMES:
    print(" -", model)


print("\nStrategies:")
for strategy in STRATEGIES:
    print(" -", strategy)


print(
    "\nTotal combinations:",
    len(MODEL_NAMES)
    *
    len(STRATEGIES)
)
```

```python
# ============================================================
# CELL 4: Single-Fold Experiment Function
# ============================================================

def run_single_fold_experiment(
    model_name,
    strategy,
    fold_number,
    train_idx,
    valid_idx
):

    # ========================================================
    # 1. Extract Fold Data
    # ========================================================

    X_train_fold = (
        X_dev
        .iloc[train_idx]
        .copy()
    )

    y_train_fold = (
        y_dev
        .iloc[train_idx]
        .copy()
    )

    X_valid_fold = (
        X_dev
        .iloc[valid_idx]
        .copy()
    )

    y_valid_fold = (
        y_dev
        .iloc[valid_idx]
        .copy()
    )


    # ========================================================
    # 2. Leakage-Safe Stage-1 Preprocessing
    # ========================================================

    pre_sampler = (
        make_pre_sampler_transformer()
    )


    X_train_stage1 = (
        pre_sampler
        .fit_transform(
            X_train_fold,
            y_train_fold
        )
    )


    X_valid_stage1 = (
        pre_sampler
        .transform(
            X_valid_fold
        )
    )


    # ========================================================
    # 3. Optional SMOTENC
    # ========================================================

    if strategy in [
        "SMOTENC",
        "SMOTENC + ENN"
    ]:

        sampler = make_smotenc()

        (
            X_train_sampled_stage1,
            y_train_sampled
        ) = sampler.fit_resample(
            X_train_stage1,
            y_train_fold
        )

    else:

        X_train_sampled_stage1 = (
            X_train_stage1
        )

        y_train_sampled = (
            y_train_fold
            .to_numpy()
        )


    # ========================================================
    # 4. Post-Sampling One-Hot Encoding
    # ========================================================

    post_encoder = (
        make_post_sampler_encoder()
    )


    X_train_encoded = (
        post_encoder
        .fit_transform(
            X_train_sampled_stage1,
            y_train_sampled
        )
    )


    X_valid_encoded = (
        post_encoder
        .transform(
            X_valid_stage1
        )
    )


    # ========================================================
    # 5. Optional ENN Cleaning
    # ========================================================

    if strategy == "SMOTENC + ENN":

        enn = make_enn()

        (
            X_train_final,
            y_train_final
        ) = enn.fit_resample(
            X_train_encoded,
            y_train_sampled
        )

    else:

        X_train_final = (
            X_train_encoded
        )

        y_train_final = (
            np.asarray(
                y_train_sampled
            )
        )


    # ========================================================
    # 6. Create Fold-Specific Model
    # ========================================================

    model = make_model(
        model_name=model_name,
        strategy=strategy,
        y_training=y_train_fold
    )


    # ========================================================
    # 7. Fit Model
    # ========================================================

    start_time = time.time()


    model.fit(
        X_train_final,
        y_train_final
    )


    fit_time = (
        time.time()
        -
        start_time
    )


    # ========================================================
    # 8. Validation Probability
    # ========================================================

    y_probability = (
        model.predict_proba(
            X_valid_encoded
        )[:, 1]
    )


    # Default threshold ONLY for diagnostic classification
    y_prediction = (
        y_probability >= 0.50
    ).astype(int)


    # ========================================================
    # 9. Metrics
    # ========================================================

    accuracy = accuracy_score(
        y_valid_fold,
        y_prediction
    )


    precision = precision_score(
        y_valid_fold,
        y_prediction,
        zero_division=0
    )


    recall = recall_score(
        y_valid_fold,
        y_prediction,
        zero_division=0
    )


    f1 = f1_score(
        y_valid_fold,
        y_prediction,
        zero_division=0
    )


    roc_auc = roc_auc_score(
        y_valid_fold,
        y_probability
    )


    pr_auc_ap = (
        average_precision_score(
            y_valid_fold,
            y_probability
        )
    )


    tn, fp, fn, tp = (
        confusion_matrix(
            y_valid_fold,
            y_prediction,
            labels=[0, 1]
        )
        .ravel()
    )


    # ========================================================
    # 10. Sampling Audit
    # ========================================================

    sampled_negative = int(
        (
            np.asarray(
                y_train_final
            ) == 0
        ).sum()
    )

    sampled_positive = int(
        (
            np.asarray(
                y_train_final
            ) == 1
        ).sum()
    )


    return {

        "Model":
            model_name,

        "Strategy":
            strategy,

        "Fold":
            fold_number,

        "Train_Rows_Original":
            int(
                len(
                    y_train_fold
                )
            ),

        "Train_NonChurn_Original":
            int(
                (
                    y_train_fold
                    == 0
                ).sum()
            ),

        "Train_Churn_Original":
            int(
                (
                    y_train_fold
                    == 1
                ).sum()
            ),

        "Train_Rows_After_Strategy":
            int(
                len(
                    y_train_final
                )
            ),

        "Train_NonChurn_After_Strategy":
            sampled_negative,

        "Train_Churn_After_Strategy":
            sampled_positive,

        "Encoded_Features":
            int(
                X_train_final.shape[1]
            ),

        "Accuracy":
            float(
                accuracy
            ),

        "Precision":
            float(
                precision
            ),

        "Recall":
            float(
                recall
            ),

        "F1":
            float(
                f1
            ),

        "ROC_AUC":
            float(
                roc_auc
            ),

        "PR_AUC_AP":
            float(
                pr_auc_ap
            ),

        "TN":
            int(tn),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TP":
            int(tp),

        "Fit_Time_Seconds":
            float(
                fit_time
            )
    }


print(
    "✅ Single-fold experiment function created."
)
```

```python
# ============================================================
# CELL 5: Smoke Test
# ============================================================

smoke_train_idx, smoke_valid_idx = (
    cv_splits[0]
)


smoke_result = (
    run_single_fold_experiment(
        model_name=
            "Logistic Regression",

        strategy=
            "Original",

        fold_number=
            1,

        train_idx=
            smoke_train_idx,

        valid_idx=
            smoke_valid_idx
    )
)


print("=" * 78)
print("PHASE 07 SMOKE TEST")
print("=" * 78)


for key, value in (
    smoke_result.items()
):

    print(
        f"{key:<32}: {value}"
    )
```

```python
# ============================================================
# CELL 6: Run Full RQ1 Experiment
# ============================================================

rq1_fold_results = []

total_combinations = (
    len(MODEL_NAMES)
    *
    len(STRATEGIES)
)

current_combination = 0


experiment_start_time = (
    time.time()
)


print("=" * 82)
print("STARTING PHASE 07 — FULL RQ1 EXPERIMENT")
print("=" * 82)

print(
    f"Models      : "
    f"{len(MODEL_NAMES)}"
)

print(
    f"Strategies  : "
    f"{len(STRATEGIES)}"
)

print(
    f"CV folds    : "
    f"{len(cv_splits)}"
)

print(
    f"Total fits  : "
    f"{total_combinations * len(cv_splits)}"
)

print("=" * 82)


for model_name in MODEL_NAMES:

    for strategy in STRATEGIES:

        current_combination += 1

        print(
            f"\n[{current_combination:02d}/"
            f"{total_combinations}] "
            f"{model_name} | {strategy}"
        )


        combination_start = (
            time.time()
        )


        for fold_number, (
            train_idx,
            valid_idx
        ) in enumerate(
            cv_splits,
            start=1
        ):

            fold_result = (
                run_single_fold_experiment(
                    model_name=
                        model_name,

                    strategy=
                        strategy,

                    fold_number=
                        fold_number,

                    train_idx=
                        train_idx,

                    valid_idx=
                        valid_idx
                )
            )


            rq1_fold_results.append(
                fold_result
            )


            print(
                f"   Fold {fold_number}: "
                f"PR-AUC="
                f"{fold_result['PR_AUC_AP']:.4f} | "
                f"ROC-AUC="
                f"{fold_result['ROC_AUC']:.4f} | "
                f"Recall="
                f"{fold_result['Recall']:.4f}"
            )


        combination_time = (
            time.time()
            -
            combination_start
        )


        print(
            f"   Completed in "
            f"{combination_time:.2f} sec"
        )


experiment_total_time = (
    time.time()
    -
    experiment_start_time
)


rq1_fold_results_df = (
    pd.DataFrame(
        rq1_fold_results
    )
)


print("\n" + "=" * 82)

print(
    "✅ FULL RQ1 EXPERIMENT COMPLETED"
)

print(
    f"Total model fits : "
    f"{len(rq1_fold_results_df)}"
)

print(
    f"Total runtime    : "
    f"{experiment_total_time / 60:.2f} minutes"
)

print("=" * 82)
```

```python
# ============================================================
# CELL 7: Inspect Raw Fold Results
# ============================================================

print(
    "Fold result shape:",
    rq1_fold_results_df.shape
)


display(
    rq1_fold_results_df.head(20)
)
```

```python
# ============================================================
# CELL 8: Aggregate Cross-Validation Results
# ============================================================

metric_columns = [
    "Accuracy",
    "Precision",
    "Recall",
    "F1",
    "ROC_AUC",
    "PR_AUC_AP"
]


aggregation_dict = {}


for metric in metric_columns:

    aggregation_dict[
        f"{metric}_Mean"
    ] = (
        metric,
        "mean"
    )

    aggregation_dict[
        f"{metric}_STD"
    ] = (
        metric,
        "std"
    )


aggregation_dict[
    "FN_Mean"
] = (
    "FN",
    "mean"
)


aggregation_dict[
    "FP_Mean"
] = (
    "FP",
    "mean"
)


aggregation_dict[
    "Fit_Time_Mean"
] = (
    "Fit_Time_Seconds",
    "mean"
)


rq1_summary = (
    rq1_fold_results_df
    .groupby(
        [
            "Model",
            "Strategy"
        ]
    )
    .agg(
        **aggregation_dict
    )
    .reset_index()
)


rq1_summary = (
    rq1_summary
    .sort_values(
        "PR_AUC_AP_Mean",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


rq1_summary.insert(
    0,
    "Overall_Rank",
    np.arange(
        1,
        len(rq1_summary)
        + 1
    )
)


display(
    rq1_summary
)
```

```python
# ============================================================
# CELL 9: Create Clean Research Results Table
# ============================================================

rq1_display = (
    rq1_summary.copy()
)


numeric_result_columns = (
    rq1_display
    .select_dtypes(
        include=[
            "number"
        ]
    )
    .columns
)


for column in numeric_result_columns:

    if column != "Overall_Rank":

        rq1_display[
            column
        ] = (
            rq1_display[
                column
            ]
            .round(4)
        )


display(
    rq1_display
)
```

```python
# ============================================================
# CELL 10: Best Strategy per Model
# ============================================================

best_strategy_per_model = (
    rq1_summary
    .sort_values(
        [
            "Model",
            "PR_AUC_AP_Mean"
        ],
        ascending=[
            True,
            False
        ]
    )
    .groupby(
        "Model",
        as_index=False
    )
    .first()
)


best_strategy_per_model = (
    best_strategy_per_model[
        [
            "Model",
            "Strategy",
            "PR_AUC_AP_Mean",
            "PR_AUC_AP_STD",
            "ROC_AUC_Mean",
            "Recall_Mean",
            "Precision_Mean",
            "F1_Mean",
            "Accuracy_Mean",
            "FN_Mean",
            "FP_Mean"
        ]
    ]
)


best_strategy_per_model = (
    best_strategy_per_model
    .sort_values(
        "PR_AUC_AP_Mean",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


display(
    best_strategy_per_model.round(4)
)
```

```python
# ============================================================
# CELL 11: Global Best RQ1 Combination
# ============================================================

rq1_champion = (
    rq1_summary
    .iloc[0]
)


print("=" * 78)
print("RQ1 CURRENT CHAMPION — BASELINE CV")
print("=" * 78)


print(
    "Model       :",
    rq1_champion[
        "Model"
    ]
)

print(
    "Strategy    :",
    rq1_champion[
        "Strategy"
    ]
)


print(
    f"PR-AUC (AP) : "
    f"{rq1_champion['PR_AUC_AP_Mean']:.6f} "
    f"± "
    f"{rq1_champion['PR_AUC_AP_STD']:.6f}"
)


print(
    f"ROC-AUC     : "
    f"{rq1_champion['ROC_AUC_Mean']:.6f}"
)


print(
    f"Recall      : "
    f"{rq1_champion['Recall_Mean']:.6f}"
)


print(
    f"Precision   : "
    f"{rq1_champion['Precision_Mean']:.6f}"
)


print(
    f"F1          : "
    f"{rq1_champion['F1_Mean']:.6f}"
)


print(
    f"Accuracy    : "
    f"{rq1_champion['Accuracy_Mean']:.6f}"
)
```

```python
# ============================================================
# CELL 12: Sampling Strategy Audit
# ============================================================

sampling_audit = (
    rq1_fold_results_df[
        [
            "Model",
            "Strategy",
            "Fold",
            "Train_Rows_Original",
            "Train_NonChurn_Original",
            "Train_Churn_Original",
            "Train_Rows_After_Strategy",
            "Train_NonChurn_After_Strategy",
            "Train_Churn_After_Strategy"
        ]
    ]
    .copy()
)


sampling_audit_summary = (
    sampling_audit
    .groupby(
        "Strategy"
    )
    .agg(
        Original_Rows_Mean=(
            "Train_Rows_Original",
            "mean"
        ),

        Final_Rows_Mean=(
            "Train_Rows_After_Strategy",
            "mean"
        ),

        Final_NonChurn_Mean=(
            "Train_NonChurn_After_Strategy",
            "mean"
        ),

        Final_Churn_Mean=(
            "Train_Churn_After_Strategy",
            "mean"
        )
    )
    .reset_index()
)


display(
    sampling_audit_summary.round(2)
)
```

```python
# ============================================================
# CELL 5: Search Iteration Configuration
# ============================================================

SEARCH_ITERATIONS = {

    "Logistic Regression":
        20,

    "Decision Tree":
        30,

    "Random Forest":
        30,

    "XGBoost":
        35,

    "LightGBM":
        35
}


print("=" * 70)
print("RANDOM SEARCH ITERATIONS")
print("=" * 70)


for model_name, n_iter in (
    SEARCH_ITERATIONS.items()
):

    print(
        f"{model_name:<22}: "
        f"{n_iter}"
    )
```

```python
# ============================================================
# CELL 6: Tunable Model Factory
# ============================================================

def make_tunable_model(
    model_name,
    strategy,
    y_training,
    params
):

    negative_count = int(
        (y_training == 0).sum()
    )

    positive_count = int(
        (y_training == 1).sum()
    )


    pos_weight = (
        negative_count
        /
        positive_count
    )


    use_weight = (
        strategy
        ==
        "Class Weight"
    )


    if model_name == "Logistic Regression":

        return LogisticRegression(

            C=params["C"],

            solver=params["solver"],

            max_iter=3000,

            class_weight=(
                "balanced"
                if use_weight
                else None
            ),

            random_state=SEED
        )


    elif model_name == "Decision Tree":

        return DecisionTreeClassifier(

            max_depth=
                params["max_depth"],

            min_samples_split=
                params["min_samples_split"],

            min_samples_leaf=
                params["min_samples_leaf"],

            criterion=
                params["criterion"],

            class_weight=(
                "balanced"
                if use_weight
                else None
            ),

            random_state=SEED
        )


    elif model_name == "Random Forest":

        return RandomForestClassifier(

            n_estimators=
                params["n_estimators"],

            max_depth=
                params["max_depth"],

            min_samples_split=
                params["min_samples_split"],

            min_samples_leaf=
                params["min_samples_leaf"],

            max_features=
                params["max_features"],

            class_weight=(
                "balanced"
                if use_weight
                else None
            ),

            random_state=SEED,

            n_jobs=-1
        )


    elif model_name == "XGBoost":

        return XGBClassifier(

            n_estimators=
                params["n_estimators"],

            learning_rate=
                params["learning_rate"],

            max_depth=
                params["max_depth"],

            min_child_weight=
                params["min_child_weight"],

            subsample=
                params["subsample"],

            colsample_bytree=
                params["colsample_bytree"],

            reg_alpha=
                params["reg_alpha"],

            reg_lambda=
                params["reg_lambda"],

            objective=
                "binary:logistic",

            eval_metric=
                "logloss",

            scale_pos_weight=(
                pos_weight
                if use_weight
                else 1.0
            ),

            random_state=SEED,

            n_jobs=-1,

            tree_method="hist"
        )


    elif model_name == "LightGBM":

        return LGBMClassifier(

            n_estimators=
                params["n_estimators"],

            learning_rate=
                params["learning_rate"],

            num_leaves=
                params["num_leaves"],

            max_depth=
                params["max_depth"],

            min_child_samples=
                params["min_child_samples"],

            subsample=
                params["subsample"],

            colsample_bytree=
                params["colsample_bytree"],

            reg_alpha=
                params["reg_alpha"],

            reg_lambda=
                params["reg_lambda"],

            class_weight=(
                "balanced"
                if use_weight
                else None
            ),

            random_state=SEED,

            n_jobs=-1,

            verbosity=-1
        )


    raise ValueError(
        f"Unknown model: {model_name}"
    )


print(
    "✅ Tunable model factory created."
)
```

```python
# ============================================================
# CELL 7: Evaluate One Hyperparameter Configuration
# ============================================================

def evaluate_tuning_configuration(
    model_name,
    strategy,
    params
):

    fold_results = []


    for fold_number, (
        train_idx,
        valid_idx
    ) in enumerate(
        cv_splits,
        start=1
    ):

        X_train_fold = (
            X_dev
            .iloc[train_idx]
            .copy()
        )

        y_train_fold = (
            y_dev
            .iloc[train_idx]
            .copy()
        )


        X_valid_fold = (
            X_dev
            .iloc[valid_idx]
            .copy()
        )

        y_valid_fold = (
            y_dev
            .iloc[valid_idx]
            .copy()
        )


        # ----------------------------------------------------
        # Stage-1 preprocessing
        # ----------------------------------------------------

        pre_sampler = (
            make_pre_sampler_transformer()
        )


        X_train_stage1 = (
            pre_sampler.fit_transform(
                X_train_fold,
                y_train_fold
            )
        )


        X_valid_stage1 = (
            pre_sampler.transform(
                X_valid_fold
            )
        )


        # ----------------------------------------------------
        # SMOTENC if required
        # ----------------------------------------------------

        if strategy in [
            "SMOTENC",
            "SMOTENC + ENN"
        ]:

            smote = (
                make_smotenc()
            )

            (
                X_train_stage1,
                y_train_processed
            ) = smote.fit_resample(
                X_train_stage1,
                y_train_fold
            )

        else:

            y_train_processed = (
                y_train_fold
                .to_numpy()
            )


        # ----------------------------------------------------
        # One-hot encoding
        # ----------------------------------------------------

        post_encoder = (
            make_post_sampler_encoder()
        )


        X_train_encoded = (
            post_encoder
            .fit_transform(
                X_train_stage1,
                y_train_processed
            )
        )


        X_valid_encoded = (
            post_encoder
            .transform(
                X_valid_stage1
            )
        )


        # ----------------------------------------------------
        # ENN if required
        # ----------------------------------------------------

        if strategy == "SMOTENC + ENN":

            enn = (
                make_enn()
            )

            (
                X_train_encoded,
                y_train_processed
            ) = enn.fit_resample(
                X_train_encoded,
                y_train_processed
            )


        # ----------------------------------------------------
        # Model
        # ----------------------------------------------------

        model = make_tunable_model(
            model_name=
                model_name,

            strategy=
                strategy,

            y_training=
                y_train_fold,

            params=
                params
        )


        model.fit(
            X_train_encoded,
            y_train_processed
        )


        probabilities = (
            model.predict_proba(
                X_valid_encoded
            )[:, 1]
        )


        predictions = (
            probabilities
            >= 0.50
        ).astype(int)


        ap = (
            average_precision_score(
                y_valid_fold,
                probabilities
            )
        )


        roc = (
            roc_auc_score(
                y_valid_fold,
                probabilities
            )
        )


        precision = (
            precision_score(
                y_valid_fold,
                predictions,
                zero_division=0
            )
        )


        recall = (
            recall_score(
                y_valid_fold,
                predictions,
                zero_division=0
            )
        )


        f1 = (
            f1_score(
                y_valid_fold,
                predictions,
                zero_division=0
            )
        )


        accuracy = (
            accuracy_score(
                y_valid_fold,
                predictions
            )
        )


        tn, fp, fn, tp = (
            confusion_matrix(
                y_valid_fold,
                predictions,
                labels=[0, 1]
            )
            .ravel()
        )


        fold_results.append({
            "Fold":
                fold_number,

            "AP":
                ap,

            "ROC_AUC":
                roc,

            "Precision":
                precision,

            "Recall":
                recall,

            "F1":
                f1,

            "Accuracy":
                accuracy,

            "FN":
                fn,

            "FP":
                fp
        })


    fold_df = pd.DataFrame(
        fold_results
    )


    return {

        "AP_Mean":
            fold_df["AP"].mean(),

        "AP_STD":
            fold_df["AP"].std(),

        "ROC_AUC_Mean":
            fold_df["ROC_AUC"].mean(),

        "Precision_Mean":
            fold_df["Precision"].mean(),

        "Recall_Mean":
            fold_df["Recall"].mean(),

        "F1_Mean":
            fold_df["F1"].mean(),

        "Accuracy_Mean":
            fold_df["Accuracy"].mean(),

        "FN_Mean":
            fold_df["FN"].mean(),

        "FP_Mean":
            fold_df["FP"].mean()
    }


print(
    "✅ Tuning evaluation function created."
)
```

```python
# ============================================================
# CELL 8: Tuning Smoke Test
# ============================================================

smoke_params = {
    "C": 1.0,
    "solver": "liblinear"
}


smoke_tuning_result = (
    evaluate_tuning_configuration(
        model_name=
            "Logistic Regression",

        strategy=
            "Original",

        params=
            smoke_params
    )
)


print("=" * 75)
print("PHASE 08 SMOKE TEST")
print("=" * 75)


print(
    "Configuration:",
    smoke_params
)


for key, value in (
    smoke_tuning_result.items()
):

    print(
        f"{key:<20}: "
        f"{value:.6f}"
    )
```

```python
# ============================================================
# CELL 9: Run Full Hyperparameter Search
# ============================================================

tuning_results = []


tuning_start_time = (
    time.time()
)


print("=" * 86)
print("STARTING PHASE 08 — HYPERPARAMETER SEARCH")
print("=" * 86)


for candidate_number, candidate in enumerate(
    TUNING_CANDIDATES,
    start=1
):

    candidate_name = (
        candidate["Candidate"]
    )

    model_name = (
        candidate["Model"]
    )

    strategy = (
        candidate["Strategy"]
    )


    n_iter = (
        SEARCH_ITERATIONS[
            model_name
        ]
    )


    parameter_samples = list(
        ParameterSampler(
            SEARCH_SPACES[
                model_name
            ],

            n_iter=
                n_iter,

            random_state=
                SEED
        )
    )


    print(
        f"\n[{candidate_number}/"
        f"{len(TUNING_CANDIDATES)}] "
        f"{candidate_name}"
    )

    print(
        f"Testing "
        f"{len(parameter_samples)} "
        f"configurations..."
    )


    candidate_start_time = (
        time.time()
    )


    for config_number, params in enumerate(
        parameter_samples,
        start=1
    ):

        metrics = (
            evaluate_tuning_configuration(
                model_name=
                    model_name,

                strategy=
                    strategy,

                params=
                    params
            )
        )


        tuning_results.append({

            "Candidate":
                candidate_name,

            "Model":
                model_name,

            "Strategy":
                strategy,

            "Config_Number":
                config_number,

            "Parameters":
                json.dumps(
                    params,
                    default=str,
                    sort_keys=True
                ),

            **metrics
        })


        if (
            config_number == 1
            or
            config_number
            % 5 == 0
            or
            config_number
            ==
            len(parameter_samples)
        ):

            print(
                f"  Config "
                f"{config_number:02d}/"
                f"{len(parameter_samples):02d} | "
                f"AP="
                f"{metrics['AP_Mean']:.5f}"
            )


    candidate_runtime = (
        time.time()
        -
        candidate_start_time
    )


    print(
        f"Completed in "
        f"{candidate_runtime / 60:.2f} min"
    )


tuning_total_time = (
    time.time()
    -
    tuning_start_time
)


tuning_results_df = (
    pd.DataFrame(
        tuning_results
    )
)


print("\n" + "=" * 86)

print(
    "✅ PHASE 08 SEARCH COMPLETED"
)

print(
    f"Configurations evaluated : "
    f"{len(tuning_results_df)}"
)

print(
    f"Total runtime            : "
    f"{tuning_total_time / 60:.2f} min"
)

print("=" * 86)
```

```python
# ============================================================
# CELL 10: Select Best Configuration per Candidate
# ============================================================

best_tuned_candidates = (
    tuning_results_df
    .sort_values(
        [
            "Candidate",
            "AP_Mean"
        ],
        ascending=[
            True,
            False
        ]
    )
    .groupby(
        "Candidate",
        as_index=False
    )
    .first()
)


best_tuned_candidates = (
    best_tuned_candidates
    .sort_values(
        "AP_Mean",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


best_tuned_candidates.insert(
    0,
    "Tuned_Rank",
    np.arange(
        1,
        len(
            best_tuned_candidates
        )
        + 1
    )
)


display(
    best_tuned_candidates[
        [
            "Tuned_Rank",
            "Candidate",
            "Model",
            "Strategy",
            "AP_Mean",
            "AP_STD",
            "ROC_AUC_Mean",
            "Precision_Mean",
            "Recall_Mean",
            "F1_Mean",
            "Accuracy_Mean",
            "FN_Mean",
            "FP_Mean",
            "Parameters"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 11: Baseline vs Tuned Comparison
# ============================================================

baseline_lookup = (
    rq1_summary[
        [
            "Model",
            "Strategy",
            "PR_AUC_AP_Mean",
            "PR_AUC_AP_STD"
        ]
    ]
    .copy()
)


tuned_vs_baseline = (
    best_tuned_candidates
    .merge(
        baseline_lookup,

        on=[
            "Model",
            "Strategy"
        ],

        how="left"
    )
)


tuned_vs_baseline[
    "AP_Improvement"
] = (
    tuned_vs_baseline[
        "AP_Mean"
    ]
    -
    tuned_vs_baseline[
        "PR_AUC_AP_Mean"
    ]
)


display(
    tuned_vs_baseline[
        [
            "Candidate",
            "Model",
            "Strategy",
            "PR_AUC_AP_Mean",
            "AP_Mean",
            "AP_Improvement",
            "Recall_Mean",
            "F1_Mean"
        ]
    ].round(6)
)
```

# PHASE 09 — Leakage-Safe Out-of-Fold Probability Generation

```python
# ============================================================
# PHASE 09 — OUT-OF-FOLD PROBABILITY GENERATION
# CELL 1: Phase Setup
# ============================================================

PHASE09_DIR = OUTPUT_DIRS["06_oof"]

os.makedirs(
    PHASE09_DIR,
    exist_ok=True
)


print("=" * 84)
print("PHASE 09 — LEAKAGE-SAFE OOF PROBABILITY GENERATION")
print("=" * 84)

print(
    f"Development samples : {len(X_dev):,}"
)

print(
    f"Locked test samples : {len(X_test):,}"
)

print(
    "Tuned candidates    :",
    len(best_tuned_candidates)
)

print(
    "CV folds            :",
    len(cv_splits)
)

print(
    "Final test locked   :",
    FINAL_TEST_LOCKED
)
```

```python
# ============================================================
# CELL 2: Prepare Tuned Candidate Specifications
# ============================================================

tuned_candidate_specs = []


for _, row in (
    best_tuned_candidates
    .sort_values(
        "Tuned_Rank"
    )
    .iterrows()
):

    tuned_candidate_specs.append({

        "Candidate":
            row["Candidate"],

        "Model":
            row["Model"],

        "Strategy":
            row["Strategy"],

        "Parameters":
            json.loads(
                row["Parameters"]
            )
    })


print("=" * 84)
print("TUNED CANDIDATES FOR OOF GENERATION")
print("=" * 84)


for spec in tuned_candidate_specs:

    print(
        f"\n{spec['Candidate']}"
    )

    print(
        " Model    :",
        spec["Model"]
    )

    print(
        " Strategy :",
        spec["Strategy"]
    )

    print(
        " Params   :",
        spec["Parameters"]
    )


print(
    "\nTotal candidates:",
    len(tuned_candidate_specs)
)
```

```python
# ============================================================
# CELL 3: Create OOF Fold Assignment
# ============================================================

oof_fold_assignment = np.full(
    len(X_dev),
    fill_value=-1,
    dtype=int
)


for fold_number, (
    train_idx,
    valid_idx
) in enumerate(
    cv_splits,
    start=1
):

    oof_fold_assignment[
        valid_idx
    ] = fold_number


print("=" * 75)
print("OOF FOLD ASSIGNMENT")
print("=" * 75)


for fold_number in range(
    1,
    len(cv_splits) + 1
):

    fold_count = int(
        (
            oof_fold_assignment
            ==
            fold_number
        ).sum()
    )

    print(
        f"Fold {fold_number}: "
        f"{fold_count:,} customers"
    )


print(
    "\nUnassigned customers:",
    int(
        (
            oof_fold_assignment
            ==
            -1
        ).sum()
    )
)
```

```python
# ============================================================
# CELL 4: OOF Probability Generation Function
# ============================================================

def generate_candidate_oof(
    candidate_name,
    model_name,
    strategy,
    parameters
):

    # --------------------------------------------------------
    # Storage
    # --------------------------------------------------------
    oof_probabilities = np.full(
        len(X_dev),
        np.nan,
        dtype=float
    )


    fold_metric_rows = []


    # --------------------------------------------------------
    # CV Loop
    # --------------------------------------------------------
    for fold_number, (
        train_idx,
        valid_idx
    ) in enumerate(
        cv_splits,
        start=1
    ):

        X_train_fold = (
            X_dev
            .iloc[train_idx]
            .copy()
        )

        y_train_fold = (
            y_dev
            .iloc[train_idx]
            .copy()
        )

        X_valid_fold = (
            X_dev
            .iloc[valid_idx]
            .copy()
        )

        y_valid_fold = (
            y_dev
            .iloc[valid_idx]
            .copy()
        )


        # ====================================================
        # 1. Stage-1 Preprocessing
        # ====================================================

        pre_sampler = (
            make_pre_sampler_transformer()
        )


        X_train_stage1 = (
            pre_sampler
            .fit_transform(
                X_train_fold,
                y_train_fold
            )
        )


        X_valid_stage1 = (
            pre_sampler
            .transform(
                X_valid_fold
            )
        )


        # ====================================================
        # 2. SMOTENC if required
        # ====================================================

        if strategy in [
            "SMOTENC",
            "SMOTENC + ENN"
        ]:

            smote = (
                make_smotenc()
            )


            (
                X_train_stage1,
                y_train_processed
            ) = smote.fit_resample(
                X_train_stage1,
                y_train_fold
            )

        else:

            y_train_processed = (
                y_train_fold
                .to_numpy()
            )


        # ====================================================
        # 3. One-Hot Encoding
        # ====================================================

        post_encoder = (
            make_post_sampler_encoder()
        )


        X_train_encoded = (
            post_encoder
            .fit_transform(
                X_train_stage1,
                y_train_processed
            )
        )


        X_valid_encoded = (
            post_encoder
            .transform(
                X_valid_stage1
            )
        )


        # ====================================================
        # 4. ENN if required
        # ====================================================

        if strategy == "SMOTENC + ENN":

            enn = (
                make_enn()
            )


            (
                X_train_encoded,
                y_train_processed
            ) = enn.fit_resample(
                X_train_encoded,
                y_train_processed
            )


        # ====================================================
        # 5. Tuned Model
        # ====================================================

        model = make_tunable_model(

            model_name=
                model_name,

            strategy=
                strategy,

            y_training=
                y_train_fold,

            params=
                parameters
        )


        model.fit(
            X_train_encoded,
            y_train_processed
        )


        # ====================================================
        # 6. Predict Validation Fold
        # ====================================================

        fold_probabilities = (
            model.predict_proba(
                X_valid_encoded
            )[:, 1]
        )


        # Store probabilities in original OOF positions
        oof_probabilities[
            valid_idx
        ] = fold_probabilities


        # ====================================================
        # 7. Fold Diagnostics
        # ====================================================

        fold_predictions = (
            fold_probabilities
            >= 0.50
        ).astype(int)


        tn, fp, fn, tp = (
            confusion_matrix(
                y_valid_fold,
                fold_predictions,
                labels=[0, 1]
            )
            .ravel()
        )


        fold_metric_rows.append({

            "Candidate":
                candidate_name,

            "Model":
                model_name,

            "Strategy":
                strategy,

            "Fold":
                fold_number,

            "Validation_Rows":
                len(valid_idx),

            "AP":
                average_precision_score(
                    y_valid_fold,
                    fold_probabilities
                ),

            "ROC_AUC":
                roc_auc_score(
                    y_valid_fold,
                    fold_probabilities
                ),

            "Precision":
                precision_score(
                    y_valid_fold,
                    fold_predictions,
                    zero_division=0
                ),

            "Recall":
                recall_score(
                    y_valid_fold,
                    fold_predictions,
                    zero_division=0
                ),

            "F1":
                f1_score(
                    y_valid_fold,
                    fold_predictions,
                    zero_division=0
                ),

            "Accuracy":
                accuracy_score(
                    y_valid_fold,
                    fold_predictions
                ),

            "TN":
                int(tn),

            "FP":
                int(fp),

            "FN":
                int(fn),

            "TP":
                int(tp)
        })


    # --------------------------------------------------------
    # Integrity
    # --------------------------------------------------------

    if np.isnan(
        oof_probabilities
    ).any():

        raise ValueError(
            f"OOF probabilities contain NaN "
            f"for {candidate_name}"
        )


    return (
        oof_probabilities,
        pd.DataFrame(
            fold_metric_rows
        )
    )


print(
    "✅ OOF generation function created."
)
```

```python
# ============================================================
# CELL 5: OOF Smoke Test
# ============================================================

smoke_spec = (
    tuned_candidate_specs[0]
)


smoke_oof_probabilities, smoke_fold_metrics = (
    generate_candidate_oof(

        candidate_name=
            smoke_spec[
                "Candidate"
            ],

        model_name=
            smoke_spec[
                "Model"
            ],

        strategy=
            smoke_spec[
                "Strategy"
            ],

        parameters=
            smoke_spec[
                "Parameters"
            ]
    )
)


print("=" * 80)
print("PHASE 09 OOF SMOKE TEST")
print("=" * 80)


print(
    "Candidate:",
    smoke_spec[
        "Candidate"
    ]
)


print(
    "OOF probabilities:",
    len(
        smoke_oof_probabilities
    )
)


print(
    "Missing probabilities:",
    int(
        np.isnan(
            smoke_oof_probabilities
        ).sum()
    )
)


print(
    "Minimum probability:",
    round(
        float(
            smoke_oof_probabilities.min()
        ),
        6
    )
)


print(
    "Maximum probability:",
    round(
        float(
            smoke_oof_probabilities.max()
        ),
        6
    )
)


print(
    "\nPooled OOF AP:",
    round(
        average_precision_score(
            y_dev,
            smoke_oof_probabilities
        ),
        6
    )
)


display(
    smoke_fold_metrics
)
```

```python
# ============================================================
# CELL 6: Generate All Tuned OOF Probabilities
# ============================================================

oof_probability_dictionary = {}

all_oof_fold_metrics = []


oof_start_time = (
    time.time()
)


print("=" * 88)
print("GENERATING TUNED OOF PROBABILITIES")
print("=" * 88)


for candidate_number, spec in enumerate(
    tuned_candidate_specs,
    start=1
):

    candidate_name = (
        spec["Candidate"]
    )


    print(
        f"\n[{candidate_number}/"
        f"{len(tuned_candidate_specs)}] "
        f"{candidate_name}"
    )


    candidate_start = (
        time.time()
    )


    probabilities, fold_metrics = (
        generate_candidate_oof(

            candidate_name=
                candidate_name,

            model_name=
                spec["Model"],

            strategy=
                spec["Strategy"],

            parameters=
                spec["Parameters"]
        )
    )


    oof_probability_dictionary[
        candidate_name
    ] = probabilities


    all_oof_fold_metrics.append(
        fold_metrics
    )


    candidate_ap = (
        average_precision_score(
            y_dev,
            probabilities
        )
    )


    candidate_roc = (
        roc_auc_score(
            y_dev,
            probabilities
        )
    )


    elapsed = (
        time.time()
        -
        candidate_start
    )


    print(
        f"   Pooled OOF AP      : "
        f"{candidate_ap:.6f}"
    )

    print(
        f"   Pooled OOF ROC-AUC : "
        f"{candidate_roc:.6f}"
    )

    print(
        f"   Runtime            : "
        f"{elapsed:.2f} sec"
    )


oof_total_time = (
    time.time()
    -
    oof_start_time
)


all_oof_fold_metrics_df = (
    pd.concat(
        all_oof_fold_metrics,
        ignore_index=True
    )
)


print("\n" + "=" * 88)

print(
    "✅ ALL TUNED OOF PROBABILITIES GENERATED"
)

print(
    f"Candidates : "
    f"{len(oof_probability_dictionary)}"
)

print(
    f"Runtime    : "
    f"{oof_total_time / 60:.2f} min"
)

print("=" * 88)
```

```python
# ============================================================
# CELL 7: Build Master OOF Prediction Table
# ============================================================

oof_predictions_df = pd.DataFrame({

    "source_index":
        X_dev.index.to_numpy(),

    "True_Churn":
        y_dev.to_numpy(),

    "OOF_Fold":
        oof_fold_assignment
})


for candidate_name, probabilities in (
    oof_probability_dictionary.items()
):

    oof_predictions_df[
        candidate_name
    ] = probabilities


print(
    "OOF prediction table shape:",
    oof_predictions_df.shape
)


display(
    oof_predictions_df.head(10)
)
```

```python
# ============================================================
# CELL 8: Pooled OOF Performance Summary
# ============================================================

oof_summary_rows = []


for spec in tuned_candidate_specs:

    candidate_name = (
        spec["Candidate"]
    )

    probabilities = (
        oof_probability_dictionary[
            candidate_name
        ]
    )


    predictions = (
        probabilities >= 0.50
    ).astype(int)


    tn, fp, fn, tp = (
        confusion_matrix(
            y_dev,
            predictions,
            labels=[0, 1]
        )
        .ravel()
    )


    oof_summary_rows.append({

        "Candidate":
            candidate_name,

        "Model":
            spec["Model"],

        "Strategy":
            spec["Strategy"],

        "OOF_AP":
            average_precision_score(
                y_dev,
                probabilities
            ),

        "OOF_ROC_AUC":
            roc_auc_score(
                y_dev,
                probabilities
            ),

        "Accuracy_050":
            accuracy_score(
                y_dev,
                predictions
            ),

        "Precision_050":
            precision_score(
                y_dev,
                predictions,
                zero_division=0
            ),

        "Recall_050":
            recall_score(
                y_dev,
                predictions,
                zero_division=0
            ),

        "F1_050":
            f1_score(
                y_dev,
                predictions,
                zero_division=0
            ),

        "TN_050":
            int(tn),

        "FP_050":
            int(fp),

        "FN_050":
            int(fn),

        "TP_050":
            int(tp)
    })


oof_summary = (
    pd.DataFrame(
        oof_summary_rows
    )
    .sort_values(
        "OOF_AP",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


oof_summary.insert(
    0,
    "OOF_AP_Rank",
    np.arange(
        1,
        len(
            oof_summary
        ) + 1
    )
)


display(
    oof_summary.round(6)
)
```

```python
# ============================================================
# CELL 9: Phase 08 vs Phase 09 AP Consistency
# ============================================================

phase08_ap_reference = (
    best_tuned_candidates[
        [
            "Candidate",
            "AP_Mean",
            "AP_STD"
        ]
    ]
    .copy()
)


oof_consistency_table = (
    oof_summary[
        [
            "Candidate",
            "OOF_AP",
            "OOF_ROC_AUC"
        ]
    ]
    .merge(
        phase08_ap_reference,
        on="Candidate",
        how="left"
    )
)


oof_consistency_table[
    "OOF_minus_MeanFold_AP"
] = (
    oof_consistency_table[
        "OOF_AP"
    ]
    -
    oof_consistency_table[
        "AP_Mean"
    ]
)


oof_consistency_table = (
    oof_consistency_table
    .sort_values(
        "OOF_AP",
        ascending=False
    )
)


display(
    oof_consistency_table.round(6)
)
```

```python
# ============================================================
# CELL 10: OOF Probability Integrity Audit
# ============================================================

probability_audit_rows = []


for candidate_name, probabilities in (
    oof_probability_dictionary.items()
):

    probability_audit_rows.append({

        "Candidate":
            candidate_name,

        "Count":
            len(
                probabilities
            ),

        "Missing":
            int(
                np.isnan(
                    probabilities
                ).sum()
            ),

        "Below_Zero":
            int(
                (
                    probabilities < 0
                ).sum()
            ),

        "Above_One":
            int(
                (
                    probabilities > 1
                ).sum()
            ),

        "Minimum":
            float(
                probabilities.min()
            ),

        "Maximum":
            float(
                probabilities.max()
            ),

        "Mean":
            float(
                probabilities.mean()
            ),

        "Median":
            float(
                np.median(
                    probabilities
                )
            )
    })


probability_audit = pd.DataFrame(
    probability_audit_rows
)


display(
    probability_audit.round(6)
)
```

# PHASE 10 — RQ2: Decision Threshold Analysis

```python
# ============================================================
# PHASE 10 — RQ2 THRESHOLD ANALYSIS
# CELL 1: Phase Setup
# ============================================================

PHASE10_DIR = OUTPUT_DIRS["07_threshold"]

os.makedirs(
    PHASE10_DIR,
    exist_ok=True
)


THRESHOLDS = np.round(
    np.arange(
        0.10,
        0.91,
        0.01
    ),
    2
)


print("=" * 86)
print("PHASE 10 — RQ2 DECISION THRESHOLD ANALYSIS")
print("=" * 86)

print(
    f"Development OOF samples : {len(y_dev):,}"
)

print(
    f"Candidates              : "
    f"{len(tuned_candidate_specs)}"
)

print(
    f"Thresholds              : "
    f"{len(THRESHOLDS)}"
)

print(
    f"Range                   : "
    f"{THRESHOLDS.min():.2f} "
    f"to "
    f"{THRESHOLDS.max():.2f}"
)

print(
    f"Total evaluations       : "
    f"{len(tuned_candidate_specs) * len(THRESHOLDS)}"
)

print(
    "\nFinal test locked       :",
    FINAL_TEST_LOCKED
)
```

```python
# ============================================================
# CELL 2: Threshold Evaluation Function
# ============================================================

def evaluate_threshold(
    y_true,
    probabilities,
    threshold
):

    predictions = (
        probabilities >= threshold
    ).astype(int)


    tn, fp, fn, tp = (
        confusion_matrix(
            y_true,
            predictions,
            labels=[0, 1]
        )
        .ravel()
    )


    accuracy = accuracy_score(
        y_true,
        predictions
    )


    precision = precision_score(
        y_true,
        predictions,
        zero_division=0
    )


    recall = recall_score(
        y_true,
        predictions,
        zero_division=0
    )


    f1 = f1_score(
        y_true,
        predictions,
        zero_division=0
    )


    specificity = (
        tn / (tn + fp)
        if (tn + fp) > 0
        else 0.0
    )


    predicted_positive_rate = (
        predictions.mean()
    )


    return {

        "Threshold":
            float(threshold),

        "Accuracy":
            float(accuracy),

        "Precision":
            float(precision),

        "Recall":
            float(recall),

        "Specificity":
            float(specificity),

        "F1":
            float(f1),

        "TN":
            int(tn),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TP":
            int(tp),

        "Predicted_Churn_Count":
            int(
                predictions.sum()
            ),

        "Predicted_Churn_Rate":
            float(
                predicted_positive_rate
            )
    }


print(
    "✅ Threshold evaluation function created."
)
```

```python
# ============================================================
# CELL 3: Threshold Smoke Test
# ============================================================

smoke_candidate = (
    "XGBoost_Original"
)


smoke_probabilities = (
    oof_probability_dictionary[
        smoke_candidate
    ]
)


smoke_threshold_result = (
    evaluate_threshold(
        y_true=
            y_dev,

        probabilities=
            smoke_probabilities,

        threshold=
            0.50
    )
)


print("=" * 78)
print("THRESHOLD SMOKE TEST")
print("=" * 78)

print(
    "Candidate:",
    smoke_candidate
)


for key, value in (
    smoke_threshold_result.items()
):

    print(
        f"{key:<24}: {value}"
    )
```

```python
# ============================================================
# CELL 4: Full Threshold Scan
# ============================================================

threshold_results = []


print("=" * 88)
print("RUNNING FULL THRESHOLD ANALYSIS")
print("=" * 88)


for candidate_number, spec in enumerate(
    tuned_candidate_specs,
    start=1
):

    candidate_name = (
        spec["Candidate"]
    )


    probabilities = (
        oof_probability_dictionary[
            candidate_name
        ]
    )


    print(
        f"[{candidate_number}/"
        f"{len(tuned_candidate_specs)}] "
        f"{candidate_name}"
    )


    for threshold in THRESHOLDS:

        metrics = (
            evaluate_threshold(
                y_true=
                    y_dev,

                probabilities=
                    probabilities,

                threshold=
                    threshold
            )
        )


        threshold_results.append({

            "Candidate":
                candidate_name,

            "Model":
                spec["Model"],

            "Strategy":
                spec["Strategy"],

            **metrics
        })


threshold_results_df = pd.DataFrame(
    threshold_results
)


print("\n✅ Threshold scan completed.")

print(
    "Total rows:",
    len(
        threshold_results_df
    )
)
```

```python
# ============================================================
# CELL 5: Inspect Threshold Results
# ============================================================

print(
    "Threshold result shape:",
    threshold_results_df.shape
)


display(
    threshold_results_df.head(20)
)
```

```python
# ============================================================
# CELL 6: Default Threshold Results
# ============================================================

default_threshold_results = (
    threshold_results_df[
        np.isclose(
            threshold_results_df[
                "Threshold"
            ],
            0.50
        )
    ]
    .copy()
)


default_threshold_results = (
    default_threshold_results
    .sort_values(
        "F1",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


display(
    default_threshold_results[
        [
            "Candidate",
            "Threshold",
            "Accuracy",
            "Precision",
            "Recall",
            "Specificity",
            "F1",
            "FN",
            "FP",
            "Predicted_Churn_Count"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 7: Best F1 Threshold per Candidate
# ============================================================

best_f1_thresholds = (
    threshold_results_df
    .sort_values(
        [
            "Candidate",
            "F1",
            "Recall",
            "Precision"
        ],
        ascending=[
            True,
            False,
            False,
            False
        ]
    )
    .groupby(
        "Candidate",
        as_index=False
    )
    .first()
)


best_f1_thresholds = (
    best_f1_thresholds[
        [
            "Candidate",
            "Model",
            "Strategy",
            "Threshold",
            "Accuracy",
            "Precision",
            "Recall",
            "Specificity",
            "F1",
            "TN",
            "FP",
            "FN",
            "TP",
            "Predicted_Churn_Count",
            "Predicted_Churn_Rate"
        ]
    ]
)


best_f1_thresholds = (
    best_f1_thresholds
    .sort_values(
        "F1",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


display(
    best_f1_thresholds.round(6)
)
```

```python
# ============================================================
# CELL 8: Default vs Best-F1 Threshold Comparison
# ============================================================

default_comparison = (
    default_threshold_results[
        [
            "Candidate",
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "FN",
            "FP",
            "Predicted_Churn_Count"
        ]
    ]
    .rename(
        columns={
            "Threshold":
                "Default_Threshold",

            "Precision":
                "Default_Precision",

            "Recall":
                "Default_Recall",

            "F1":
                "Default_F1",

            "FN":
                "Default_FN",

            "FP":
                "Default_FP",

            "Predicted_Churn_Count":
                "Default_Predicted_Churn"
        }
    )
)


f1_comparison = (
    best_f1_thresholds[
        [
            "Candidate",
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "FN",
            "FP",
            "Predicted_Churn_Count"
        ]
    ]
    .rename(
        columns={
            "Threshold":
                "Best_F1_Threshold",

            "Precision":
                "Best_F1_Precision",

            "Recall":
                "Best_F1_Recall",

            "F1":
                "Best_F1",

            "FN":
                "Best_F1_FN",

            "FP":
                "Best_F1_FP",

            "Predicted_Churn_Count":
                "Best_F1_Predicted_Churn"
        }
    )
)


default_vs_f1 = (
    default_comparison
    .merge(
        f1_comparison,
        on="Candidate",
        how="inner"
    )
)


default_vs_f1[
    "FN_Change"
] = (
    default_vs_f1[
        "Best_F1_FN"
    ]
    -
    default_vs_f1[
        "Default_FN"
    ]
)


default_vs_f1[
    "FP_Change"
] = (
    default_vs_f1[
        "Best_F1_FP"
    ]
    -
    default_vs_f1[
        "Default_FP"
    ]
)


default_vs_f1[
    "F1_Change"
] = (
    default_vs_f1[
        "Best_F1"
    ]
    -
    default_vs_f1[
        "Default_F1"
    ]
)


display(
    default_vs_f1.round(6)
)
```

```python
# ============================================================
# CELL 9: Threshold Near 80% Recall
# ============================================================

TARGET_RECALL = 0.80


recall80_rows = []


for candidate_name in (
    threshold_results_df[
        "Candidate"
    ].unique()
):

    candidate_data = (
        threshold_results_df[
            threshold_results_df[
                "Candidate"
            ]
            ==
            candidate_name
        ]
        .copy()
    )


    candidate_data[
        "Recall_Distance"
    ] = (
        candidate_data[
            "Recall"
        ]
        -
        TARGET_RECALL
    ).abs()


    nearest_row = (
        candidate_data
        .sort_values(
            [
                "Recall_Distance",
                "Precision"
            ],
            ascending=[
                True,
                False
            ]
        )
        .iloc[0]
    )


    recall80_rows.append(
        nearest_row
    )


recall80_thresholds = pd.DataFrame(
    recall80_rows
)


recall80_thresholds = (
    recall80_thresholds[
        [
            "Candidate",
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "FN",
            "FP",
            "Predicted_Churn_Count"
        ]
    ]
)


display(
    recall80_thresholds.round(6)
)
```

# PHASE 11 — Scenario-Based Retention Decision Cost Analysis

```python
# ============================================================
# PHASE 11 — SCENARIO-BASED RETENTION DECISION COST
# CELL 1: Phase Setup
# ============================================================

PHASE11_DIR = OUTPUT_DIRS["08_utility"]

os.makedirs(
    PHASE11_DIR,
    exist_ok=True
)


COST_SCENARIOS = {
    "FN2_FP1": 2,
    "FN5_FP1": 5,
    "FN10_FP1": 10
}


N_DECISION = len(y_dev)


print("=" * 88)
print("PHASE 11 — SCENARIO-BASED RETENTION DECISION COST")
print("=" * 88)

print(
    f"OOF development customers : {N_DECISION:,}"
)

print(
    f"Candidates                : "
    f"{threshold_results_df['Candidate'].nunique()}"
)

print(
    f"Thresholds / candidate    : "
    f"{threshold_results_df['Threshold'].nunique()}"
)

print(
    f"Decision scenarios        : "
    f"{len(COST_SCENARIOS)}"
)


print("\nCost scenarios:")

for scenario, ratio in COST_SCENARIOS.items():

    print(
        f" - {scenario}: "
        f"FN cost = {ratio} × FP cost"
    )


print(
    "\nFinal test locked:",
    FINAL_TEST_LOCKED
)
```

```python
# ============================================================
# CELL 2: Calculate Decision Cost for Every Threshold
# ============================================================

decision_cost_rows = []


for _, row in threshold_results_df.iterrows():

    for scenario_name, fn_fp_ratio in (
        COST_SCENARIOS.items()
    ):

        weighted_cost = (
            fn_fp_ratio
            *
            int(row["FN"])
            +
            int(row["FP"])
        )


        normalized_cost = (
            weighted_cost
            /
            N_DECISION
        )


        cost_per_1000 = (
            normalized_cost
            *
            1000
        )


        decision_cost_rows.append({

            "Candidate":
                row["Candidate"],

            "Model":
                row["Model"],

            "Strategy":
                row["Strategy"],

            "Threshold":
                float(
                    row["Threshold"]
                ),

            "Scenario":
                scenario_name,

            "FN_FP_Ratio":
                fn_fp_ratio,

            "Precision":
                float(
                    row["Precision"]
                ),

            "Recall":
                float(
                    row["Recall"]
                ),

            "F1":
                float(
                    row["F1"]
                ),

            "Specificity":
                float(
                    row["Specificity"]
                ),

            "FN":
                int(
                    row["FN"]
                ),

            "FP":
                int(
                    row["FP"]
                ),

            "TP":
                int(
                    row["TP"]
                ),

            "TN":
                int(
                    row["TN"]
                ),

            "Weighted_Decision_Cost":
                int(
                    weighted_cost
                ),

            "Normalized_Decision_Cost":
                float(
                    normalized_cost
                ),

            "Cost_per_1000_Customers":
                float(
                    cost_per_1000
                )
        })


decision_cost_df = pd.DataFrame(
    decision_cost_rows
)


print(
    "Decision-cost table shape:",
    decision_cost_df.shape
)


print(
    "Total evaluations:",
    len(
        decision_cost_df
    )
)
```

```python
# ============================================================
# CELL 3: Inspect Decision-Cost Results
# ============================================================

display(
    decision_cost_df.head(20)
)


print(
    "\nNormalized cost range:"
)

print(
    "Minimum:",
    round(
        decision_cost_df[
            "Normalized_Decision_Cost"
        ].min(),
        6
    )
)

print(
    "Maximum:",
    round(
        decision_cost_df[
            "Normalized_Decision_Cost"
        ].max(),
        6
    )
)
```

```python
# ============================================================
# CELL 4: Default Threshold Decision Costs
# ============================================================

default_cost_results = (
    decision_cost_df[
        np.isclose(
            decision_cost_df[
                "Threshold"
            ],
            0.50
        )
    ]
    .copy()
)


default_cost_results = (
    default_cost_results
    .sort_values(
        [
            "Scenario",
            "Normalized_Decision_Cost"
        ]
    )
    .reset_index(
        drop=True
    )
)


display(
    default_cost_results[
        [
            "Scenario",
            "Candidate",
            "Threshold",
            "Recall",
            "Precision",
            "FN",
            "FP",
            "Weighted_Decision_Cost",
            "Normalized_Decision_Cost"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 5: Best Cost Threshold per Candidate and Scenario
# ============================================================

best_cost_thresholds = (
    decision_cost_df
    .sort_values(
        [
            "Candidate",
            "Scenario",
            "Normalized_Decision_Cost",
            "F1",
            "Recall",
            "Precision"
        ],
        ascending=[
            True,
            True,
            True,
            False,
            False,
            False
        ]
    )
    .groupby(
        [
            "Candidate",
            "Scenario"
        ],
        as_index=False
    )
    .first()
)


best_cost_thresholds = (
    best_cost_thresholds[
        [
            "Candidate",
            "Model",
            "Strategy",
            "Scenario",
            "FN_FP_Ratio",
            "Threshold",
            "Precision",
            "Recall",
            "F1",
            "Specificity",
            "FN",
            "FP",
            "TP",
            "TN",
            "Weighted_Decision_Cost",
            "Normalized_Decision_Cost",
            "Cost_per_1000_Customers"
        ]
    ]
)


display(
    best_cost_thresholds.round(6)
)
```

```python
# ============================================================
# CELL 6: Scenario-Wise Candidate Ranking
# ============================================================

scenario_ranking_parts = []


for scenario_name in COST_SCENARIOS.keys():

    scenario_table = (
        best_cost_thresholds[
            best_cost_thresholds[
                "Scenario"
            ]
            ==
            scenario_name
        ]
        .copy()
    )


    # Add OOF AP as secondary reference
    scenario_table = (
        scenario_table
        .merge(
            oof_summary[
                [
                    "Candidate",
                    "OOF_AP",
                    "OOF_ROC_AUC"
                ]
            ],

            on="Candidate",

            how="left"
        )
    )


    scenario_table = (
        scenario_table
        .sort_values(
            [
                "Normalized_Decision_Cost",
                "OOF_AP",
                "F1"
            ],
            ascending=[
                True,
                False,
                False
            ]
        )
        .reset_index(
            drop=True
        )
    )


    scenario_table[
        "Scenario_Rank"
    ] = np.arange(
        1,
        len(
            scenario_table
        ) + 1
    )


    scenario_ranking_parts.append(
        scenario_table
    )


scenario_rankings = pd.concat(
    scenario_ranking_parts,
    ignore_index=True
)


display(
    scenario_rankings[
        [
            "Scenario",
            "Scenario_Rank",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost",
            "FN",
            "FP",
            "Recall",
            "Precision",
            "F1",
            "OOF_AP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 7: Best Candidate in Each Cost Scenario
# ============================================================

scenario_winners = (
    scenario_rankings[
        scenario_rankings[
            "Scenario_Rank"
        ]
        ==
        1
    ]
    .copy()
)


print("=" * 88)
print("BEST MODEL + THRESHOLD BY DECISION-COST SCENARIO")
print("=" * 88)


for _, row in (
    scenario_winners
    .sort_values(
        "FN_FP_Ratio"
    )
    .iterrows()
):

    print(
        f"\nScenario "
        f"FN:FP = "
        f"{int(row['FN_FP_Ratio'])}:1"
    )

    print(
        " Candidate :",
        row[
            "Candidate"
        ]
    )

    print(
        f" Threshold : "
        f"{row['Threshold']:.2f}"
    )

    print(
        f" Cost      : "
        f"{row['Normalized_Decision_Cost']:.6f}"
    )

    print(
        f" Recall    : "
        f"{row['Recall']:.4f}"
    )

    print(
        f" Precision : "
        f"{row['Precision']:.4f}"
    )

    print(
        f" FN / FP   : "
        f"{int(row['FN'])} / "
        f"{int(row['FP'])}"
    )
```

```python
# ============================================================
# CELL 8: Robust Cross-Scenario Ranking
# ============================================================

robust_rank_summary = (
    scenario_rankings
    .groupby(
        [
            "Candidate",
            "Model",
            "Strategy"
        ]
    )
    .agg(

        Mean_Scenario_Rank=(
            "Scenario_Rank",
            "mean"
        ),

        Best_Scenario_Rank=(
            "Scenario_Rank",
            "min"
        ),

        Worst_Scenario_Rank=(
            "Scenario_Rank",
            "max"
        ),

        Scenario_Wins=(
            "Scenario_Rank",
            lambda x:
                int(
                    (x == 1).sum()
                )
        ),

        Mean_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "mean"
        ),

        Worst_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "max"
        ),

        Mean_Recall=(
            "Recall",
            "mean"
        ),

        Mean_Precision=(
            "Precision",
            "mean"
        ),

        Mean_F1=(
            "F1",
            "mean"
        ),

        OOF_AP=(
            "OOF_AP",
            "first"
        )
    )
    .reset_index()
)


robust_rank_summary = (
    robust_rank_summary
    .sort_values(
        [
            "Mean_Scenario_Rank",
            "Worst_Scenario_Rank",
            "Mean_Normalized_Cost",
            "OOF_AP"
        ],
        ascending=[
            True,
            True,
            True,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)


robust_rank_summary.insert(
    0,
    "Robust_Rank",
    np.arange(
        1,
        len(
            robust_rank_summary
        ) + 1
    )
)


display(
    robust_rank_summary.round(6)
)
```

```python
# ============================================================
# CELL 9: Provisional Robust Candidate
# ============================================================

provisional_robust_candidate = (
    robust_rank_summary
    .iloc[0]
)


print("=" * 84)
print("PROVISIONAL CROSS-SCENARIO ROBUST CANDIDATE")
print("=" * 84)


print(
    "Candidate :",
    provisional_robust_candidate[
        "Candidate"
    ]
)

print(
    "Model     :",
    provisional_robust_candidate[
        "Model"
    ]
)

print(
    "Strategy  :",
    provisional_robust_candidate[
        "Strategy"
    ]
)

print(
    f"Mean Rank : "
    f"{provisional_robust_candidate['Mean_Scenario_Rank']:.3f}"
)

print(
    f"Worst Rank: "
    f"{int(provisional_robust_candidate['Worst_Scenario_Rank'])}"
)

print(
    f"Scenario Wins: "
    f"{int(provisional_robust_candidate['Scenario_Wins'])}"
)

print(
    f"OOF AP    : "
    f"{provisional_robust_candidate['OOF_AP']:.6f}"
)
```

```python
# ============================================================
# CELL 10: Default vs Cost-Optimized Threshold
# ============================================================

default_cost_for_merge = (
    default_cost_results[
        [
            "Candidate",
            "Scenario",
            "Threshold",
            "FN",
            "FP",
            "Recall",
            "Precision",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Threshold":
                "Default_Threshold",

            "FN":
                "Default_FN",

            "FP":
                "Default_FP",

            "Recall":
                "Default_Recall",

            "Precision":
                "Default_Precision",

            "Normalized_Decision_Cost":
                "Default_Normalized_Cost"
        }
    )
)


optimized_cost_for_merge = (
    best_cost_thresholds[
        [
            "Candidate",
            "Scenario",
            "Threshold",
            "FN",
            "FP",
            "Recall",
            "Precision",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Threshold":
                "Optimized_Threshold",

            "FN":
                "Optimized_FN",

            "FP":
                "Optimized_FP",

            "Recall":
                "Optimized_Recall",

            "Precision":
                "Optimized_Precision",

            "Normalized_Decision_Cost":
                "Optimized_Normalized_Cost"
        }
    )
)


default_vs_cost_optimized = (
    default_cost_for_merge
    .merge(
        optimized_cost_for_merge,

        on=[
            "Candidate",
            "Scenario"
        ],

        how="inner"
    )
)


default_vs_cost_optimized[
    "Cost_Reduction"
] = (
    default_vs_cost_optimized[
        "Default_Normalized_Cost"
    ]
    -
    default_vs_cost_optimized[
        "Optimized_Normalized_Cost"
    ]
)


default_vs_cost_optimized[
    "Cost_Reduction_Percentage"
] = (

    default_vs_cost_optimized[
        "Cost_Reduction"
    ]

    /

    default_vs_cost_optimized[
        "Default_Normalized_Cost"
    ]

    *

    100
)


default_vs_cost_optimized[
    "FN_Change"
] = (
    default_vs_cost_optimized[
        "Optimized_FN"
    ]
    -
    default_vs_cost_optimized[
        "Default_FN"
    ]
)


default_vs_cost_optimized[
    "FP_Change"
] = (
    default_vs_cost_optimized[
        "Optimized_FP"
    ]
    -
    default_vs_cost_optimized[
        "Default_FP"
    ]
)


display(
    default_vs_cost_optimized.round(6)
)
```

```python
# ============================================================
# CELL 11: Robust Candidate Across Cost Scenarios
# ============================================================

robust_candidate_name = (
    provisional_robust_candidate[
        "Candidate"
    ]
)


robust_candidate_scenarios = (
    scenario_rankings[
        scenario_rankings[
            "Candidate"
        ]
        ==
        robust_candidate_name
    ]
    .sort_values(
        "FN_FP_Ratio"
    )
    .copy()
)


display(
    robust_candidate_scenarios[
        [
            "Scenario",
            "FN_FP_Ratio",
            "Scenario_Rank",
            "Threshold",
            "Normalized_Decision_Cost",
            "Recall",
            "Precision",
            "F1",
            "FN",
            "FP",
            "OOF_AP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 12: Optimal Threshold Boundary Audit
# ============================================================

THRESHOLD_MIN = float(
    THRESHOLDS.min()
)

THRESHOLD_MAX = float(
    THRESHOLDS.max()
)


best_cost_thresholds[
    "At_Grid_Boundary"
] = (

    np.isclose(
        best_cost_thresholds[
            "Threshold"
        ],
        THRESHOLD_MIN
    )

    |

    np.isclose(
        best_cost_thresholds[
            "Threshold"
        ],
        THRESHOLD_MAX
    )
)


scenario_winner_boundary = (
    scenario_winners[
        "Threshold"
    ].apply(
        lambda x:
            np.isclose(
                x,
                THRESHOLD_MIN
            )
            or
            np.isclose(
                x,
                THRESHOLD_MAX
            )
    )
)


robust_candidate_boundary = (
    robust_candidate_scenarios[
        "Threshold"
    ].apply(
        lambda x:
            np.isclose(
                x,
                THRESHOLD_MIN
            )
            or
            np.isclose(
                x,
                THRESHOLD_MAX
            )
    )
)


boundary_issue = bool(
    scenario_winner_boundary.any()
    or
    robust_candidate_boundary.any()
)


print("=" * 84)
print("OPTIMAL THRESHOLD BOUNDARY AUDIT")
print("=" * 84)


print(
    f"Threshold search range: "
    f"{THRESHOLD_MIN:.2f} "
    f"to "
    f"{THRESHOLD_MAX:.2f}"
)


print(
    "\nScenario winner at boundary:",
    bool(
        scenario_winner_boundary.any()
    )
)


print(
    "Robust candidate threshold at boundary:",
    bool(
        robust_candidate_boundary.any()
    )
)


print(
    "\nBoundary issue requiring wider scan:",
    boundary_issue
)


if boundary_issue:

    print(
        "\n⚠️ At least one important optimum "
        "is at the search boundary."
    )

    print(
        "Do NOT finalize the threshold yet."
    )

else:

    print(
        "\n✅ Important optima are inside "
        "the searched threshold range."
    )
```

# PHASE 11B — Extended Threshold Boundary Analysis

```python
# ============================================================
# PHASE 11B — EXTENDED THRESHOLD BOUNDARY ANALYSIS
# CELL 1: Extended Threshold Grid
# ============================================================

PHASE11B_DIR = os.path.join(
    PHASE11_DIR,
    "extended_threshold_analysis"
)

os.makedirs(
    PHASE11B_DIR,
    exist_ok=True
)


EXTENDED_THRESHOLDS = np.round(
    np.arange(
        0.01,
        1.00,
        0.01
    ),
    2
)


print("=" * 88)
print("PHASE 11B — EXTENDED THRESHOLD ANALYSIS")
print("=" * 88)

print(
    "Old threshold range : "
    f"{THRESHOLDS.min():.2f} → "
    f"{THRESHOLDS.max():.2f}"
)

print(
    "New threshold range : "
    f"{EXTENDED_THRESHOLDS.min():.2f} → "
    f"{EXTENDED_THRESHOLDS.max():.2f}"
)

print(
    "Threshold step      : 0.01"
)

print(
    "Threshold count     :",
    len(EXTENDED_THRESHOLDS)
)

print(
    "Candidates          :",
    len(tuned_candidate_specs)
)

print(
    "Final test locked   :",
    FINAL_TEST_LOCKED
)
```

```python
# ============================================================
# CELL 2: Extended Threshold Scan
# ============================================================

extended_threshold_rows = []


for spec in tuned_candidate_specs:

    candidate_name = spec["Candidate"]

    probabilities = (
        oof_probability_dictionary[
            candidate_name
        ]
    )


    for threshold in EXTENDED_THRESHOLDS:

        metrics = evaluate_threshold(
            y_true=y_dev,
            probabilities=probabilities,
            threshold=threshold
        )


        extended_threshold_rows.append({

            "Candidate":
                candidate_name,

            "Model":
                spec["Model"],

            "Strategy":
                spec["Strategy"],

            **metrics
        })


extended_threshold_results_df = pd.DataFrame(
    extended_threshold_rows
)


print("✅ Extended threshold scan completed.")

print(
    "Threshold evaluations:",
    len(
        extended_threshold_results_df
    )
)

print(
    "Expected:",
    7 * 99
)
```

```python
# ============================================================
# CELL 3: Extended Scenario-Based Decision Costs
# ============================================================

extended_cost_rows = []


for _, row in (
    extended_threshold_results_df
    .iterrows()
):

    for scenario_name, ratio in (
        COST_SCENARIOS.items()
    ):

        weighted_cost = (
            ratio * int(row["FN"])
            +
            int(row["FP"])
        )


        normalized_cost = (
            weighted_cost
            /
            N_DECISION
        )


        extended_cost_rows.append({

            "Candidate":
                row["Candidate"],

            "Model":
                row["Model"],

            "Strategy":
                row["Strategy"],

            "Threshold":
                float(
                    row["Threshold"]
                ),

            "Scenario":
                scenario_name,

            "FN_FP_Ratio":
                ratio,

            "Precision":
                float(
                    row["Precision"]
                ),

            "Recall":
                float(
                    row["Recall"]
                ),

            "F1":
                float(
                    row["F1"]
                ),

            "Specificity":
                float(
                    row["Specificity"]
                ),

            "FN":
                int(
                    row["FN"]
                ),

            "FP":
                int(
                    row["FP"]
                ),

            "TP":
                int(
                    row["TP"]
                ),

            "TN":
                int(
                    row["TN"]
                ),

            "Weighted_Decision_Cost":
                int(
                    weighted_cost
                ),

            "Normalized_Decision_Cost":
                float(
                    normalized_cost
                )
        })


extended_decision_cost_df = pd.DataFrame(
    extended_cost_rows
)


print(
    "✅ Extended decision-cost analysis completed."
)

print(
    "Total cost evaluations:",
    len(
        extended_decision_cost_df
    )
)

print(
    "Expected:",
    7 * 99 * 3
)
```

```python
# ============================================================
# CELL 4: Extended Best Thresholds
# ============================================================

extended_best_cost_thresholds = (
    extended_decision_cost_df
    .sort_values(
        [
            "Candidate",
            "Scenario",
            "Normalized_Decision_Cost",
            "F1",
            "Recall",
            "Precision"
        ],
        ascending=[
            True,
            True,
            True,
            False,
            False,
            False
        ]
    )
    .groupby(
        [
            "Candidate",
            "Scenario"
        ],
        as_index=False
    )
    .first()
)


print(
    "Candidate-scenario optima:",
    len(
        extended_best_cost_thresholds
    )
)


display(
    extended_best_cost_thresholds[
        [
            "Candidate",
            "Scenario",
            "FN_FP_Ratio",
            "Threshold",
            "Normalized_Decision_Cost",
            "Recall",
            "Precision",
            "F1",
            "FN",
            "FP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 5: Extended Scenario Rankings
# ============================================================

extended_scenario_parts = []


for scenario_name in COST_SCENARIOS.keys():

    scenario_table = (
        extended_best_cost_thresholds[
            extended_best_cost_thresholds[
                "Scenario"
            ] == scenario_name
        ]
        .copy()
    )


    scenario_table = scenario_table.merge(
        oof_summary[
            [
                "Candidate",
                "OOF_AP",
                "OOF_ROC_AUC"
            ]
        ],
        on="Candidate",
        how="left"
    )


    scenario_table = (
        scenario_table
        .sort_values(
            [
                "Normalized_Decision_Cost",
                "OOF_AP",
                "F1"
            ],
            ascending=[
                True,
                False,
                False
            ]
        )
        .reset_index(
            drop=True
        )
    )


    scenario_table[
        "Scenario_Rank"
    ] = np.arange(
        1,
        len(scenario_table) + 1
    )


    extended_scenario_parts.append(
        scenario_table
    )


extended_scenario_rankings = pd.concat(
    extended_scenario_parts,
    ignore_index=True
)


extended_scenario_winners = (
    extended_scenario_rankings[
        extended_scenario_rankings[
            "Scenario_Rank"
        ] == 1
    ]
    .copy()
)


display(
    extended_scenario_rankings[
        [
            "Scenario",
            "Scenario_Rank",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost",
            "Recall",
            "Precision",
            "FN",
            "FP",
            "OOF_AP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 6: Extended Robust Candidate Ranking
# ============================================================

extended_robust_rank_summary = (
    extended_scenario_rankings
    .groupby(
        [
            "Candidate",
            "Model",
            "Strategy"
        ]
    )
    .agg(

        Mean_Scenario_Rank=(
            "Scenario_Rank",
            "mean"
        ),

        Best_Scenario_Rank=(
            "Scenario_Rank",
            "min"
        ),

        Worst_Scenario_Rank=(
            "Scenario_Rank",
            "max"
        ),

        Scenario_Wins=(
            "Scenario_Rank",
            lambda x:
                int(
                    (x == 1).sum()
                )
        ),

        Mean_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "mean"
        ),

        Worst_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "max"
        ),

        Mean_Recall=(
            "Recall",
            "mean"
        ),

        Mean_Precision=(
            "Precision",
            "mean"
        ),

        Mean_F1=(
            "F1",
            "mean"
        ),

        OOF_AP=(
            "OOF_AP",
            "first"
        )
    )
    .reset_index()
)


extended_robust_rank_summary = (
    extended_robust_rank_summary
    .sort_values(
        [
            "Mean_Scenario_Rank",
            "Worst_Scenario_Rank",
            "Mean_Normalized_Cost",
            "OOF_AP"
        ],
        ascending=[
            True,
            True,
            True,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)


extended_robust_rank_summary.insert(
    0,
    "Robust_Rank",
    np.arange(
        1,
        len(
            extended_robust_rank_summary
        ) + 1
    )
)


display(
    extended_robust_rank_summary.round(6)
)


extended_robust_candidate = (
    extended_robust_rank_summary
    .iloc[0]
)
```

```python
# ============================================================
# CELL 7: Extended Boundary Audit
# ============================================================

EXTENDED_MIN = float(
    EXTENDED_THRESHOLDS.min()
)

EXTENDED_MAX = float(
    EXTENDED_THRESHOLDS.max()
)


extended_winner_boundary = (
    extended_scenario_winners[
        "Threshold"
    ]
    .apply(
        lambda x:
            np.isclose(
                x,
                EXTENDED_MIN
            )
            or
            np.isclose(
                x,
                EXTENDED_MAX
            )
    )
)


extended_robust_name = (
    extended_robust_candidate[
        "Candidate"
    ]
)


extended_robust_scenarios = (
    extended_scenario_rankings[
        extended_scenario_rankings[
            "Candidate"
        ]
        ==
        extended_robust_name
    ]
    .copy()
)


extended_robust_boundary = (
    extended_robust_scenarios[
        "Threshold"
    ]
    .apply(
        lambda x:
            np.isclose(
                x,
                EXTENDED_MIN
            )
            or
            np.isclose(
                x,
                EXTENDED_MAX
            )
    )
)


extended_boundary_issue = bool(
    extended_winner_boundary.any()
    or
    extended_robust_boundary.any()
)


print("=" * 86)
print("EXTENDED THRESHOLD BOUNDARY AUDIT")
print("=" * 86)

print(
    f"Search range : "
    f"{EXTENDED_MIN:.2f} "
    f"→ "
    f"{EXTENDED_MAX:.2f}"
)

print(
    "Scenario winner at boundary :",
    bool(
        extended_winner_boundary.any()
    )
)

print(
    "Robust candidate at boundary:",
    bool(
        extended_robust_boundary.any()
    )
)

print(
    "Boundary issue:",
    extended_boundary_issue
)


if extended_boundary_issue:

    print(
        "\n⚠️ Boundary issue still exists."
    )

else:

    print(
        "\n✅ All important optima are "
        "inside the extended grid."
    )
```

```python
# ============================================================
# CELL 8: Old vs Extended Scenario Winners
# ============================================================

old_winners_compare = (
    scenario_winners[
        [
            "Scenario",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Candidate":
                "Old_Candidate",

            "Threshold":
                "Old_Threshold",

            "Normalized_Decision_Cost":
                "Old_Cost"
        }
    )
)


new_winners_compare = (
    extended_scenario_winners[
        [
            "Scenario",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Candidate":
                "New_Candidate",

            "Threshold":
                "New_Threshold",

            "Normalized_Decision_Cost":
                "New_Cost"
        }
    )
)


old_vs_extended = (
    old_winners_compare
    .merge(
        new_winners_compare,
        on="Scenario",
        how="inner"
    )
)


old_vs_extended[
    "Cost_Change"
] = (
    old_vs_extended[
        "New_Cost"
    ]
    -
    old_vs_extended[
        "Old_Cost"
    ]
)


display(
    old_vs_extended.round(6)
)
```

```python
# ============================================================
# CELL 9: Save Phase 11B Outputs
# ============================================================

extended_threshold_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_extended_threshold_results.csv"
)


extended_cost_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_extended_decision_costs.csv"
)


extended_best_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_best_candidate_scenario_thresholds.csv"
)


extended_rankings_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_scenario_rankings.csv"
)


extended_winners_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_scenario_winners.csv"
)


extended_robust_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_robust_ranking.csv"
)


old_vs_extended_path = os.path.join(
    PHASE11B_DIR,
    "phase11b_old_vs_extended_winners.csv"
)


extended_threshold_results_df.to_csv(
    extended_threshold_path,
    index=False
)


extended_decision_cost_df.to_csv(
    extended_cost_path,
    index=False
)


extended_best_cost_thresholds.to_csv(
    extended_best_path,
    index=False
)


extended_scenario_rankings.to_csv(
    extended_rankings_path,
    index=False
)


extended_scenario_winners.to_csv(
    extended_winners_path,
    index=False
)


extended_robust_rank_summary.to_csv(
    extended_robust_path,
    index=False
)


old_vs_extended.to_csv(
    old_vs_extended_path,
    index=False
)


print(
    "✅ Phase 11B outputs saved."
)
```

```python
# ============================================================
# PHASE 11B — FINAL SUMMARY
# Send me the COMPLETE output
# ============================================================

print("=" * 100)
print("PHASE 11B — EXTENDED THRESHOLD ANALYSIS SUMMARY")
print("=" * 100)


print("\n[1] EXTENDED SEARCH DESIGN")

print(
    f"Old range  : "
    f"{THRESHOLDS.min():.2f}"
    f" → "
    f"{THRESHOLDS.max():.2f}"
)

print(
    f"New range  : "
    f"{EXTENDED_THRESHOLDS.min():.2f}"
    f" → "
    f"{EXTENDED_THRESHOLDS.max():.2f}"
)

print(
    "Step       : 0.01"
)

print(
    "Thresholds :",
    len(
        EXTENDED_THRESHOLDS
    )
)

print(
    "Threshold evaluations:",
    len(
        extended_threshold_results_df
    )
)

print(
    "Cost evaluations:",
    len(
        extended_decision_cost_df
    )
)


print("\n[2] EXTENDED SCENARIO WINNERS")


for _, row in (
    extended_scenario_winners
    .sort_values(
        "FN_FP_Ratio"
    )
    .iterrows()
):

    print(
        f"\nFN:FP = "
        f"{int(row['FN_FP_Ratio'])}:1"
    )

    print(
        f"  Candidate : "
        f"{row['Candidate']}"
    )

    print(
        f"  Threshold : "
        f"{row['Threshold']:.2f}"
    )

    print(
        f"  Cost      : "
        f"{row['Normalized_Decision_Cost']:.6f}"
    )

    print(
        f"  Recall    : "
        f"{row['Recall']:.4f}"
    )

    print(
        f"  Precision : "
        f"{row['Precision']:.4f}"
    )

    print(
        f"  FN        : "
        f"{int(row['FN'])}"
    )

    print(
        f"  FP        : "
        f"{int(row['FP'])}"
    )


print("\n[3] EXTENDED ROBUST RANKING")


for _, row in (
    extended_robust_rank_summary
    .iterrows()
):

    print(
        f"#{int(row['Robust_Rank'])} "
        f"{row['Candidate']:<34} "
        f"MeanRank="
        f"{row['Mean_Scenario_Rank']:.2f} | "
        f"WorstRank="
        f"{int(row['Worst_Scenario_Rank'])} | "
        f"Wins="
        f"{int(row['Scenario_Wins'])} | "
        f"MeanCost="
        f"{row['Mean_Normalized_Cost']:.6f} | "
        f"AP="
        f"{row['OOF_AP']:.6f}"
    )


print("\n[4] ROBUST CANDIDATE")

print(
    "Candidate :",
    extended_robust_candidate[
        "Candidate"
    ]
)

print(
    "Model     :",
    extended_robust_candidate[
        "Model"
    ]
)

print(
    "Strategy  :",
    extended_robust_candidate[
        "Strategy"
    ]
)

print(
    "Mean Rank :",
    round(
        float(
            extended_robust_candidate[
                "Mean_Scenario_Rank"
            ]
        ),
        4
    )
)


print("\n[5] ROBUST CANDIDATE SCENARIO THRESHOLDS")


for _, row in (
    extended_robust_scenarios
    .sort_values(
        "FN_FP_Ratio"
    )
    .iterrows()
):

    print(
        f"FN:FP "
        f"{int(row['FN_FP_Ratio'])}:1 | "
        f"Rank={int(row['Scenario_Rank'])} | "
        f"t={row['Threshold']:.2f} | "
        f"Cost={row['Normalized_Decision_Cost']:.6f} | "
        f"Recall={row['Recall']:.4f} | "
        f"Precision={row['Precision']:.4f} | "
        f"FN={int(row['FN'])} | "
        f"FP={int(row['FP'])}"
    )


print("\n[6] OLD → EXTENDED WINNER COMPARISON")


for _, row in (
    old_vs_extended
    .iterrows()
):

    print(
        f"\n{row['Scenario']}"
    )

    print(
        f"  Old : "
        f"{row['Old_Candidate']} "
        f"| t={row['Old_Threshold']:.2f} "
        f"| Cost={row['Old_Cost']:.6f}"
    )

    print(
        f"  New : "
        f"{row['New_Candidate']} "
        f"| t={row['New_Threshold']:.2f} "
        f"| Cost={row['New_Cost']:.6f}"
    )


print("\n[7] EXTENDED BOUNDARY AUDIT")

print(
    "Scenario winner boundary :",
    bool(
        extended_winner_boundary.any()
    )
)

print(
    "Robust candidate boundary:",
    bool(
        extended_robust_boundary.any()
    )
)

print(
    "Boundary issue           :",
    extended_boundary_issue
)


print("\n[8] TEST SAFETY")

print(
    "Final test used   : False"
)

print(
    "Final test locked :",
    FINAL_TEST_LOCKED
)


phase11b_files = [
    extended_threshold_path,
    extended_cost_path,
    extended_best_path,
    extended_rankings_path,
    extended_winners_path,
    extended_robust_path,
    old_vs_extended_path
]


phase11b_checks = {

    "Exactly 99 thresholds":
        len(
            EXTENDED_THRESHOLDS
        ) == 99,

    "Exactly 693 threshold evaluations":
        len(
            extended_threshold_results_df
        ) == 693,

    "Exactly 2079 cost evaluations":
        len(
            extended_decision_cost_df
        ) == 2079,

    "Exactly 21 candidate-scenario optima":
        len(
            extended_best_cost_thresholds
        ) == 21,

    "Exactly three scenario winners":
        len(
            extended_scenario_winners
        ) == 3,

    "Seven candidates robustly ranked":
        len(
            extended_robust_rank_summary
        ) == 7,

    "Final test remains locked":
        FINAL_TEST_LOCKED is True,

    "All Phase 11B files saved":
        all(
            os.path.isfile(path)
            for path in phase11b_files
        )
}


print("\n" + "-" * 100)
print("PHASE 11B INTEGRITY CHECKS")
print("-" * 100)


for name, status in (
    phase11b_checks.items()
):

    print(
        (
            "✅"
            if status
            else "❌"
        ),
        name
    )


print("\n" + "=" * 100)


if (
    all(
        phase11b_checks.values()
    )
    and
    not extended_boundary_issue
):

    print(
        "✅ PHASE 11B STATUS: PASSED — "
        "BOUNDARY ISSUE RESOLVED"
    )

elif all(
    phase11b_checks.values()
):

    print(
        "⚠️ PHASE 11B STATUS: PASSED, "
        "BUT BOUNDARY ISSUE STILL EXISTS"
    )

else:

    print(
        "❌ PHASE 11B STATUS: REVIEW REQUIRED"
    )


print("=" * 100)
```

# PHASE 12 — Final Model + Threshold Selection

```python
# ============================================================
# PHASE 12 — FINAL MODEL + THRESHOLD SELECTION
# CELL 1: Phase Setup
# ============================================================

PHASE12_DIR = OUTPUT_DIRS["09_final_model"]

os.makedirs(
    PHASE12_DIR,
    exist_ok=True
)


print("=" * 90)
print("PHASE 12 — ROBUST MODEL + THRESHOLD SELECTION")
print("=" * 90)

print(
    f"Development samples : {len(y_dev):,}"
)

print(
    f"Candidates          : "
    f"{extended_decision_cost_df['Candidate'].nunique()}"
)

print(
    f"Thresholds          : "
    f"{extended_decision_cost_df['Threshold'].nunique()}"
)

print(
    f"Cost scenarios      : "
    f"{extended_decision_cost_df['Scenario'].nunique()}"
)

print(
    "\nSelection principle:"
)

print(
    "Minimize worst-case relative decision-cost regret"
)

print(
    "\nFinal test used     : False"
)

print(
    "Final test locked   :",
    FINAL_TEST_LOCKED
)
```

```python
# ============================================================
# CELL 2: Scenario-Wise Global Best Cost
# ============================================================

scenario_best_cost_reference = (
    extended_decision_cost_df
    .groupby(
        [
            "Scenario",
            "FN_FP_Ratio"
        ],
        as_index=False
    )
    .agg(
        Scenario_Global_Best_Cost=(
            "Normalized_Decision_Cost",
            "min"
        )
    )
)


display(
    scenario_best_cost_reference
    .round(6)
)
```

```python
# ============================================================
# CELL 3: Verify Scenario Best-Cost References
# ============================================================

scenario_reference_check = (
    extended_scenario_winners[
        [
            "Scenario",
            "FN_FP_Ratio",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost"
        ]
    ]
    .merge(
        scenario_best_cost_reference,
        on=[
            "Scenario",
            "FN_FP_Ratio"
        ],
        how="left"
    )
)


scenario_reference_check[
    "Cost_Match"
] = np.isclose(
    scenario_reference_check[
        "Normalized_Decision_Cost"
    ],
    scenario_reference_check[
        "Scenario_Global_Best_Cost"
    ]
)


display(
    scenario_reference_check
    .round(6)
)


print(
    "\nAll scenario references verified:",
    bool(
        scenario_reference_check[
            "Cost_Match"
        ].all()
    )
)
```

```python
# ============================================================
# CELL 4: Calculate Scenario Regret
# ============================================================

robust_combination_costs = (
    extended_decision_cost_df
    .merge(
        scenario_best_cost_reference,
        on=[
            "Scenario",
            "FN_FP_Ratio"
        ],
        how="left"
    )
)


robust_combination_costs[
    "Absolute_Regret"
] = (
    robust_combination_costs[
        "Normalized_Decision_Cost"
    ]
    -
    robust_combination_costs[
        "Scenario_Global_Best_Cost"
    ]
)


robust_combination_costs[
    "Relative_Regret"
] = (
    robust_combination_costs[
        "Absolute_Regret"
    ]
    /
    robust_combination_costs[
        "Scenario_Global_Best_Cost"
    ]
)


print(
    "Regret-analysis rows:",
    len(
        robust_combination_costs
    )
)


print(
    "\nRelative regret range:"
)

print(
    "Minimum:",
    round(
        robust_combination_costs[
            "Relative_Regret"
        ].min(),
        6
    )
)

print(
    "Maximum:",
    round(
        robust_combination_costs[
            "Relative_Regret"
        ].max(),
        6
    )
)


display(
    robust_combination_costs.head()
    .round(6)
)
```

```python
# ============================================================
# CELL 5: Cross-Scenario Combination Summary
# ============================================================

robust_combination_summary = (
    robust_combination_costs
    .groupby(
        [
            "Candidate",
            "Model",
            "Strategy",
            "Threshold"
        ],
        as_index=False
    )
    .agg(

        Mean_Relative_Regret=(
            "Relative_Regret",
            "mean"
        ),

        Max_Relative_Regret=(
            "Relative_Regret",
            "max"
        ),

        Mean_Absolute_Regret=(
            "Absolute_Regret",
            "mean"
        ),

        Max_Absolute_Regret=(
            "Absolute_Regret",
            "max"
        ),

        Mean_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "mean"
        ),

        Worst_Normalized_Cost=(
            "Normalized_Decision_Cost",
            "max"
        ),

        Precision=(
            "Precision",
            "first"
        ),

        Recall=(
            "Recall",
            "first"
        ),

        F1=(
            "F1",
            "first"
        ),

        Specificity=(
            "Specificity",
            "first"
        ),

        FN=(
            "FN",
            "first"
        ),

        FP=(
            "FP",
            "first"
        )
    )
)


# ------------------------------------------------------------
# Add OOF ranking quality
# ------------------------------------------------------------

robust_combination_summary = (
    robust_combination_summary
    .merge(
        oof_summary[
            [
                "Candidate",
                "OOF_AP",
                "OOF_ROC_AUC"
            ]
        ],
        on="Candidate",
        how="left"
    )
)


print(
    "Unique model-threshold combinations:",
    len(
        robust_combination_summary
    )
)


display(
    robust_combination_summary
    .head()
    .round(6)
)
```

```python
# ============================================================
# CELL 6: Robust Model-Threshold Ranking
# ============================================================

robust_combination_ranking = (
    robust_combination_summary
    .sort_values(
        [
            "Max_Relative_Regret",
            "Mean_Relative_Regret",
            "Mean_Normalized_Cost",
            "OOF_AP",
            "F1"
        ],
        ascending=[
            True,
            True,
            True,
            False,
            False
        ]
    )
    .reset_index(
        drop=True
    )
)


robust_combination_ranking.insert(
    0,
    "Robust_Combination_Rank",
    np.arange(
        1,
        len(
            robust_combination_ranking
        ) + 1
    )
)


selected_combination = (
    robust_combination_ranking
    .iloc[0]
)


print("=" * 90)
print("SELECTED ROBUST MODEL + THRESHOLD")
print("=" * 90)


print(
    "Candidate :",
    selected_combination[
        "Candidate"
    ]
)

print(
    "Model     :",
    selected_combination[
        "Model"
    ]
)

print(
    "Strategy  :",
    selected_combination[
        "Strategy"
    ]
)

print(
    f"Threshold : "
    f"{selected_combination['Threshold']:.2f}"
)

print(
    f"Max Relative Regret  : "
    f"{selected_combination['Max_Relative_Regret']:.6f}"
)

print(
    f"Mean Relative Regret : "
    f"{selected_combination['Mean_Relative_Regret']:.6f}"
)

print(
    f"Mean Decision Cost   : "
    f"{selected_combination['Mean_Normalized_Cost']:.6f}"
)

print(
    f"OOF AP               : "
    f"{selected_combination['OOF_AP']:.6f}"
)

print(
    f"OOF ROC-AUC          : "
    f"{selected_combination['OOF_ROC_AUC']:.6f}"
)

print(
    f"Recall               : "
    f"{selected_combination['Recall']:.6f}"
)

print(
    f"Precision            : "
    f"{selected_combination['Precision']:.6f}"
)

print(
    f"F1                   : "
    f"{selected_combination['F1']:.6f}"
)

print(
    f"FN / FP              : "
    f"{int(selected_combination['FN'])} / "
    f"{int(selected_combination['FP'])}"
)
```

```python
# ============================================================
# CELL 7: Top Robust Combinations
# ============================================================

top_robust_combinations = (
    robust_combination_ranking
    .head(10)
    .copy()
)


display(
    top_robust_combinations[
        [
            "Robust_Combination_Rank",
            "Candidate",
            "Threshold",
            "Max_Relative_Regret",
            "Mean_Relative_Regret",
            "Mean_Normalized_Cost",
            "OOF_AP",
            "Recall",
            "Precision",
            "F1",
            "FN",
            "FP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 8: Selected Combination Scenario Breakdown
# ============================================================

SELECTED_CANDIDATE = str(
    selected_combination[
        "Candidate"
    ]
)


SELECTED_MODEL = str(
    selected_combination[
        "Model"
    ]
)


SELECTED_STRATEGY = str(
    selected_combination[
        "Strategy"
    ]
)


SELECTED_THRESHOLD = float(
    selected_combination[
        "Threshold"
    ]
)


selected_scenario_performance = (
    robust_combination_costs[
        (
            robust_combination_costs[
                "Candidate"
            ]
            ==
            SELECTED_CANDIDATE
        )
        &
        (
            np.isclose(
                robust_combination_costs[
                    "Threshold"
                ],
                SELECTED_THRESHOLD
            )
        )
    ]
    .sort_values(
        "FN_FP_Ratio"
    )
    .copy()
)


display(
    selected_scenario_performance[
        [
            "Scenario",
            "FN_FP_Ratio",
            "Threshold",
            "Normalized_Decision_Cost",
            "Scenario_Global_Best_Cost",
            "Absolute_Regret",
            "Relative_Regret",
            "Recall",
            "Precision",
            "F1",
            "FN",
            "FP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# CELL 9: Robust Selection vs Scenario-Specific Optimum
# ============================================================

selected_vs_scenario_winner = (
    selected_scenario_performance[
        [
            "Scenario",
            "FN_FP_Ratio",
            "Candidate",
            "Threshold",
            "Normalized_Decision_Cost",
            "Relative_Regret",
            "FN",
            "FP",
            "Recall",
            "Precision"
        ]
    ]
    .rename(
        columns={
            "Candidate":
                "Robust_Candidate",

            "Threshold":
                "Robust_Threshold",

            "Normalized_Decision_Cost":
                "Robust_Cost",

            "FN":
                "Robust_FN",

            "FP":
                "Robust_FP",

            "Recall":
                "Robust_Recall",

            "Precision":
                "Robust_Precision"
        }
    )
    .merge(

        extended_scenario_winners[
            [
                "Scenario",
                "Candidate",
                "Threshold",
                "Normalized_Decision_Cost",
                "FN",
                "FP"
            ]
        ]
        .rename(
            columns={
                "Candidate":
                    "Scenario_Winner",

                "Threshold":
                    "Scenario_Optimal_Threshold",

                "Normalized_Decision_Cost":
                    "Scenario_Best_Cost",

                "FN":
                    "Scenario_Best_FN",

                "FP":
                    "Scenario_Best_FP"
            }
        ),

        on="Scenario",

        how="left"
    )
)


display(
    selected_vs_scenario_winner
    .round(6)
)
```

```python
# ============================================================
# CELL 10: Selected Threshold vs Default 0.50
# ============================================================

selected_default_cost = (
    extended_decision_cost_df[
        (
            extended_decision_cost_df[
                "Candidate"
            ]
            ==
            SELECTED_CANDIDATE
        )
        &
        (
            np.isclose(
                extended_decision_cost_df[
                    "Threshold"
                ],
                0.50
            )
        )
    ]
    .sort_values(
        "FN_FP_Ratio"
    )
    .copy()
)


selected_threshold_cost = (
    selected_scenario_performance
    .sort_values(
        "FN_FP_Ratio"
    )
    .copy()
)


selected_default_vs_robust = (
    selected_default_cost[
        [
            "Scenario",
            "FN_FP_Ratio",
            "Normalized_Decision_Cost",
            "FN",
            "FP"
        ]
    ]
    .rename(
        columns={
            "Normalized_Decision_Cost":
                "Default_Cost",

            "FN":
                "Default_FN",

            "FP":
                "Default_FP"
        }
    )
    .merge(

        selected_threshold_cost[
            [
                "Scenario",
                "Normalized_Decision_Cost",
                "FN",
                "FP"
            ]
        ]
        .rename(
            columns={
                "Normalized_Decision_Cost":
                    "Robust_Cost",

                "FN":
                    "Robust_FN",

                "FP":
                    "Robust_FP"
            }
        ),

        on="Scenario",

        how="inner"
    )
)


selected_default_vs_robust[
    "Cost_Reduction"
] = (
    selected_default_vs_robust[
        "Default_Cost"
    ]
    -
    selected_default_vs_robust[
        "Robust_Cost"
    ]
)


selected_default_vs_robust[
    "Cost_Reduction_Percentage"
] = (

    selected_default_vs_robust[
        "Cost_Reduction"
    ]

    /

    selected_default_vs_robust[
        "Default_Cost"
    ]

    *

    100
)


display(
    selected_default_vs_robust
    .round(6)
)
```

```python
# ============================================================
# CELL 11: Retrieve Final Hyperparameters
# ============================================================

selected_parameter_row = (
    best_tuned_candidates[
        best_tuned_candidates[
            "Candidate"
        ]
        ==
        SELECTED_CANDIDATE
    ]
)


if len(
    selected_parameter_row
) != 1:

    raise ValueError(
        "Selected candidate hyperparameters "
        "could not be uniquely identified."
    )


SELECTED_PARAMETERS = json.loads(
    selected_parameter_row
    .iloc[0][
        "Parameters"
    ]
)


print("=" * 82)
print("SELECTED MODEL SPECIFICATION")
print("=" * 82)

print(
    "Candidate :",
    SELECTED_CANDIDATE
)

print(
    "Model     :",
    SELECTED_MODEL
)

print(
    "Strategy  :",
    SELECTED_STRATEGY
)

print(
    f"Threshold : "
    f"{SELECTED_THRESHOLD:.2f}"
)

print(
    "\nHyperparameters:"
)


for key, value in (
    SELECTED_PARAMETERS.items()
):

    print(
        f" - {key}: {value}"
    )
```

```python
# ============================================================
# CELL 12: Final Selection Safety Audit
# ============================================================

selected_at_boundary = bool(

    np.isclose(
        SELECTED_THRESHOLD,
        EXTENDED_THRESHOLDS.min()
    )

    or

    np.isclose(
        SELECTED_THRESHOLD,
        EXTENDED_THRESHOLDS.max()
    )
)


print("=" * 82)
print("FINAL SELECTION SAFETY AUDIT")
print("=" * 82)


print(
    "Selected threshold:",
    SELECTED_THRESHOLD
)


print(
    "Threshold at grid boundary:",
    selected_at_boundary
)


print(
    "Selection based only on OOF data:",
    True
)


print(
    "Final test used:",
    False
)


print(
    "Final test remains locked:",
    FINAL_TEST_LOCKED
)


if selected_at_boundary:

    print(
        "\n❌ Selected threshold lies "
        "at extended grid boundary."
    )

else:

    print(
        "\n✅ Selected threshold lies "
        "inside the validated grid."
    )
```

```python
# ============================================================
# CELL 3: Create Fresh Final Preprocessing Objects
# ============================================================

final_pre_sampler = (
    make_pre_sampler_transformer()
)


final_post_encoder = (
    make_post_sampler_encoder()
)


print(
    "✅ Fresh final preprocessing "
    "objects created."
)
```

```python
# ============================================================
# CELL 4: Fit Stage-1 on Full Development Data
# ============================================================

X_dev_stage1_final = (
    final_pre_sampler
    .fit_transform(
        X_dev,
        y_dev
    )
)


print("=" * 82)
print("FINAL DEVELOPMENT PREPROCESSING")
print("=" * 82)

print(
    "Input shape :",
    X_dev.shape
)

print(
    "Stage-1 shape:",
    X_dev_stage1_final.shape
)

print(
    "NaN after preprocessing:",
    int(
        np.isnan(
            X_dev_stage1_final
        ).sum()
    )
)
```

```python
# ============================================================
# CELL 5: Final TotalCharges Imputation Audit
# ============================================================

final_numeric_pipeline = (
    final_pre_sampler
    .named_transformers_[
        "continuous"
    ]
)


final_numeric_imputer = (
    final_numeric_pipeline
    .named_steps[
        "imputer"
    ]
)


final_imputer_statistics = (
    final_numeric_imputer
    .statistics_
)


final_imputation_table = pd.DataFrame({

    "Feature":
        CONTINUOUS_FEATURES,

    "Development_Imputation_Value":
        final_imputer_statistics
})


display(
    final_imputation_table
)


print(
    "\nFull development TotalCharges median:",
    X_dev[
        "TotalCharges"
    ].median()
)
```

```python
# ============================================================
# CELL 6: Final Development One-Hot Encoding
# ============================================================

X_dev_final_encoded = (
    final_post_encoder
    .fit_transform(
        X_dev_stage1_final,
        y_dev
    )
)


print("=" * 82)
print("FINAL DEVELOPMENT ENCODING")
print("=" * 82)

print(
    "Encoded shape:",
    X_dev_final_encoded.shape
)

print(
    "NaN count:",
    int(
        np.isnan(
            X_dev_final_encoded
        ).sum()
    )
)


print(
    "\nResampling performed:",
    False
)
```

```python
# ============================================================
# CELL 7: Create Locked Final Model
# ============================================================

final_model = make_tunable_model(

    model_name=
        SELECTED_MODEL,

    strategy=
        SELECTED_STRATEGY,

    y_training=
        y_dev,

    params=
        SELECTED_PARAMETERS
)


print("=" * 82)
print("FINAL MODEL")
print("=" * 82)

print(final_model)

print(
    "\n✅ Locked XGBoost model created."
)
```

```python
# ============================================================
# CELL 8: Fit Final Model
# ============================================================

final_fit_start = time.time()


final_model.fit(
    X_dev_final_encoded,
    y_dev
)


final_fit_seconds = (
    time.time()
    -
    final_fit_start
)


print(
    "✅ Final model fitted on full "
    "development set."
)

print(
    f"Training samples : "
    f"{len(y_dev):,}"
)

print(
    f"Fit time         : "
    f"{final_fit_seconds:.2f} sec"
)
```

```python
# ============================================================
# CELL 9: OPEN LOCKED FINAL TEST
# ============================================================

print("=" * 90)
print("🔓 OPENING LOCKED FINAL TEST FOR FINAL EVALUATION")
print("=" * 90)


# ------------------------------------------------------------
# Transform test using DEVELOPMENT-fitted preprocessor
# ------------------------------------------------------------

X_test_stage1_final = (
    final_pre_sampler
    .transform(
        X_test
    )
)


X_test_final_encoded = (
    final_post_encoder
    .transform(
        X_test_stage1_final
    )
)


# ------------------------------------------------------------
# Final unseen probabilities
# ------------------------------------------------------------

final_test_probabilities = (
    final_model
    .predict_proba(
        X_test_final_encoded
    )[:, 1]
)


FINAL_TEST_OPENED = True


print(
    "Final test transformed shape:",
    X_test_final_encoded.shape
)


print(
    "Missing probabilities:",
    int(
        np.isnan(
            final_test_probabilities
        ).sum()
    )
)


print(
    "Probability minimum:",
    round(
        float(
            final_test_probabilities.min()
        ),
        6
    )
)


print(
    "Probability maximum:",
    round(
        float(
            final_test_probabilities.max()
        ),
        6
    )
)


print(
    "\n🔓 FINAL_TEST_OPENED:",
    FINAL_TEST_OPENED
)


print(
    "\n⚠️ Configuration is now frozen permanently."
)
```

```python
# ============================================================
# CELL 10: Final Test Ranking Metrics
# ============================================================

FINAL_TEST_AP = (
    average_precision_score(
        y_test,
        final_test_probabilities
    )
)


FINAL_TEST_ROC_AUC = (
    roc_auc_score(
        y_test,
        final_test_probabilities
    )
)


print("=" * 82)
print("FINAL TEST — THRESHOLD-INDEPENDENT METRICS")
print("=" * 82)


print(
    f"Average Precision : "
    f"{FINAL_TEST_AP:.6f}"
)

print(
    f"ROC-AUC           : "
    f"{FINAL_TEST_ROC_AUC:.6f}"
)


print(
    "\nOOF AP reference  : "
    f"{selected_combination['OOF_AP']:.6f}"
)


print(
    "OOF ROC reference : "
    f"{selected_combination['OOF_ROC_AUC']:.6f}"
)
```

```python
# ============================================================
# CELL 11: Final Threshold Evaluation Function
# ============================================================

def final_threshold_metrics(
    threshold
):

    predictions = (
        final_test_probabilities
        >=
        threshold
    ).astype(int)


    tn, fp, fn, tp = (
        confusion_matrix(
            y_test,
            predictions,
            labels=[0, 1]
        )
        .ravel()
    )


    return {

        "Threshold":
            float(threshold),

        "Accuracy":
            accuracy_score(
                y_test,
                predictions
            ),

        "Precision":
            precision_score(
                y_test,
                predictions,
                zero_division=0
            ),

        "Recall":
            recall_score(
                y_test,
                predictions,
                zero_division=0
            ),

        "F1":
            f1_score(
                y_test,
                predictions,
                zero_division=0
            ),

        "TN":
            int(tn),

        "FP":
            int(fp),

        "FN":
            int(fn),

        "TP":
            int(tp),

        "Predicted_Churn":
            int(
                predictions.sum()
            )
    }


print(
    "✅ Final threshold evaluation "
    "function created."
)
```

```python
# ============================================================
# CELL 12: Default vs Selected Threshold
# ============================================================

final_default_metrics = (
    final_threshold_metrics(
        0.50
    )
)


final_selected_metrics = (
    final_threshold_metrics(
        SELECTED_THRESHOLD
    )
)


final_threshold_comparison = pd.DataFrame(
    [
        {
            "Threshold_Type":
                "Default",

            **final_default_metrics
        },

        {
            "Threshold_Type":
                "Selected_Robust",

            **final_selected_metrics
        }
    ]
)


display(
    final_threshold_comparison
    .round(6)
)
```

```python
# ============================================================
# CELL 13: Final Selected-Threshold Results
# ============================================================

print("=" * 88)
print("FINAL TEST RESULT — LOCKED THRESHOLD")
print("=" * 88)


print(
    "Model     :",
    SELECTED_MODEL
)

print(
    "Strategy  :",
    SELECTED_STRATEGY
)

print(
    f"Threshold : "
    f"{SELECTED_THRESHOLD:.2f}"
)


print(
    f"\nAccuracy  : "
    f"{final_selected_metrics['Accuracy']:.6f}"
)

print(
    f"Precision : "
    f"{final_selected_metrics['Precision']:.6f}"
)

print(
    f"Recall    : "
    f"{final_selected_metrics['Recall']:.6f}"
)

print(
    f"F1        : "
    f"{final_selected_metrics['F1']:.6f}"
)

print(
    f"ROC-AUC   : "
    f"{FINAL_TEST_ROC_AUC:.6f}"
)

print(
    f"AP        : "
    f"{FINAL_TEST_AP:.6f}"
)


print(
    "\nConfusion Matrix:"
)

print(
    f"TN = {final_selected_metrics['TN']}"
)

print(
    f"FP = {final_selected_metrics['FP']}"
)

print(
    f"FN = {final_selected_metrics['FN']}"
)

print(
    f"TP = {final_selected_metrics['TP']}"
)
```

```python
# ============================================================
# CELL 14: Final Test Scenario Decision Costs
# ============================================================

final_cost_rows = []


for threshold_name, metrics in [

    (
        "Default_0.50",
        final_default_metrics
    ),

    (
        "Selected_0.18",
        final_selected_metrics
    )
]:

    for scenario_name, ratio in (
        COST_SCENARIOS.items()
    ):

        weighted_cost = (
            ratio
            *
            metrics["FN"]
            +
            metrics["FP"]
        )


        normalized_cost = (
            weighted_cost
            /
            len(y_test)
        )


        final_cost_rows.append({

            "Threshold_Type":
                threshold_name,

            "Scenario":
                scenario_name,

            "FN_FP_Ratio":
                ratio,

            "Threshold":
                metrics["Threshold"],

            "FN":
                metrics["FN"],

            "FP":
                metrics["FP"],

            "Weighted_Decision_Cost":
                weighted_cost,

            "Normalized_Decision_Cost":
                normalized_cost
        })


final_test_costs = pd.DataFrame(
    final_cost_rows
)


display(
    final_test_costs
    .round(6)
)
```

```python
# ============================================================
# CELL 15: Default vs Selected Decision Cost
# ============================================================

default_test_cost = (
    final_test_costs[
        final_test_costs[
            "Threshold_Type"
        ]
        ==
        "Default_0.50"
    ][
        [
            "Scenario",
            "FN_FP_Ratio",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Normalized_Decision_Cost":
                "Default_Cost"
        }
    )
)


selected_test_cost = (
    final_test_costs[
        final_test_costs[
            "Threshold_Type"
        ]
        ==
        "Selected_0.18"
    ][
        [
            "Scenario",
            "Normalized_Decision_Cost"
        ]
    ]
    .rename(
        columns={
            "Normalized_Decision_Cost":
                "Selected_Cost"
        }
    )
)


final_cost_comparison = (
    default_test_cost
    .merge(
        selected_test_cost,
        on="Scenario",
        how="inner"
    )
)


final_cost_comparison[
    "Cost_Reduction"
] = (
    final_cost_comparison[
        "Default_Cost"
    ]
    -
    final_cost_comparison[
        "Selected_Cost"
    ]
)


final_cost_comparison[
    "Cost_Reduction_Percentage"
] = (

    final_cost_comparison[
        "Cost_Reduction"
    ]

    /

    final_cost_comparison[
        "Default_Cost"
    ]

    *

    100
)


display(
    final_cost_comparison
    .round(6)
)
```

```python
# ============================================================
# CELL 16: Final Confusion Matrix Figure
# ============================================================

final_cm = np.array([
    [
        final_selected_metrics["TN"],
        final_selected_metrics["FP"]
    ],
    [
        final_selected_metrics["FN"],
        final_selected_metrics["TP"]
    ]
])


fig, ax = plt.subplots(
    figsize=(6, 5)
)


image = ax.imshow(
    final_cm
)


ax.set_xticks(
    [0, 1]
)

ax.set_yticks(
    [0, 1]
)


ax.set_xticklabels(
    [
        "Predicted Non-Churn",
        "Predicted Churn"
    ]
)


ax.set_yticklabels(
    [
        "Actual Non-Churn",
        "Actual Churn"
    ]
)


for i in range(2):

    for j in range(2):

        ax.text(
            j,
            i,
            str(
                final_cm[i, j]
            ),
            ha="center",
            va="center"
        )


ax.set_title(
    f"Final Test Confusion Matrix "
    f"(Threshold = {SELECTED_THRESHOLD:.2f})"
)


plt.tight_layout()


final_cm_path = os.path.join(
    PHASE13_DIR,
    "final_test_confusion_matrix.png"
)


plt.savefig(
    final_cm_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()
plt.close()


print(
    "✅ Final confusion matrix saved."
)
```

```python
# ============================================================
# CELL 20: Save Final Evaluation Metadata
# ============================================================

phase13_metadata = {

    "candidate":
        SELECTED_CANDIDATE,

    "model":
        SELECTED_MODEL,

    "strategy":
        SELECTED_STRATEGY,

    "selected_threshold":
        SELECTED_THRESHOLD,

    "selected_hyperparameters":
        SELECTED_PARAMETERS,

    "development_samples":
        int(
            len(y_dev)
        ),

    "final_test_samples":
        int(
            len(y_test)
        ),

    "final_test_ap":
        float(
            FINAL_TEST_AP
        ),

    "final_test_roc_auc":
        float(
            FINAL_TEST_ROC_AUC
        ),

    "selected_threshold_accuracy":
        float(
            final_selected_metrics[
                "Accuracy"
            ]
        ),

    "selected_threshold_precision":
        float(
            final_selected_metrics[
                "Precision"
            ]
        ),

    "selected_threshold_recall":
        float(
            final_selected_metrics[
                "Recall"
            ]
        ),

    "selected_threshold_f1":
        float(
            final_selected_metrics[
                "F1"
            ]
        ),

    "selected_threshold_tn":
        int(
            final_selected_metrics[
                "TN"
            ]
        ),

    "selected_threshold_fp":
        int(
            final_selected_metrics[
                "FP"
            ]
        ),

    "selected_threshold_fn":
        int(
            final_selected_metrics[
                "FN"
            ]
        ),

    "selected_threshold_tp":
        int(
            final_selected_metrics[
                "TP"
            ]
        ),

    "test_used_for_tuning":
        False,

    "test_used_for_threshold_selection":
        False,

    "configuration_changed_after_test":
        False,

    "actual_financial_cost_claimed":
        False
}


phase13_metadata_path = os.path.join(
    PHASE13_DIR,
    "phase13_final_evaluation.json"
)


with open(
    phase13_metadata_path,
    "w"
) as f:

    json.dump(
        phase13_metadata,
        f,
        indent=4
    )


print(
    "✅ Final evaluation metadata saved."
)
```

```python
# ============================================================
# PHASE 13 — FINAL SUMMARY
# Run LAST and send me the COMPLETE output
# ============================================================

print("=" * 106)
print("PHASE 13 — FINAL LOCKED-TEST EVALUATION SUMMARY")
print("=" * 106)


# ------------------------------------------------------------
# 1. Locked Configuration
# ------------------------------------------------------------
print("\n[1] LOCKED MODEL CONFIGURATION")

print(
    "Candidate :",
    SELECTED_CANDIDATE
)

print(
    "Model     :",
    SELECTED_MODEL
)

print(
    "Strategy  :",
    SELECTED_STRATEGY
)

print(
    f"Threshold : "
    f"{SELECTED_THRESHOLD:.2f}"
)


# ------------------------------------------------------------
# 2. Dataset
# ------------------------------------------------------------
print("\n[2] FINAL EVALUATION DATA")

print(
    f"Development samples : "
    f"{len(y_dev):,}"
)

print(
    f"Final test samples  : "
    f"{len(y_test):,}"
)

print(
    f"Test Non-Churn      : "
    f"{int((y_test == 0).sum()):,}"
)

print(
    f"Test Churn          : "
    f"{int((y_test == 1).sum()):,}"
)


# ------------------------------------------------------------
# 3. Ranking Metrics
# ------------------------------------------------------------
print("\n[3] FINAL TEST RANKING PERFORMANCE")

print(
    f"Average Precision : "
    f"{FINAL_TEST_AP:.6f}"
)

print(
    f"ROC-AUC           : "
    f"{FINAL_TEST_ROC_AUC:.6f}"
)

print(
    f"\nDevelopment OOF AP      : "
    f"{selected_combination['OOF_AP']:.6f}"
)

print(
    f"Development OOF ROC-AUC : "
    f"{selected_combination['OOF_ROC_AUC']:.6f}"
)


# ------------------------------------------------------------
# 4. Default Threshold
# ------------------------------------------------------------
print("\n[4] DEFAULT THRESHOLD = 0.50")

print(
    f"Accuracy  : "
    f"{final_default_metrics['Accuracy']:.6f}"
)

print(
    f"Precision : "
    f"{final_default_metrics['Precision']:.6f}"
)

print(
    f"Recall    : "
    f"{final_default_metrics['Recall']:.6f}"
)

print(
    f"F1        : "
    f"{final_default_metrics['F1']:.6f}"
)

print(
    f"TN / FP   : "
    f"{final_default_metrics['TN']} / "
    f"{final_default_metrics['FP']}"
)

print(
    f"FN / TP   : "
    f"{final_default_metrics['FN']} / "
    f"{final_default_metrics['TP']}"
)


# ------------------------------------------------------------
# 5. Selected Threshold
# ------------------------------------------------------------
print(
    f"\n[5] LOCKED SELECTED THRESHOLD = "
    f"{SELECTED_THRESHOLD:.2f}"
)

print(
    f"Accuracy  : "
    f"{final_selected_metrics['Accuracy']:.6f}"
)

print(
    f"Precision : "
    f"{final_selected_metrics['Precision']:.6f}"
)

print(
    f"Recall    : "
    f"{final_selected_metrics['Recall']:.6f}"
)

print(
    f"F1        : "
    f"{final_selected_metrics['F1']:.6f}"
)

print(
    f"TN / FP   : "
    f"{final_selected_metrics['TN']} / "
    f"{final_selected_metrics['FP']}"
)

print(
    f"FN / TP   : "
    f"{final_selected_metrics['FN']} / "
    f"{final_selected_metrics['TP']}"
)


# ------------------------------------------------------------
# 6. Threshold Effect
# ------------------------------------------------------------
print("\n[6] THRESHOLD EFFECT")

print(
    f"Recall : "
    f"{final_default_metrics['Recall']:.4f}"
    f" → "
    f"{final_selected_metrics['Recall']:.4f}"
)

print(
    f"F1     : "
    f"{final_default_metrics['F1']:.4f}"
    f" → "
    f"{final_selected_metrics['F1']:.4f}"
)

print(
    f"FN     : "
    f"{final_default_metrics['FN']}"
    f" → "
    f"{final_selected_metrics['FN']}"
)

print(
    f"FP     : "
    f"{final_default_metrics['FP']}"
    f" → "
    f"{final_selected_metrics['FP']}"
)


# ------------------------------------------------------------
# 7. Scenario Decision Costs
# ------------------------------------------------------------
print("\n[7] FINAL TEST SCENARIO DECISION COSTS")


for _, row in (
    final_cost_comparison
    .sort_values(
        "FN_FP_Ratio"
    )
    .iterrows()
):

    print(
        f"FN:FP "
        f"{int(row['FN_FP_Ratio'])}:1 "
        f"| Default="
        f"{row['Default_Cost']:.6f} "
        f"| Selected="
        f"{row['Selected_Cost']:.6f} "
        f"| Reduction="
        f"{row['Cost_Reduction_Percentage']:+.2f}%"
    )


# ------------------------------------------------------------
# 8. Generalization Difference
# ------------------------------------------------------------
print("\n[8] OOF → FINAL TEST GENERALIZATION")

print(
    f"AP difference      : "
    f"{FINAL_TEST_AP - selected_combination['OOF_AP']:+.6f}"
)

print(
    f"ROC-AUC difference : "
    f"{FINAL_TEST_ROC_AUC - selected_combination['OOF_ROC_AUC']:+.6f}"
)


# ------------------------------------------------------------
# 9. Safety
# ------------------------------------------------------------
print("\n[9] FINAL EVALUATION SAFETY")

print(
    "Final test opened                  :",
    FINAL_TEST_OPENED
)

print(
    "Test used during model selection   : False"
)

print(
    "Test used during tuning            : False"
)

print(
    "Test used during threshold search  : False"
)

print(
    "Configuration changed after test   : False"
)

print(
    "Actual monetary cost claimed       : False"
)


# ------------------------------------------------------------
# 10. Saved Outputs
# ------------------------------------------------------------
print("\n[10] SAVED OUTPUTS")


phase13_files = [

    final_predictions_path,
    final_threshold_comparison_path,
    final_costs_path,
    final_cost_comparison_path,
    final_cm_path,
    final_pre_sampler_path,
    final_post_encoder_path,
    final_model_path,
    phase13_metadata_path
]


for path in phase13_files:

    print(
        f"{os.path.basename(path):<46}"
        f": {os.path.isfile(path)}"
    )


# ------------------------------------------------------------
# Integrity Checks
# ------------------------------------------------------------

phase13_checks = {

    "Final test opened":
        FINAL_TEST_OPENED is True,

    "Exactly 1409 test samples evaluated":
        len(
            final_test_probabilities
        ) == 1409,

    "No missing test probabilities":
        not np.isnan(
            final_test_probabilities
        ).any(),

    "All probabilities within [0,1]":
        (
            (
                final_test_probabilities >= 0
            )
            &
            (
                final_test_probabilities <= 1
            )
        ).all(),

    "Selected threshold unchanged":
        np.isclose(
            SELECTED_THRESHOLD,
            0.18
        ),

    "Selected candidate unchanged":
        SELECTED_CANDIDATE
        ==
        "XGBoost_Original",

    "Confusion matrix totals test size":
        (
            final_selected_metrics["TN"]
            +
            final_selected_metrics["FP"]
            +
            final_selected_metrics["FN"]
            +
            final_selected_metrics["TP"]
        )
        ==
        len(y_test),

    "AP valid":
        0
        <=
        FINAL_TEST_AP
        <=
        1,

    "ROC-AUC valid":
        0
        <=
        FINAL_TEST_ROC_AUC
        <=
        1,

    "All Phase 13 outputs saved":
        all(
            os.path.isfile(
                path
            )
            for path in phase13_files
        )
}


print("\n" + "-" * 106)
print("PHASE 13 INTEGRITY CHECKS")
print("-" * 106)


for name, status in (
    phase13_checks.items()
):

    print(
        (
            "✅"
            if status
            else "❌"
        ),
        name
    )


print("\n" + "=" * 106)


if all(
    phase13_checks.values()
):

    print(
        "✅ PHASE 13 STATUS: PASSED — "
        "FINAL TEST EVALUATION COMPLETE"
    )

else:

    print(
        "❌ PHASE 13 STATUS: REVIEW REQUIRED"
    )


print("=" * 106)
```

# PHASE 14 — RQ3: SHAP Explainability

```python
# ============================================================
# PHASE 14 — SHAP EXPLAINABILITY
# CELL 1: Setup
# ============================================================

import shap

PHASE14_DIR = OUTPUT_DIRS["10_shap"]

os.makedirs(
    PHASE14_DIR,
    exist_ok=True
)


print("=" * 88)
print("PHASE 14 — SHAP EXPLAINABILITY")
print("=" * 88)

print("Model      :", SELECTED_MODEL)
print("Strategy   :", SELECTED_STRATEGY)
print("Threshold  :", SELECTED_THRESHOLD)

print(
    f"Final test samples : {len(X_test):,}"
)

print(
    "Encoded features   :",
    X_test_final_encoded.shape[1]
)

print(
    "\nSHAP is used only for post-hoc explanation."
)

print(
    "No tuning/model/threshold changes will be made."
)
```

```python
# ============================================================
# CELL 2: Reconstruct Human-Readable Encoded Feature Names
# ============================================================

# ------------------------------------------------------------
# Find OrdinalEncoder from pre-sampler categorical pipeline
# ------------------------------------------------------------

final_categorical_pipeline = (
    final_pre_sampler
    .named_transformers_[
        "categorical"
    ]
)


ordinal_encoder_final = None


for step_name, step_object in (
    final_categorical_pipeline.steps
):

    if isinstance(
        step_object,
        OrdinalEncoder
    ):

        ordinal_encoder_final = (
            step_object
        )


if ordinal_encoder_final is None:

    raise ValueError(
        "OrdinalEncoder could not be found."
    )


# ------------------------------------------------------------
# Find OneHotEncoder from final post-encoder
# ------------------------------------------------------------

post_categorical_transformer = (
    final_post_encoder
    .named_transformers_[
        "categorical"
    ]
)


if isinstance(
    post_categorical_transformer,
    OneHotEncoder
):

    onehot_encoder_final = (
        post_categorical_transformer
    )

else:

    onehot_encoder_final = None

    if hasattr(
        post_categorical_transformer,
        "steps"
    ):

        for _, step_object in (
            post_categorical_transformer.steps
        ):

            if isinstance(
                step_object,
                OneHotEncoder
            ):

                onehot_encoder_final = (
                    step_object
                )


if onehot_encoder_final is None:

    raise ValueError(
        "OneHotEncoder could not be found."
    )


# ------------------------------------------------------------
# Build readable names
# ------------------------------------------------------------

readable_feature_names = (
    list(
        CONTINUOUS_FEATURES
    )
)


raw_feature_mapping = (
    list(
        CONTINUOUS_FEATURES
    )
)


for feature_position, raw_feature in enumerate(
    CATEGORICAL_FEATURES
):

    original_categories = (
        ordinal_encoder_final
        .categories_[
            feature_position
        ]
    )


    encoded_codes = (
        onehot_encoder_final
        .categories_[
            feature_position
        ]
    )


    for code in encoded_codes:

        try:

            category_position = int(
                round(
                    float(code)
                )
            )


            if (
                0
                <=
                category_position
                <
                len(original_categories)
            ):

                category_label = (
                    original_categories[
                        category_position
                    ]
                )

            else:

                category_label = (
                    f"Code_{code}"
                )

        except Exception:

            category_label = (
                str(code)
            )


        readable_feature_names.append(
            f"{raw_feature}={category_label}"
        )


        raw_feature_mapping.append(
            raw_feature
        )


print("=" * 82)
print("READABLE FEATURE NAME AUDIT")
print("=" * 82)

print(
    "Expected encoded features:",
    X_test_final_encoded.shape[1]
)

print(
    "Generated names          :",
    len(
        readable_feature_names
    )
)


if (
    len(readable_feature_names)
    !=
    X_test_final_encoded.shape[1]
):

    raise ValueError(
        "Readable feature count does not "
        "match encoded feature count."
    )


print(
    "\n✅ Feature-name mapping verified."
)
```

```python
# ============================================================
# CELL 3: Human-Readable Encoded Test Matrix
# ============================================================

X_test_shap_df = pd.DataFrame(
    X_test_final_encoded,
    columns=
        readable_feature_names,
    index=
        X_test.index
)


encoded_feature_mapping_df = pd.DataFrame({

    "Encoded_Index":
        np.arange(
            len(
                readable_feature_names
            )
        ),

    "Encoded_Feature":
        readable_feature_names,

    "Raw_Feature":
        raw_feature_mapping
})


print(
    "SHAP matrix shape:",
    X_test_shap_df.shape
)


display(
    encoded_feature_mapping_df
    .head(20)
)
```

```python
# ============================================================
# CELL 4: Compute SHAP Values
# ============================================================

shap_start_time = (
    time.time()
)


shap_explainer = (
    shap.TreeExplainer(
        final_model
    )
)


raw_shap_values = (
    shap_explainer
    .shap_values(
        X_test_final_encoded
    )
)


# ------------------------------------------------------------
# Handle SHAP version differences
# ------------------------------------------------------------

if isinstance(
    raw_shap_values,
    list
):

    shap_values_array = (
        np.asarray(
            raw_shap_values[-1]
        )
    )

else:

    shap_values_array = (
        np.asarray(
            raw_shap_values
        )
    )


# Some versions may return an extra output dimension
if shap_values_array.ndim == 3:

    shap_values_array = (
        shap_values_array[
            :,
            :,
            -1
        ]
    )


expected_value = (
    shap_explainer
    .expected_value
)


if isinstance(
    expected_value,
    (
        list,
        tuple,
        np.ndarray
    )
):

    expected_value = float(
        np.asarray(
            expected_value
        )
        .reshape(-1)[-1]
    )

else:

    expected_value = float(
        expected_value
    )


shap_runtime = (
    time.time()
    -
    shap_start_time
)


print("=" * 82)
print("SHAP COMPUTATION")
print("=" * 82)

print(
    "SHAP shape:",
    shap_values_array.shape
)

print(
    "Expected:",
    X_test_final_encoded.shape
)

print(
    "Expected/base value:",
    round(
        expected_value,
        6
    )
)

print(
    f"Runtime: "
    f"{shap_runtime:.2f} sec"
)


if (
    shap_values_array.shape
    !=
    X_test_final_encoded.shape
):

    raise ValueError(
        "SHAP value dimensions do not match "
        "the encoded test matrix."
    )


print(
    "\n✅ SHAP values successfully generated."
)
```

```python
# ============================================================
# CELL 5: Build SHAP Explanation Object
# ============================================================

shap_explanation = shap.Explanation(

    values=
        shap_values_array,

    base_values=
        np.repeat(
            expected_value,
            len(
                X_test_shap_df
            )
        ),

    data=
        X_test_shap_df
        .to_numpy(),

    feature_names=
        readable_feature_names
)


print(
    "✅ SHAP Explanation object created."
)
```

```python
# ============================================================
# CELL 6: Encoded-Level Global SHAP Importance
# ============================================================

encoded_mean_abs_shap = (
    np.abs(
        shap_values_array
    )
    .mean(
        axis=0
    )
)


encoded_shap_importance = pd.DataFrame({

    "Encoded_Feature":
        readable_feature_names,

    "Raw_Feature":
        raw_feature_mapping,

    "Mean_Absolute_SHAP":
        encoded_mean_abs_shap
})


encoded_shap_importance = (
    encoded_shap_importance
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


encoded_shap_importance.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(
            encoded_shap_importance
        ) + 1
    )
)


display(
    encoded_shap_importance
    .head(20)
    .round(6)
)
```

```python
# ============================================================
# PHASE 14
# FINAL CELL 7: Numerically Stable Raw-Feature Grouped SHAP
# ============================================================

# ------------------------------------------------------------
# Original 19-feature order
# ------------------------------------------------------------

raw_feature_order = (
    CONTINUOUS_FEATURES
    +
    CATEGORICAL_FEATURES
)


# ------------------------------------------------------------
# Convert SHAP values to float64
# for numerically stable aggregation
# ------------------------------------------------------------

shap_values_float64 = np.asarray(
    shap_values_array,
    dtype=np.float64
)


grouped_shap_columns = []
grouped_shap_rows = []


# ------------------------------------------------------------
# Aggregate encoded SHAP components
# back to original raw features
# ------------------------------------------------------------

for raw_feature in raw_feature_order:

    feature_indices = [
        i
        for i, feature_name
        in enumerate(raw_feature_mapping)
        if feature_name == raw_feature
    ]


    if len(feature_indices) == 0:

        raise ValueError(
            f"No encoded components found "
            f"for raw feature: {raw_feature}"
        )


    # --------------------------------------------------------
    # IMPORTANT:
    # Sum component SHAP values for EACH customer first.
    #
    # Correct grouping:
    #
    # mean(
    #     abs(
    #         sum(component SHAP values)
    #     )
    # )
    # --------------------------------------------------------

    per_customer_grouped_shap = (
        shap_values_float64[
            :,
            feature_indices
        ]
        .sum(
            axis=1,
            dtype=np.float64
        )
    )


    grouped_shap_columns.append(
        per_customer_grouped_shap
    )


    grouped_shap_rows.append({

        "Raw_Feature":
            raw_feature,

        "Mean_Absolute_SHAP":
            float(
                np.mean(
                    np.abs(
                        per_customer_grouped_shap
                    )
                )
            ),

        "Encoded_Components":
            int(
                len(
                    feature_indices
                )
            )
    })


# ============================================================
# Create grouped SHAP matrix
# Shape should be:
# 1409 customers × 19 raw features
# ============================================================

grouped_shap_values_array = (
    np.column_stack(
        grouped_shap_columns
    )
    .astype(
        np.float64
    )
)


# ============================================================
# Create global raw-feature importance table
# ============================================================

grouped_shap_importance = (
    pd.DataFrame(
        grouped_shap_rows
    )
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=False
    )
    .reset_index(
        drop=True
    )
)


grouped_shap_importance.insert(
    0,
    "Rank",
    np.arange(
        1,
        len(
            grouped_shap_importance
        ) + 1
    )
)


# ============================================================
# Mapping Coverage Audit
# ============================================================

all_group_indices = []


for raw_feature in raw_feature_order:

    indices = [
        i
        for i, feature_name
        in enumerate(raw_feature_mapping)
        if feature_name == raw_feature
    ]

    all_group_indices.extend(
        indices
    )


mapping_complete = (
    sorted(
        all_group_indices
    )
    ==
    list(
        range(
            shap_values_float64.shape[1]
        )
    )
)


mapping_unique = (
    len(
        all_group_indices
    )
    ==
    len(
        set(
            all_group_indices
        )
    )
)


# ============================================================
# SHAP Grouping Additivity Audit
# ============================================================

encoded_total = (
    shap_values_float64
    .sum(
        axis=1,
        dtype=np.float64
    )
)


grouped_total = (
    grouped_shap_values_array
    .sum(
        axis=1,
        dtype=np.float64
    )
)


grouping_difference = (
    grouped_total
    -
    encoded_total
)


grouping_max_abs_difference = float(
    np.max(
        np.abs(
            grouping_difference
        )
    )
)


grouping_mean_abs_difference = float(
    np.mean(
        np.abs(
            grouping_difference
        )
    )
)


# ------------------------------------------------------------
# 1e-6 tolerance accounts only for
# harmless floating-point summation differences.
# ------------------------------------------------------------

grouping_additivity_verified = bool(
    np.allclose(
        grouped_total,
        encoded_total,
        atol=1e-6,
        rtol=1e-6
    )
)


# ============================================================
# Display Audit
# ============================================================

print("=" * 90)
print("FINAL RAW-FEATURE SHAP AGGREGATION")
print("=" * 90)

print(
    "Encoded SHAP shape          :",
    shap_values_float64.shape
)

print(
    "Grouped SHAP shape          :",
    grouped_shap_values_array.shape
)

print(
    "Raw features                :",
    len(
        grouped_shap_importance
    )
)

print(
    "All encoded indices covered :",
    mapping_complete
)

print(
    "No duplicated indices       :",
    mapping_unique
)

print(
    f"Max additivity difference   : "
    f"{grouping_max_abs_difference:.12e}"
)

print(
    f"Mean additivity difference  : "
    f"{grouping_mean_abs_difference:.12e}"
)

print(
    "Grouped SHAP additivity verified:",
    grouping_additivity_verified
)


print(
    "\nTOP RAW FEATURES:"
)


display(
    grouped_shap_importance
    .round(6)
)


# ============================================================
# Safety Checks
# ============================================================

if not mapping_complete:

    raise ValueError(
        "Not all encoded SHAP features "
        "were assigned to a raw feature."
    )


if not mapping_unique:

    raise ValueError(
        "Some encoded SHAP features were "
        "assigned more than once."
    )


if (
    grouped_shap_values_array.shape
    !=
    (
        len(X_test),
        19
    )
):

    raise ValueError(
        "Grouped SHAP matrix must have "
        "shape (1409, 19)."
    )


if not grouping_additivity_verified:

    raise ValueError(
        "Grouped SHAP additivity exceeds "
        "the accepted numerical tolerance."
    )


print(
    "\n✅ Raw-feature SHAP grouping "
    "successfully verified."
)
```

```python
# ============================================================
# PHASE 14
# CELL 8: Corrected Grouped Raw-Feature SHAP Plot
# ============================================================

top_grouped = (
    grouped_shap_importance
    .head(15)
    .sort_values(
        "Mean_Absolute_SHAP",
        ascending=True
    )
)


fig, ax = plt.subplots(
    figsize=(9, 7)
)


ax.barh(
    top_grouped[
        "Raw_Feature"
    ],

    top_grouped[
        "Mean_Absolute_SHAP"
    ]
)


ax.set_title(
    "Global SHAP Importance — Original Customer Features"
)


ax.set_xlabel(
    "Mean Absolute Grouped SHAP Value"
)


ax.set_ylabel(
    "Original Feature"
)


ax.grid(
    axis="x",
    alpha=0.20
)


plt.tight_layout()


grouped_shap_plot_path = os.path.join(
    PHASE14_DIR,
    "phase14_grouped_global_shap.png"
)


plt.savefig(
    grouped_shap_plot_path,
    dpi=300,
    bbox_inches="tight"
)


plt.show()

plt.close()


print(
    "✅ Corrected grouped global SHAP plot saved."
)

print(
    grouped_shap_plot_path
)
```

```python
# ============================================================
# CELL 11: Select Objective Local Explanation Cases
# ============================================================

y_test_array = (
    y_test.to_numpy()
)


selected_predictions_array = (
    final_test_probabilities
    >=
    SELECTED_THRESHOLD
).astype(int)


# ------------------------------------------------------------
# True Positive
# ------------------------------------------------------------

tp_positions = np.where(
    (
        y_test_array == 1
    )
    &
    (
        selected_predictions_array == 1
    )
)[0]


tp_case_position = (
    tp_positions[
        np.argmax(
            final_test_probabilities[
                tp_positions
            ]
        )
    ]
)


# ------------------------------------------------------------
# Borderline case
# ------------------------------------------------------------

borderline_case_position = int(
    np.argmin(
        np.abs(
            final_test_probabilities
            -
            SELECTED_THRESHOLD
        )
    )
)


# ------------------------------------------------------------
# False Negative
# ------------------------------------------------------------

fn_positions = np.where(
    (
        y_test_array == 1
    )
    &
    (
        selected_predictions_array == 0
    )
)[0]


if len(fn_positions) > 0:

    fn_case_position = (
        fn_positions[
            np.argmax(
                final_test_probabilities[
                    fn_positions
                ]
            )
        ]
    )

else:

    fn_case_position = None


local_case_rows = [

    {
        "Case_Type":
            "High_Confidence_TP",

        "Position":
            int(
                tp_case_position
            )
    },

    {
        "Case_Type":
            "Borderline",

        "Position":
            int(
                borderline_case_position
            )
    }
]


if fn_case_position is not None:

    local_case_rows.append({

        "Case_Type":
            "False_Negative",

        "Position":
            int(
                fn_case_position
            )
    })


local_cases = pd.DataFrame(
    local_case_rows
)


local_cases[
    "Source_Index"
] = local_cases[
    "Position"
].apply(
    lambda pos:
        X_test.index[
            int(pos)
        ]
)


local_cases[
    "True_Churn"
] = local_cases[
    "Position"
].apply(
    lambda pos:
        int(
            y_test_array[
                int(pos)
            ]
        )
)


local_cases[
    "Predicted_Probability"
] = local_cases[
    "Position"
].apply(
    lambda pos:
        float(
            final_test_probabilities[
                int(pos)
            ]
        )
)


local_cases[
    "Prediction"
] = (
    local_cases[
        "Predicted_Probability"
    ]
    >=
    SELECTED_THRESHOLD
).astype(int)


display(
    local_cases
    .round(6)
)
```

```python
# ============================================================
# CELL 12: SHAP Local Waterfall Explanations
# ============================================================

local_waterfall_paths = []


for _, case in (
    local_cases.iterrows()
):

    case_type = (
        case[
            "Case_Type"
        ]
    )


    position = int(
        case[
            "Position"
        ]
    )


    case_explanation = shap.Explanation(

        values=
            shap_values_array[
                position
            ],

        base_values=
            expected_value,

        data=
            X_test_shap_df
            .iloc[
                position
            ]
            .to_numpy(),

        feature_names=
            readable_feature_names
    )


    shap.plots.waterfall(
        case_explanation,
        max_display=12,
        show=False
    )


    plt.title(
        f"{case_type} — "
        f"p(churn)="
        f"{final_test_probabilities[position]:.3f}"
    )


    plt.tight_layout()


    waterfall_path = os.path.join(
        PHASE14_DIR,
        f"phase14_local_{case_type.lower()}.png"
    )


    plt.savefig(
        waterfall_path,
        dpi=300,
        bbox_inches="tight"
    )


    plt.show()
    plt.close()


    local_waterfall_paths.append(
        waterfall_path
    )


print(
    "✅ Local SHAP waterfall plots saved:",
    len(
        local_waterfall_paths
    )
)
```

```python
# ============================================================
# CELL 13: Local Top SHAP Contributions
# ============================================================

local_contribution_rows = []


for _, case in (
    local_cases.iterrows()
):

    case_type = (
        case[
            "Case_Type"
        ]
    )


    position = int(
        case[
            "Position"
        ]
    )


    case_values = (
        shap_values_array[
            position
        ]
    )


    case_table = pd.DataFrame({

        "Encoded_Feature":
            readable_feature_names,

        "Raw_Feature":
            raw_feature_mapping,

        "Feature_Value":
            X_test_shap_df
            .iloc[
                position
            ]
            .to_numpy(),

        "SHAP_Value":
            case_values
    })


    case_table[
        "Absolute_SHAP"
    ] = (
        case_table[
            "SHAP_Value"
        ].abs()
    )


    case_table = (
        case_table
        .sort_values(
            "Absolute_SHAP",
            ascending=False
        )
        .head(12)
    )


    case_table[
        "Case_Type"
    ] = case_type


    case_table[
        "Source_Index"
    ] = (
        case[
            "Source_Index"
        ]
    )


    case_table[
        "Predicted_Probability"
    ] = (
        case[
            "Predicted_Probability"
        ]
    )


    local_contribution_rows.append(
        case_table
    )


local_contributions = pd.concat(
    local_contribution_rows,
    ignore_index=True
)


display(
    local_contributions[
        [
            "Case_Type",
            "Encoded_Feature",
            "SHAP_Value",
            "Absolute_SHAP"
        ]
    ].round(6)
)
```

```python
# ============================================================
# ZIP THE ENTIRE telco_churn_research FOLDER
# ============================================================

import os
import shutil

SOURCE_DIR = "/kaggle/working/telco_churn_research"
ZIP_BASE = "/kaggle/working/telco_churn_research_full"

# Safety check
if not os.path.isdir(SOURCE_DIR):
    raise FileNotFoundError(
        f"Folder not found: {SOURCE_DIR}"
    )

# Remove old zip if it exists
old_zip = ZIP_BASE + ".zip"

if os.path.isfile(old_zip):
    os.remove(old_zip)

# Create zip of entire folder
ZIP_PATH = shutil.make_archive(
    base_name=ZIP_BASE,
    format="zip",
    root_dir=os.path.dirname(SOURCE_DIR),
    base_dir=os.path.basename(SOURCE_DIR)
)

# Count files
file_count = 0
total_size = 0

for root, dirs, files in os.walk(SOURCE_DIR):
    for file in files:
        file_path = os.path.join(root, file)
        file_count += 1
        total_size += os.path.getsize(file_path)

print("=" * 90)
print("FULL PROJECT ZIP EXPORT")
print("=" * 90)

print("Source folder :", SOURCE_DIR)
print("Total files   :", file_count)

print(
    "Folder size   :",
    f"{total_size / (1024**2):.2f} MB"
)

print(
    "ZIP created   :",
    os.path.isfile(ZIP_PATH)
)

print(
    "ZIP size      :",
    f"{os.path.getsize(ZIP_PATH) / (1024**2):.2f} MB"
)

print("ZIP path      :", ZIP_PATH)

print("=" * 90)
print("✅ FULL telco_churn_research FOLDER PACKED SUCCESSFULLY")
```
